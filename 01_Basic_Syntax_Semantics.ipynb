### What is Python

Python is a programming language. It is a way to give instructions to a computer using text, instead of clicking buttons.

> Real-life analogy: Think of Python like English instructions for a robot chef. You write a recipe (code), and the robot (computer) follows it exactly, step by step, in order.

#### Why Python specifically

- It reads almost like plain English, so it is easy to learn
- One short Python instruction often does the work of many lines in other languages
- Used everywhere: websites, AI, data analysis, automation, games

> Interview Tip: If asked "Why Python?", mention three points: readability, large standard library, and a huge community and ecosystem (used in web development, data science, automation, and AI/ML).

### Variables — Storing Data with a Name

A variable is a name you give to a piece of data, so you can use that data again later without retyping it every time.

> Real-life analogy: A variable is like a labeled jar in your kitchen. You put sugar in a jar and label it "sugar". Whenever a recipe needs sugar, you just say "get the sugar jar" instead of describing the sugar every time.

### How Variables Work

When you write:

```python
age = 31
```

Python does three things:

1. Creates a value, `31`, in memory
2. Creates a name, `age`
3. Points the name `age` to that value, so you can use `age` anywhere instead of typing `31`

> Important Point: In Python, a variable is just a label pointing to a value. The value lives in memory, and the variable name is how you refer to it. This is different from languages like C, where a variable is treated more like a fixed memory box.

### Assigning Multiple Variables

```python
name = "Sandipan"
city = "Pune"
age = 31
```

You can also assign multiple variables in a single line:

```python
name, city, age = "Sandipan", "Pune", 31
```

### Reassigning Variables

A variable's value can change at any time. The jar label stays the same, but you can empty it and put something new inside.

```python
score = 10
print(score)   # 10

score = 25
print(score)   # 25
```

### Rules for Naming Variables

- Can contain letters, numbers, and underscores, example `user_name`, `age2`
- Cannot start with a number, example `2age` is invalid
- Cannot contain spaces, use `first_name` instead of `first name`
- Case-sensitive, so `Name` and `name` are treated as different variables
- Cannot use Python's reserved keywords as variable names, example `class`, `for`, `if`, `def`

### Good Naming Practice

- Use meaningful names, example `total_price` instead of `tp`
- Use lowercase with underscores for multi-word names, example `first_name` instead of `firstName` (this style is called snake_case, and it's the standard convention in Python)

> Interview Tip: A common interview question is "Is Python statically typed or dynamically typed?" Answer: Python is dynamically typed, meaning you don't declare a variable's type in advance, and the same variable name can even be reassigned to a different type later, example `x = 5` then `x = "hello"` is valid in Python.

> Important Point: Because variables can be reassigned to a different type, always be careful in long programs not to accidentally reuse a variable name for something unrelated. This can cause confusing bugs.

````python
# Variables in Python

# Basic assignment
age = 31
name = "Sandipan"
city = "Pune"

print(name, "is", age, "years old and lives in", city)
````

````python
# Multiple assignment in one line
name, city, age = "Sandipan", "Pune", 31
print(name, city, age)
````

### Data Types in Python

### 1. Introduction to Data Types

A data type tells Python what kind of value is stored in a variable, and what you are allowed to do with it.

> Real-life analogy: Think of data types like different containers in a kitchen. A jar holds dry items like sugar, a bottle holds liquids, a tray holds vegetables. You would not pour milk into a jar made for sugar. Similarly, Python expects you to use the right operations for the right type of data.

Every value in Python belongs to a specific data type, and Python checks this automatically whenever you run your code.

### 2. Importance of Data Types in Programming

Data types matter because they decide:

- How much memory a value takes
- What operations are valid on that value, for example you can add two numbers but not two lists with a plus sign in the same way
- How the value behaves when compared, stored, or converted
- How errors are caught early, before they become bigger bugs

> Important Point: Choosing the right data type makes your program faster, uses less memory, and avoids unexpected bugs. For example, using a set instead of a list to check membership repeatedly is much faster.

> Interview Tip: A common question is "Why does Python need data types if it is dynamically typed?" Answer: Even though you do not declare types explicitly, Python still enforces type rules at runtime. Dynamically typed means the type is decided automatically and can change, not that type rules disappear.

### 3. Basic Data Types

#### Integers

Whole numbers, with no decimal point, positive or negative.

> Real-life analogy: Counting apples in a basket. You cannot have half an apple while counting whole apples, only complete units.

- Example: `25`, `-3`, `0`
- Used for counting, indexing, and whole-number math

#### Floating-point Numbers

Numbers with a decimal point.

> Real-life analogy: Measuring weight on a kitchen scale, where the reading can be something like 2.75 kilograms, not just a whole number.

- Example: `3.14`, `-0.5`, `99.99`
- Used for measurements, prices, scientific calculations

> Important Point: Floating-point numbers can have small precision errors due to how computers store decimals in binary, for example `0.1 + 0.2` may show `0.30000000000000004` instead of exactly `0.3`. This is a classic interview question.

#### Strings

Text data, written inside single or double quotes.

> Real-life analogy: A name tag. It can hold any sequence of characters, like your name, an address, or a sentence.

- Example: `"Hello"`, `'Pune'`, `"123"` (this is text, not a number, because it is in quotes)
- Strings are sequences of characters, so you can access individual letters using indexing

#### Booleans

Only two possible values, `True` or `False`.

> Real-life analogy: A light switch. It is either ON or OFF, with nothing in between.

- Example: `True`, `False`
- Used heavily in conditions, comparisons, and decision-making, example `age > 18` evaluates to a boolean

> Interview Tip: In Python, `True` is internally treated as `1` and `False` as `0`. So `True + True` gives `2`. This is a common trick question.

---

### 4. Advanced Data Types

These are called collection types because each one can hold multiple values together inside a single variable.

#### Lists

An ordered, changeable collection of items.

> Real-life analogy: A shopping list written on paper. You can add items, remove items, or change an item, and the order in which you wrote them is preserved.

- Example: `fruits = ["apple", "banana", "mango"]`
- Allows duplicate values
- Items can be changed after creation, this is called mutable

#### Tuples

An ordered, unchangeable collection of items.

> Real-life analogy: A printed receipt. Once it is printed, you cannot edit the items on it, you can only read it.

- Example: `coordinates = (10, 20)`
- Allows duplicate values
- Cannot be changed after creation, this is called immutable
- Slightly faster than lists because Python knows they will not change

#### Sets

An unordered collection of unique items, with no duplicates allowed.

> Real-life analogy: A bag of unique stickers. If you try to add a sticker you already have, the bag simply ignores it, since duplicates are not allowed.

- Example: `unique_numbers = {1, 2, 3, 3, 2}` automatically becomes `{1, 2, 3}`
- No guaranteed order
- Very fast for checking whether an item exists

#### Dictionaries

A collection of key-value pairs, like a real dictionary where a word maps to its meaning.

> Real-life analogy: A phone contact list. The contact name is the key, and the phone number is the value. You look up a person by name to get their number, you do not search through every number one by one.

- Example: `student = {"name": "Sandipan", "age": 31, "city": "Pune"}`
- Keys must be unique, values can repeat
- Extremely fast lookups by key

> Important Point: Lists and tuples are accessed by position, called index, for example `fruits[0]`. Dictionaries are accessed by key, for example `student["name"]`. Sets are not accessed by position or key at all, since they are unordered.

> Interview Tip: A very common interview question is "What is the difference between a list and a tuple?" Answer: Lists are mutable and slightly slower, tuples are immutable and slightly faster, and tuples are often used for fixed data like coordinates or database records.

We will cover lists, tuples, sets, and dictionaries in much more depth as separate full lessons later in this course. This section is just an introduction so you can see the full picture of Python data types together.

### 5. Type Conversion

Type conversion means changing a value from one data type to another.

There are two kinds:

- Implicit conversion, done automatically by Python, example `5 + 2.0` automatically becomes `7.0`, an int combined with a float
- Explicit conversion, done manually by you, using functions like `int()`, `float()`, `str()`, `list()`, `tuple()`, `set()`

> Real-life analogy: Implicit conversion is like a chef automatically adjusting a recipe when you give them a slightly different ingredient. Explicit conversion is like you yourself deciding to swap one ingredient for another on purpose.

> Important Point: Converting a string with a decimal point directly to int will cause an error, example `int("3.14")` fails. You must convert to float first, then to int, example `int(float("3.14"))`.

> Interview Tip: Be ready to explain what happens when converting a list to a set, example `set([1, 2, 2, 3])` becomes `{1, 2, 3}`, removing duplicates and losing order. This is a common practical question.

### Operators in Python

### 1. Introduction to Operators

An operator is a special symbol that performs an operation on one or more values, called operands.

> Real-life analogy: Think of operators like tools in a kitchen. A knife (operator) cuts a vegetable (operand) into pieces (result). Different tools do different jobs, just like different operators do different jobs on your data.

Python operators are grouped into categories based on what they do:

> Important Point: This lesson covers Arithmetic, Comparison, and Logical operators in depth, since these are the most commonly used and most asked about in interviews. Assignment, identity, membership, and bitwise operators will be covered later as we progress.

---

### 2. Arithmetic Operators

Arithmetic operators perform mathematical calculations, just like a calculator.

```python
a = 10
b = 3
```

#### Addition `+`

Adds two values together.

> Real-life analogy: Combining two piles of coins into one bigger pile.

- `a + b` gives `13`

#### Subtraction `-`

Finds the difference between two values.

> Real-life analogy: Spending money from your wallet, the remaining amount is the result of subtraction.

- `a - b` gives `7`

#### Multiplication `*`

Repeats addition a certain number of times.

> Real-life analogy: Buying 3 packs of 10 pencils each. Instead of adding 10 three times, you multiply, `10 * 3`.

- `a * b` gives `30`

#### Division `/`

Splits a value into equal parts, and always gives a decimal result, called a float, even if the numbers divide evenly.

> Real-life analogy: Sharing a pizza of 10 slices among 3 friends, each person's exact share, including leftovers as fractions, is what division gives you.

- `a / b` gives `3.3333333333333335`

#### Floor Division `//`

Divides and removes anything after the decimal point, keeping only the whole number part.

> Real-life analogy: Distributing 10 candies equally among 3 friends, where candies cannot be broken into pieces. Each friend gets `10 // 3` which is `3` whole candies.

- `a // b` gives `3`

#### Modulus `%`

Gives the remainder left over after division.

> Real-life analogy: Continuing the candy example, after giving 3 candies to each of the 3 friends, `9` candies are distributed, and `1` candy remains. That leftover `1` is exactly what `%` gives you.

- `a % b` gives `1`

> Important Point: Modulus is extremely useful for checking even or odd numbers, example `number % 2 == 0` means the number is even, and for cyclic patterns, like checking every 5th item in a loop.

#### Exponentiation `**`

Raises a number to the power of another number.

> Real-life analogy: Doubling bacteria in a petri dish repeatedly. If bacteria double 3 times starting from 2, that is `2 ** 3`, which means 2 multiplied by itself 3 times.

- `a ** b` gives `1000`, meaning 10 multiplied by itself 3 times

> Interview Tip: A common question is "What is the difference between `/` and `//` in Python?" Answer: `/` always returns a float, even for exact division, while `//` returns the floor value, the largest whole number less than or equal to the result, and the return type depends on the operand types.

---

### 3. Comparison Operators

Comparison operators compare two values and always return a boolean result, either `True` or `False`.

> Real-life analogy: A weighing scale comparing two bags. It does not tell you the exact weight difference, it just tells you which side is heavier, lighter, or if they are equal.

```python
x = 10
y = 20
```

#### Equal to `==`

Checks if two values are the same.

- `x == y` gives `False`

> Important Point: A single `=` is used for assignment, like `x = 10`. A double `==` is used for comparison, like `x == 10`. Mixing these up is one of the most common beginner mistakes.

#### Not equal to `!=`

Checks if two values are different.

- `x != y` gives `True`

#### Greater than `>`

Checks if the left value is bigger than the right value.

- `x > y` gives `False`

#### Less than `<`

Checks if the left value is smaller than the right value.

- `x < y` gives `True`

#### Greater than or equal to `>=`

Checks if the left value is bigger than or the same as the right value.

- `x >= 10` gives `True`

#### Less than or equal to `<=`

Checks if the left value is smaller than or the same as the right value.

- `x <= 9` gives `False`

> Interview Tip: Comparison operators are the foundation of all decision-making in programming, since `if` statements and loops rely entirely on these boolean results. Make sure you are fully comfortable with them before moving to conditionals.

---

### 4. Logical Operators

Logical operators combine multiple conditions, called boolean expressions, into one final `True` or `False` result.

> Real-life analogy: Think of a bouncer at a club checking two conditions, age and dress code. AND means both conditions must be true for entry. OR means at least one is enough. NOT simply flips a decision, true becomes false and false becomes true.

```python
age = 25
has_id = True
```

#### AND

Returns `True` only if both conditions are `True`. If even one condition is `False`, the whole result is `False`.

- `age > 18 and has_id == True` gives `True`
- `age > 18 and has_id == False` gives `False`

#### OR

Returns `True` if at least one condition is `True`. Only returns `False` if both conditions are `False`.

- `age > 18 or has_id == False` gives `True`

#### NOT

Reverses the boolean value. `True` becomes `False`, and `False` becomes `True`.

- `not has_id` gives `False`, since `has_id` is `True`

> Important Point: Python uses the words `and`, `or`, `not` directly, unlike some languages that use symbols like `&&`, `||`, `!`. Using `&&` or `||` in Python will cause a syntax error.

> Interview Tip: A common interview question involves something called short-circuit evaluation. With `and`, if the first condition is `False`, Python does not even check the second condition, since the result is already decided. With `or`, if the first condition is `True`, Python skips checking the second condition. This matters for performance and for avoiding errors, for example checking `x != 0 and 10 / x > 1` safely avoids

````python
# 2. Arithmetic Operators

a = 10
b = 3

print(a + b)    # 13  Addition
print(a - b)    # 7   Subtraction
print(a * b)    # 30  Multiplication
print(a / b)    # 3.3333333333333335  Division, always returns float
print(a // b)   # 3   Floor Division
print(a % b)    # 1   Modulus, remainder
print(a ** b)   # 1000  Exponentiation
````

````python
# 3. Comparison Operators

x = 10
y = 20

print(x == y)   # False
print(x != y)   # True
print(x > y)    # False
print(x < y)    # True
print(x >= 10)  # True
print(x <= 9)   # False
````

````python
# 4. Logical Operators

age = 25
has_id = True

print(age > 18 and has_id == True)    # True, both conditions true
print(age > 18 and has_id == False)   # False, one condition false
print(age > 18 or has_id == False)    # True, at least one condition true
print(not has_id)                     # False, reverses True to False
````
