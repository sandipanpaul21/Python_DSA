```python
team_kolkata = {"sukanya","sukanta","jharna"}
team_pune    = {"sukanya", "gour", "susmita"}

# Adding / removing
team_pune.add("sandipan")
team_pune.discard("Ghost")          # no error even though absent
print("Pune team:", team_pune)

# Set math - who knows what?
print("Either:  ", team_pune | team_kolkata)     # union
print("Both:   ", team_pune & team_kolkata)     # intersection
print("Only Pune:   ", team_pune - team_kolkata)     # difference
print("Exactly one:   ", team_pune ^ team_kolkata)     # symmetric difference

# Relationship checks
juniors = {"sandipan"}
print("Juniors all in Pune team?", juniors.issubset(team_pune))
print("Teams share nobody?", team_pune.isdisjoint(team_kolkata))
```
