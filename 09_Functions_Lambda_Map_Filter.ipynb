# Python Functions (Basic to Advanced: including Lambda, Map, Filter)

## 1. Introduction

- **What is it?**
  - A **function** is a named, reusable block of code that performs a specific task and can be run whenever needed.
- **Definition**
  - A function is a **block of organized, reusable code defined using the `def` keyword** (or `lambda` for anonymous ones) that takes optional inputs (parameters), performs some logic, and optionally returns a result.
- **Why do we need it?**
  - Without functions, you'd repeat the same code everywhere you need that logic. Functions let you write logic **once** and reuse it, making programs shorter, easier to test, and easier to fix (one place to change, not many).
- **Where is it used?**
  - Literally everywhere: validating input, calculations, processing data, handling events in games, transforming data with `map`/`filter`, building APIs, structuring large programs into manageable pieces.

> A function is a **mini-program inside your program** — give it inputs, it does a job, it (optionally) hands back a result.

## 2. Real-Life Analogy

- A function is like a **coffee vending machine**.
  - You give it **input** (money, choice of coffee type) — these are the **arguments**.
  - It performs a fixed **process** internally (grinding, brewing) — this is the **function body**.
  - It gives you **output** (a cup of coffee) — this is the **return value**.
  - You can use the same machine (function) as many times as you want, with different inputs, without knowing exactly how it works inside.
- **Lambda functions** are like a **quick hand-written sticky note recipe** — a tiny, one-line, throwaway instruction you use once and discard, instead of writing a full recipe card (a full `def` function).

## 3. Explanation

- A function is defined once using `def function_name(parameters):` and a code block, then **called** (used) by writing `function_name(arguments)`.
- **Parameters** are the placeholder names listed in the function definition; **arguments** are the actual values you pass in when calling the function.
- Functions can accept **positional arguments**, **keyword arguments**, **default values**, and an **arbitrary number of arguments** (`*args`, `**kwargs`).
- A function can **return** a value using the `return` statement; if there's no `return`, the function returns `None` by default.
- A **lambda function** is a small, unnamed (anonymous) function defined in a single line using the `lambda` keyword — used for short, throwaway logic, often passed directly into other functions.
- `map()` and `filter()` are built-in functions that **apply a function to every item of an iterable** (`map`) or **keep only items that satisfy a condition** (`filter`) — both are commonly used together with `lambda`.

## 4. Syntax

### Defining and calling a basic function
```python
def function_name(parameter1, parameter2):
    """Docstring: explains what the function does."""
    # function body
    return result
```
- `def` -- keyword that starts a function definition.
- `function_name` -- should be lowercase_with_underscores (PEP 8).
- `(parameter1, parameter2)` -- placeholders for the data the function needs.
- Triple-quoted string right after the def line -- a **docstring**, recommended for documenting what the function does.
- `return` -- sends a value back to the caller; ends the function immediately when reached.

### Default parameter values
```python
def describe_pet(pet_name, animal_type='dog'):
    ...
```
- If the caller doesn't provide `animal_type`, Python uses `'dog'`.
- Parameters **with** default values must come **after** parameters without defaults.

### Arbitrary number of arguments
```python
def make_pizza(*toppings):        # collects extra positional args into a tuple
    ...

def build_profile(first, last, **user_info):   # collects extra keyword args into a dict
    ...
```
- `*args` — packs any number of positional arguments into a **tuple**.
- `**kwargs` — packs any number of keyword arguments into a **dictionary**.

### Lambda functions
```python
square = lambda x: x ** 2
add = lambda x, y: x + y
```
- `lambda` — keyword to define an anonymous function.
- `x` (before the colon) — the parameter(s), just like a normal function.
- `x ** 2` (after the colon) — a **single expression** that is automatically returned; no `return` keyword, no multi-line logic allowed.

### map() and filter()
```python
map(function, iterable)      # applies function to every item; returns a map object (lazy)
filter(function, iterable)    # keeps items where function(item) is True; returns a filter object (lazy)
```
- Both return **iterator objects**, not lists — wrap with `list(...)` to see the results immediately.

### 5. Examples

**Basic Example** — a simple function with a parameter and return value

````python
def get_formatted_name(first_name, last_name):
    """Return a full name, neatly formatted."""
    full_name = first_name + ' ' + last_name
    return full_name.title()

musician = get_formatted_name('jimi', 'hendrix')
print(musician)
````

**Intermediate Example** — default values, `*args`, and `**kwargs`

````python
def describe_pet(pet_name, animal_type='dog'):
    """Display information about a pet."""
    print(f"I have a {animal_type}. My {animal_type}'s name is {pet_name.title()}.")

describe_pet('willie')                       # uses default 'dog'
describe_pet('harry', animal_type='hamster') # overrides default


def make_pizza(size, *toppings):
    """Summarize the pizza being made."""
    print(f"\nMaking a {size}-inch pizza with:")
    for topping in toppings:
        print(f"- {topping}")

make_pizza(12, 'pepperoni')
make_pizza(16, 'mushrooms', 'green peppers', 'extra cheese')


def build_profile(first, last, **user_info):
    """Build a dictionary containing everything we know about a user."""
    user_info['first_name'] = first
    user_info['last_name'] = last
    return user_info

user_profile = build_profile('albert', 'einstein', location='princeton', field='physics')
print(user_profile)
````

**Real-World Example** — using `lambda`, `map()`, and `filter()` together to process data

````python
students = [
    {'name': 'Alice', 'marks': 85},
    {'name': 'Bob', 'marks': 42},
    {'name': 'Charlie', 'marks': 91},
    {'name': 'Diana', 'marks': 55},
]

# Use filter() with a lambda to get only students who passed (marks >= 50)
passed_students = list(filter(lambda s: s['marks'] >= 50, students))
print("Passed:", [s['name'] for s in passed_students])

# Use map() with a lambda to create a list of just the names, uppercased
names_upper = list(map(lambda s: s['name'].upper(), students))
print("Names:", names_upper)

# Combine both: get uppercase names of only the passed students
passed_names_upper = list(map(lambda s: s['name'].upper(), passed_students))
print("Passed names:", passed_names_upper)
````

- Line 8: `filter(lambda s: s['marks'] >= 50, students)` checks each dictionary and keeps only those where the lambda returns `True`.
- Line 12: `map(lambda s: s['name'].upper(), students)` transforms every student dictionary into just its uppercase name.
- Both `filter()` and `map()` return **lazy iterator objects** — wrapping them in `list(...)` forces Python to actually compute and store the results.
- This pattern (`filter` then `map`, or vice versa) is very common for quick, readable data processing without writing full `for` loops.

## 6. Internal Working

- When Python encounters a `def` statement, it creates a **function object** and binds it to the function's name — the function body is **not executed yet**, only stored.
- Each time a function is **called**, Python creates a **new stack frame** (a block of memory) to hold that call's local variables, parameters, and where to return to afterward. This is why local variables inside a function don't exist outside it (**local scope**).
- When the function finishes (hits `return` or the end of its body), its stack frame is **popped off the call stack**, and any return value is passed back to the caller.
- `*args` and `**kwargs` are handled by Python's argument-binding logic: extra positional arguments get collected into a **tuple**, extra keyword arguments into a **dictionary**, before the function body ever runs.
- `lambda` creates a function object exactly like `def` does internally — the only real difference is that `lambda` is restricted to a **single expression** and has no name of its own (though you can still assign it to a variable).
- `map()` and `filter()` return **iterator objects** which compute their results **lazily**, one item at a time, only when you actually iterate over them (e.g., via `list()`, a `for` loop, or `next()`) — they don't build the whole result up front.

## 7. Time & Space Complexity

- **Function call overhead** → each call has a small constant-time cost (**O(1)**) to set up and tear down its stack frame — negligible for most programs, but matters in tight loops with millions of calls.
- **`map(func, iterable)`** → **O(n)** time to apply `func` to each of `n` items; **O(n)** space once materialized into a list (the iterator itself is O(1) space until consumed).
- **`filter(func, iterable)`** → **O(n)** time to test each of `n` items; output space depends on how many items pass (**O(k)** where k ≤ n).
- **Recursive functions** → complexity depends on the recursion depth and branching (covered in detail in the Recursion topic); each recursive call also adds a new stack frame, so deep recursion uses **O(depth)** space on the call stack.
- Overall, using `map`/`filter`/`lambda` instead of manual loops does **not** change the algorithmic complexity — it's the same O(n) work, just written more concisely.

## 8. Common Mistakes

- **Forgetting `return`**: a function without `return` gives back `None`; assigning its result to a variable without checking this can cause silent bugs.
- **Putting default-valued parameters before non-default ones**: `def f(a=1, b):` is a `SyntaxError` — non-default parameters must come first.
- **Using a mutable default argument** like `def f(items=[])`: the same list object is reused across every call, causing values to build up unexpectedly. Fix: use `def f(items=None): items = items or []`.
- **Overusing `lambda` for complex logic**: cramming multi-step logic into a lambda hurts readability — if it needs more than one expression, use a regular `def` function.
- **Confusing `map()`/`filter()` output with a list**: forgetting to wrap them in `list(...)` means you get an iterator object printed instead of the actual values.
- **Shadowing variable names**: naming a function parameter the same as a variable outside the function can cause confusing bugs due to local vs. global scope.
- **Forgetting that `*args` must come before `**kwargs`** in a function definition, and both must come after regular positional/default parameters.

## 9. Best Practices

- Give every function a **clear, descriptive, lowercase_with_underscores name** that describes what it does (`calculate_total`, not `calc` or `func1`).
- Always add a **docstring** (`"""..."""`) explaining what the function does, its parameters, and what it returns.
- Keep functions **small and focused on one task** — if a function is doing many unrelated things, split it into smaller functions.
- Prefer **explicit `return` statements** over relying on side effects (like `print`) inside a function meant to compute a value.
- Use `lambda` only for **short, simple, throwaway logic**, typically as an argument to `map()`, `filter()`, `sorted()`, etc. — use `def` for anything more complex or reused elsewhere.
- Prefer **list comprehensions** over `map()`/`filter()` when readability improves (e.g., `[s['name'] for s in students if s['marks'] >= 50]` is often clearer than combining `filter` and `map`).
- Use type hints for clarity in larger projects: `def add(a: int, b: int) -> int:`.

````python
# Best practice example: list comprehension vs map+filter, both valid
students = [{'name': 'Alice', 'marks': 85}, {'name': 'Bob', 'marks': 42}]

# Using map + filter
result_1 = list(map(lambda s: s['name'], filter(lambda s: s['marks'] >= 50, students)))

# Using a list comprehension (often more readable)
result_2 = [s['name'] for s in students if s['marks'] >= 50]

print(result_1 == result_2)
````

## 10. Interview Questions

**Beginner**
1. **Q: What is the difference between a parameter and an argument?**
   A: A parameter is the placeholder variable name listed in a function's definition; an argument is the actual value passed in when the function is called.
2. **Q: What does a function return if there is no explicit `return` statement?**
   A: It returns `None`.
3. **Q: What is a lambda function?**
   A: A small, anonymous, single-expression function defined using the `lambda` keyword, without needing a `def` or a name.

**Intermediate**

4. **Q: What is the difference between `*args` and `**kwargs`?**
   A: `*args` collects extra **positional** arguments into a tuple; `**kwargs` collects extra **keyword** arguments into a dictionary.
5. **Q: What is the difference between `map()` and `filter()`?**
   A: `map()` applies a function to every item and returns a transformed version of each; `filter()` applies a function that returns True/False and keeps only the items where it's True.
6. **Q: Why is using a mutable default argument (like a list) risky?**
   A: The default object is created only once, at function definition time, and shared across all calls that don't provide their own value — causing unexpected accumulation of data between unrelated calls.

**Advanced**

7. **Q: Why can't a lambda function contain multiple statements?**
   A: By design, `lambda` is limited to a single expression, which is implicitly returned — it exists for short, inline use, not for full function logic with multiple statements, loops, or assignments.
8. **Q: What happens internally when a function is called in Python?**
   A: Python creates a new stack frame containing the function's local variables and parameters, executes the function body using that frame, and pops the frame off the call stack once the function returns, passing the return value back to the caller.
9. **Q: Why are `map()` and `filter()` considered "lazy"?**
   A: They return iterator objects that compute each result only when requested (e.g., during iteration or when converted with `list()`), rather than computing and storing the entire result immediately — this saves memory for large datasets.

## 12. Revision

- Function = reusable named block of code, defined with `def`, called with `function_name(args)`.
- Parameters (definition) vs. arguments (call) — don't confuse the two.
- Types of arguments: positional, keyword, default-valued, `*args` (tuple of extras), `**kwargs` (dict of extras).
- `return` sends a value back; no `return` means the function returns `None`.
- `lambda` = anonymous, single-expression function: `lambda params: expression`.
- `map(func, iterable)` → transforms every item; `filter(func, iterable)` → keeps items where `func` is True. Both are lazy iterators — use `list(...)` to see results.
- Internally: each function call gets its own **stack frame**; local variables live and die with that call.
- List comprehensions often read more clearly than combining `map`+`filter`+`lambda`, but all approaches have the same O(n) complexity.
- **Interview keyword to remember**: "map/filter are lazy iterators; lambda is a single-expression anonymous function; mutable default arguments are a classic gotcha."
