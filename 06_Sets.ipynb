# Python Sets

## 1. Introduction

- **What is it?**
  - A **set** is a built-in Python data type that stores a collection of **unique, unordered** items.
- **Definition**
  - A set is an **unordered, mutable collection of unique, hashable items**, written inside curly braces `{}` or created with `set()`.
- **Why do we need it?**
  - Lists allow duplicates and are ordered; sometimes you only care about **uniqueness** and **fast membership checks**, not order. A set is optimized exactly for that.
- **Where is it used?**
  - Removing duplicates from data, checking membership quickly, comparing groups of data (common items, different items), tagging systems, graph algorithms (visited nodes), and set-covering / greedy algorithms.

> A set answers one question really well: **"is this item in here, and is it unique?"**

## 2. Real-Life Analogy

- A set is like a **guest list at a private party** where no name can be repeated — each guest is checked in only once, no matter how many times they try to enter.
- Another analogy: a **Venn diagram**.
  - Two circles (two sets) — the overlap is the **intersection**, everything inside either circle is the **union**, and the part of one circle not touching the other is the **difference**.
- Real example from the reference book: a fruit basket and a vegetable basket — tomato exists in both, so it's the **intersection**.

## 3. Explanation

- A set stores items **without duplicates** — adding the same item twice has no effect.
- A set has **no guaranteed order** — you cannot rely on the order items are stored or printed in (unlike a list).
- Items in a set must be **hashable** (immutable types like numbers, strings, tuples) — you cannot put a list or another set inside a set.
- Sets are **mutable** — you can add or remove items after creation, but the *items themselves* must be immutable.
- Python supports classic **mathematical set operations**: union, intersection, difference, and symmetric difference.

## 4. Syntax

### Creating a set
```python
my_set = {1, 2, 3}          # set literal
my_set = set([1, 2, 2, 3])  # from a list -> {1, 2, 3}, duplicates removed
empty_set = set()            # CANNOT use {} for empty set -- that creates a dict!
```

### Common set methods
```python
my_set.add(item)         # add a single item
my_set.update(iterable)  # add multiple items from a list/set
my_set.remove(item)      # remove item; raises KeyError if not found
my_set.discard(item)     # remove item; does NOT raise error if not found
my_set.pop()               # removes and returns an arbitrary item
my_set.clear()              # empty the set
len(my_set)                  # number of items
item in my_set                # membership check -- very fast
```

### Set operations (mathematical)
```python
a | b   # union: items in a OR b
a & b   # intersection: items in BOTH a AND b
a - b   # difference: items in a but NOT in b
a ^ b   # symmetric difference: items in a or b, but NOT both
```
- `|`, `&`, `-`, `^` are operators; equivalent methods also exist: `.union()`, `.intersection()`, `.difference()`, `.symmetric_difference()`.

### 5. Examples

**Basic Example** — creating a set and removing duplicates

````python
numbers = [1, 2, 2, 3, 3, 3]
unique_numbers = set(numbers)
print(unique_numbers)   # {1, 2, 3} -- order not guaranteed
````

**Intermediate Example** — union, intersection, difference

````python
fruits = set(["avocado", "tomato", "banana"])
vegetables = set(["beets", "carrots", "tomato"])

print(fruits | vegetables)   # union: everything from both
print(fruits & vegetables)   # intersection: common items -> tomato
print(fruits - vegetables)   # difference: in fruits but not vegetables
print(vegetables - fruits)   # difference the other way
````

**Real-World Example** — removing duplicate results while looping (based on a survey/poll)

````python
favorite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python',
}

print("The following languages have been mentioned:")
for language in set(favorite_languages.values()):
    print(language.title())
````

- Line 1-6: a dictionary mapping people to their favorite language; notice `'jen'` and `'phil'` both picked `'python'`.
- Line 9: `favorite_languages.values()` returns `['python', 'c', 'ruby', 'python']` — has a duplicate.
- Wrapping it in `set(...)` removes the duplicate `'python'`, so each language prints **only once**, regardless of how many people chose it.

## 6. Internal Working

- Internally, a Python set is implemented using a **hash table**, just like a dictionary (in fact, a set behaves like a dictionary with only keys and no values).
- When you add an item, Python computes its **hash value** using `hash(item)` and uses that hash to decide which "bucket" (slot) to store it in.
- To check `item in my_set`, Python computes the hash of `item` and jumps **directly** to the bucket where it would be stored — it does **not** need to scan every item, which is why membership checks are so fast.
- Because sets rely on hashing, only **hashable (immutable)** objects can be stored — this is why you can't put a list inside a set (lists are mutable and unhashable), but you can put a tuple.
- Set order appears "random" because it's based on hash values and internal bucket placement, not insertion order.

## 7. Time & Space Complexity

- **Add an item** `my_set.add(x)` → **O(1)** average — hash lookup decides the bucket directly.
- **Remove an item** `my_set.remove(x)` / `.discard(x)` → **O(1)** average.
- **Membership check** `x in my_set` → **O(1)** average — this is the single biggest reason to use a set instead of a list (list membership check is O(n)).
- **Union / Intersection / Difference** → **O(len(a) + len(b))** — Python must look through elements of both sets once.
- **Space Complexity** → **O(n)** for `n` items, plus some extra reserved space for the hash table (similar overhead to dictionaries).
- Worst case for add/remove/lookup is technically O(n) if there are many hash collisions, but this is rare in practice with Python's hashing.

## 8. Common Mistakes

- **Using `{}` for an empty set** — `{}` actually creates an **empty dictionary**, not a set. Use `set()` instead.
- **Assuming sets keep insertion order** — never rely on the order items print in; if you need order, use a list.
- **Trying to store a mutable object (like a list) inside a set** — raises `TypeError: unhashable type: 'list'`. Use a tuple instead if you need to store multiple values together.
- **Using `.remove()` on an item that doesn't exist** — raises `KeyError`. Use `.discard()` if you're not sure the item is present.
- **Trying to access a set item by index** — sets are unordered, so `my_set[0]` is **not valid** (`TypeError`).
- **Confusing `-` (difference) direction** — `a - b` is different from `b - a`; order matters for difference (but not for union or intersection).

## 9. Best Practices

- Use a set when you need **uniqueness** or **fast membership testing**, not when order matters.
- Use `set()` to quickly de-duplicate a list: `unique_items = set(my_list)`.
- Prefer `.discard()` over `.remove()` when you're unsure if the item exists, to avoid unnecessary error handling.
- Use set operators (`|`, `&`, `-`, `^`) for concise, readable set-math instead of manual loops with `if` checks.
- Convert back to a `list()` if you need to preserve results in an ordered, indexable form after set operations.
- For large-scale membership checking (e.g., "has this ID been seen before?"), a set is usually the right tool over a list.

## 10. Interview Questions

**Beginner**
1. **Q: What is the main difference between a list and a set?**
   A: A list is ordered and allows duplicates; a set is unordered and only stores unique items.
2. **Q: How do you create an empty set?**
   A: Using `set()`. Using `{}` creates an empty dictionary instead.
3. **Q: Can you access a set item using an index like `my_set[0]`?**
   A: No, sets are unordered and unindexed, so this raises a `TypeError`.

**Intermediate**

4. **Q: What is the time complexity of checking membership in a set versus a list?**
   A: O(1) average for a set (hash-based lookup) versus O(n) for a list (linear scan).
5. **Q: What is the difference between `.remove()` and `.discard()` on a set?**
   A: `.remove(x)` raises a `KeyError` if `x` isn't in the set; `.discard(x)` does nothing if `x` isn't present.
6. **Q: How would you find common elements between two lists efficiently?**
   A: Convert both lists to sets and use the intersection operator: `set(list1) & set(list2)`.

**Advanced**

7. **Q: Why can't a list be stored inside a set?**
   A: Sets use hashing internally, and hashing requires immutable objects. Lists are mutable, so they are unhashable and cannot be added to a set. Tuples, which are immutable, can be used instead.
8. **Q: How does Python achieve O(1) average lookup time for sets?**
   A: Sets are backed by a hash table. Python computes `hash(item)` to determine which bucket the item belongs to, then checks that specific bucket directly instead of scanning the whole collection.
9. **Q: What's the difference between `symmetric_difference` (`^`) and regular `difference` (`-`)?**
   A: `a - b` gives items only in `a`; `a ^ b` gives items that are in exactly one of the two sets (in `a` or `b`, but not both).

## 12. Revision

- Set = unordered, mutable collection of **unique**, hashable items.
- Create with `{1, 2, 3}` or `set(iterable)`; empty set MUST use `set()`, not `{}`.
- Core operations: `add`, `update`, `remove`, `discard`, `pop`, `clear`.
- Mathematical operators: `|` union, `&` intersection, `-` difference, `^` symmetric difference.
- Internally backed by a **hash table** → average **O(1)** add/remove/membership check.
- Items must be **hashable/immutable** (numbers, strings, tuples) — no lists or sets inside a set.
- No indexing — sets are not ordered or sliceable.
- **Interview keyword to remember**: "sets give O(1) average membership testing because of hashing, unlike O(n) for lists."
