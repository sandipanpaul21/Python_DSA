## 1. Introduction

- **Operators** are symbols that perform operations on values/variables (add, compare, combine conditions, etc.)
- **Input** means getting data from the user/outside world into your program
- **Output** means displaying/sending data out of your program
- Why needed?
  - Every calculation, filter, or condition in data analysis uses operators (e.g. `df["salary"] > 50000`)
  - Input/Output is how your program communicates with the outside world (files, users, APIs)
- Where used?
  - Filtering rows in Pandas uses comparison + logical operators directly
  - Feature engineering uses arithmetic operators
  - Debugging/logging relies on output (`print`)

## 2. Real-Life Analogy

- **Arithmetic operators** are like a calculator's buttons (+, -, *, /)
- **Comparison operators** are like a bouncer at a club checking a condition: "Is age >= 18?" - lets you in (`True`) or not (`False`)
- **Logical operators** (`and`, `or`, `not`) are like combining multiple bouncer rules: "Must be 18+ AND have a ticket" - both conditions must pass
- **Input** is like a waiter taking your order (program pauses and waits for you to respond)
- **Output** is like the waiter bringing your food out (program shows you the result)

## 3. Explanation

- **Types of operators in Python:**
  - Arithmetic: `+ - * / // % **`
  - Comparison: `== != > < >= <=`
  - Logical: `and or not`
  - Assignment: `= += -= *= /=`
  - Membership: `in`, `not in`
  - Identity: `is`, `is not`
- `/` always gives a `float` result (true division); `//` gives **floor division** (rounds down to nearest whole number)
- `%` gives the **remainder** (modulo) - very commonly used to check even/odd, cyclic patterns
- `**` is exponentiation (power)

> **Trick to remember:** "//` floors, `%` leaves the leftover" - floor division and modulo are a pair; `a = (a // b) * b + (a % b)` always holds true.

## 4. Syntax

```python
# Arithmetic
a + b, a - b, a * b, a / b, a // b, a % b, a ** b

# Comparison (returns True/False)
a == b, a != b, a > b, a < b, a >= b, a <= b

# Logical (combine boolean expressions)
condition1 and condition2
condition1 or condition2
not condition1

# Input - always returns a string
value = input("Enter something: ")

# Output
print(value)                          # basic
print(value, end=" ")                 # custom ending (default is newline)
print(f"Value is {value}")            # f-string formatting
```

````python
a, b = 17, 5
print(a + b, a - b, a * b)
print(a / b)    # true division -> float
print(a // b)   # floor division -> int
print(a % b)    # remainder
print(a ** b)   # power
````

## 5. Examples

### Basic Example

````python
# Basic: comparison and logical operators
age = 20
has_ticket = True

can_enter = age >= 18 and has_ticket
print(can_enter)
````

### Intermediate Example

````python
# Intermediate: taking numeric input safely and computing
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print(f"Sum: {num1 + num2}")
print(f"Product: {num1 * num2}")
print(f"Is num1 divisible by num2: {num1 % num2 == 0}")
````

- `int(input(...))` casts the string input directly into an integer in one line - a very common pattern
- `num1 % num2 == 0` checks divisibility - remainder is 0 means num2 divides num1 exactly

### Real-World Example

````python
# Real-world: simple data validation check before adding a record
# Mimics a check you'd run before inserting a row into a dataset

salary = 45000
department = "Sales"
years_experience = 3

is_valid_record = (
    salary > 0
    and department in ["Sales", "Engineering", "HR", "Finance"]
    and years_experience >= 0
)

print(f"Record valid: {is_valid_record}")
````

- `department in [...]` uses the **membership operator** `in` to check if a value exists in a list - extremely common when validating categorical data
- Combining multiple conditions with `and` is exactly how real data validation/filtering logic is written
- Parentheses `()` around the whole expression just make it readable across multiple lines

## 6. Internal Working

- Comparison and logical operators return **`bool` objects** (`True`/`False`), which Python internally treats as `1`/`0`
- `and`/`or` use **short-circuit evaluation**:
  - `and` stops and returns the first falsy value it finds (doesn't check the rest)
  - `or` stops and returns the first truthy value it finds
- `input()` internally reads from `stdin` (standard input stream) and always returns type `str`
- `print()` internally writes to `stdout` (standard output stream)

> **Trick to remember:** Short-circuiting means `False and expensive_function()` never even calls `expensive_function()`. This is a real performance/interview trick - put cheap/likely-to-fail conditions first in an `and` chain.

````python
def expensive_check():
    print("This ran!")
    return True

result = False and expensive_check()   # expensive_check() never runs due to short-circuit
print(result)
````

## 7. Time and Space Complexity

- Arithmetic and comparison operators on numbers: **O(1)** time
- `in` operator on a **list**: **O(n)** - Python scans the list one item at a time
- `in` operator on a **set/dict**: **O(1)** average - uses hashing internally (important interview point!)
- `print()`/`input()`: effectively O(1) relative to program logic (I/O cost is separate from algorithmic complexity)

> **Interview line:** If you're checking membership repeatedly, convert your list to a set first - it changes `in` from O(n) to O(1) average case.

## 8. Common Mistakes

- Using `=` (assignment) when you meant `==` (comparison) - a very common bug, e.g. `if age = 18:` is a syntax error, but easy to type by mistake
- Forgetting `input()` returns a string - `input() + 5` crashes with `TypeError`
- Using `/` when you actually wanted whole number division (`//`) - can silently introduce float bugs
- Writing `if a == True:` instead of just `if a:` - unnecessary and considered non-Pythonic
- Using `in` on a large list repeatedly in a loop instead of converting to a set first (performance mistake)

## 9. Best Practices

- Write `if is_valid:` instead of `if is_valid == True:` - cleaner and Pythonic
- Use parentheses to group complex logical conditions for readability, even when not strictly required
- Convert lists to sets before doing repeated membership checks (`in`) for performance
- Always validate/cast `input()` before using it in calculations
- Use f-strings for output formatting instead of comma-separated `print()` arguments when formatting matters

## 10. Interview Questions

**Beginner**
- Q: What is the difference between `/` and `//`?
  A: `/` performs true division and always returns a float; `//` performs floor division and returns the result rounded down to the nearest whole number.
- Q: What does the `%` operator do?
  A: Returns the remainder after division (modulo) - commonly used to check even/odd numbers or cyclic patterns.

**Intermediate**
- Q: What is short-circuit evaluation in `and`/`or`?
  A: Python stops evaluating as soon as the result is determined - `and` stops at the first falsy value, `or` stops at the first truthy value, skipping evaluation of the remaining expressions.
- Q: Why is checking membership (`in`) faster on a set than on a list?
  A: Sets use hash tables internally, giving O(1) average-time lookups, while lists require a linear O(n) scan through each element.

**Advanced**
- Q: What is the difference between `is` and `==`?
  A: `==` checks value equality (do they have the same value), while `is` checks identity (are they literally the same object in memory). Two equal values aren't always the same object.
- Q: How does Python's `%` behave differently from some other languages when negative numbers are involved?
  A: Python's `%` result always takes the sign of the divisor (e.g. `-7 % 3` gives `2`, not `-1` as in C/Java), because Python's floor division always rounds toward negative infinity.

## 12. Revision Summary

- Operator categories: Arithmetic, Comparison, Logical, Assignment, Membership, Identity
- `/` = true division (float), `//` = floor division (int/rounded down), `%` = remainder
- `and`/`or` short-circuit - stop early once result is known
- `in` is O(n) on a list, O(1) average on a set/dict - convert for performance
- `input()` always returns a string - cast before using in math
- `print()` supports f-strings, `end=`, and multiple comma-separated arguments
- `is` checks identity, `==` checks value equality - don't confuse them
