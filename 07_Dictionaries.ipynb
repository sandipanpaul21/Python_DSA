# Python Dictionaries

## 1. Introduction

- **What is it?**
  - A **dictionary** is a built-in Python data type that stores data as **key-value pairs**.
- **Definition**
  - A dictionary is an **unordered (insertion-ordered since Python 3.7+ for iteration), mutable collection of key-value pairs**, written inside curly braces `{}`, where each **key** is unique and maps to one **value**.
- **Why do we need it?**
  - Lists let you access data only by position (index). Sometimes you want to access data by a **meaningful label** instead — like looking up a person's age by their name, not by remembering "index 3".
- **Where is it used?**
  - Storing structured records (a person's details, a product's info), counting occurrences of items, caching/memoization, JSON-like data (APIs), configuration settings, and fast key-based lookups.

> A dictionary is Python's version of a **real dictionary book**: you look up a **word (key)** to get its **meaning (value)**.

## 2. Real-Life Analogy

- A dictionary is like a **phone contact list**.
  - The **name** (key) is unique and easy to remember.
  - The **phone number** (value) is what you actually want, but you find it by searching for the name, not by scrolling through numbered positions.
- Another analogy: a **labeled filing cabinet**.
  - Each drawer has a label (key): "Tax Documents", "Medical Records".
  - Inside each drawer is the actual content (value).
  - You go straight to the drawer by its label — you don't search through every drawer one by one.

## 3. Explanation

- A dictionary stores pairs: `key: value`. Each **key must be unique** — if you assign a new value to an existing key, the old value is overwritten.
- **Keys must be immutable/hashable** (strings, numbers, tuples) — just like set items. **Values can be anything**: a number, string, list, or even another dictionary.
- Dictionaries are **mutable** — add, change, or remove key-value pairs anytime.
- Since **Python 3.7**, dictionaries **remember insertion order** when you loop through them (this is now a guaranteed language feature, not an implementation detail).
- You access a value using its key in square brackets: `my_dict[key]`, similar to how you'd use an index for a list, except here the "index" is any hashable value you choose.

## 4. Syntax

### Creating a dictionary
```python
my_dict = {'key1': 'value1', 'key2': 'value2'}
empty_dict = {}                          # this DOES create an empty dict (not a set)
my_dict = dict(key1='value1', key2='value2')  # alternate constructor
```

### Accessing and modifying values
```python
my_dict[key]              # access value by key -- raises KeyError if key missing
my_dict.get(key, default) # access safely -- returns default (or None) if key missing
my_dict[key] = value       # add a new key-value pair OR update an existing key
del my_dict[key]            # remove a key-value pair -- raises KeyError if missing
my_dict.pop(key, default)    # remove and return value for key, with optional default
```

### Common dictionary methods
```python
my_dict.keys()     # view of all keys
my_dict.values()   # view of all values
my_dict.items()    # view of all (key, value) pairs
len(my_dict)         # number of key-value pairs
key in my_dict         # membership check -- checks KEYS only, very fast
```

### Looping through a dictionary
```python
for key, value in my_dict.items():
    print(key, value)

for key in my_dict:        # loops through keys by default
    print(key)

for value in my_dict.values():
    print(value)
```

### 5. Examples

**Basic Example** — creating and accessing a dictionary

````python
alien_0 = {'color': 'green', 'points': 5}
print(alien_0['color'])
print(alien_0['points'])
````

**Intermediate Example** — adding, modifying, and removing key-value pairs

````python
alien_0 = {'color': 'green', 'points': 5}
print(alien_0)

# Add new key-value pairs
alien_0['x_position'] = 0
alien_0['y_position'] = 25
print(alien_0)

# Modify an existing value
alien_0['color'] = 'yellow'
print(alien_0)

# Remove a key-value pair
del alien_0['points']
print(alien_0)
````

**Real-World Example** — looping through a poll result and nesting a list inside a dictionary

````python
favorite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python',
}

for name, language in favorite_languages.items():
    print(name.title() + "'s favorite language is " + language.title() + ".")

# A dictionary that stores a LIST as one of its values
pizza = {
    'crust': 'thick',
    'toppings': ['mushrooms', 'extra cheese'],
}

print("\nYou ordered a " + pizza['crust'] + "-crust pizza with the following toppings:")
for topping in pizza['toppings']:
    print("\t" + topping)
````

- Line 1-6: dictionary mapping each person's name (key) to their favorite language (value).
- Line 8-9: `.items()` returns each key-value pair together, so we unpack them into `name` and `language` in the `for` loop.
- Line 12-15: this dictionary's `'toppings'` key holds a **list**, showing that values are not limited to simple types.
- Line 18-19: we grab the list with `pizza['toppings']` and loop through it separately to print each topping.

## 6. Internal Working

- Internally, a Python dictionary is implemented using a **hash table**, exactly like a set — but each hash "slot" stores a **key-value pair** instead of just a key.
- When you write `my_dict[key]`, Python computes `hash(key)` and jumps directly to the bucket that should contain it — it doesn't scan every pair.
- If two different keys produce the same hash (a **collision**), Python handles it internally using a technique called **open addressing**, so both key-value pairs still get stored correctly and can be retrieved.
- Since **Python 3.7**, dictionaries maintain a separate internal structure that also tracks **insertion order**, so iterating over `.items()`, `.keys()`, or `.values()` returns pairs in the order they were added — this was an implementation detail in 3.6 and became a guaranteed language feature in 3.7.
- Because keys must be hashable, you cannot use a list as a key (unhashable), but you can use a tuple.

## 7. Time & Space Complexity

- **Access by key** `my_dict[key]` → **O(1)** average — direct hash-based lookup.
- **Insert/update a key** `my_dict[key] = value` → **O(1)** average.
- **Delete a key** `del my_dict[key]` / `.pop(key)` → **O(1)** average.
- **Membership check** `key in my_dict` → **O(1)** average — checks keys only, not values.
- **Looping through all items** `.items()` → **O(n)** — must visit every pair once.
- **Space Complexity** → **O(n)** for `n` key-value pairs, plus some extra reserved space (similar to sets).
- Worst case for lookups is O(n) if there are many hash collisions, but this is rare in practice.

## 8. Common Mistakes

- **Accessing a missing key with `[]`** — `my_dict['missing_key']` raises `KeyError`. Use `.get('missing_key', default_value)` to avoid crashing.
- **Assuming dictionary keys can be lists** — raises `TypeError: unhashable type: 'list'`. Use a tuple if you need a compound key.
- **Confusing `.keys()`, `.values()`, `.items()`** — `.keys()` gives only keys, `.values()` gives only values, `.items()` gives both as pairs; using the wrong one causes unpacking errors in loops.
- **Modifying a dictionary while looping over it** — adding/removing keys during iteration raises `RuntimeError: dictionary changed size during iteration`. Loop over a copy: `for key in list(my_dict.keys())`.
- **Relying on old Python versions (<3.7) for insertion order** — order was not guaranteed before 3.7; always check which Python version code needs to support.
- **Overwriting a key accidentally** — assigning to an existing key silently replaces its value instead of adding a new pair; make sure key names are unique on purpose.

## 9. Best Practices

- Use **descriptive, meaningful key names** (`'first_name'`, not `'fn'`).
- Prefer `.get(key, default)` over `[key]` when a key might not exist, to avoid try/except boilerplate.
- Use `.setdefault(key, default)` when you want to insert a default value only if the key is missing.
- Keep dictionaries **flat when possible** — avoid deeply nested dictionaries-in-dictionaries-in-lists; it hurts readability.
- Use dictionary/**comprehensions** for concise transformations: `{k: v.upper() for k, v in my_dict.items()}`.
- Use `in` for key membership checks — it's O(1) and far more Pythonic than looping manually.
- When counting occurrences, consider `collections.Counter` or `dict.get(key, 0) + 1` pattern instead of manual if-else checks.

````python
# Dictionary comprehension example
scores = {'alice': 85, 'bob': 45, 'charlie': 92}
passing = {name: score for name, score in scores.items() if score >= 50}
print(passing)
````

## 10. Interview Questions

**Beginner**
1. **Q: What is a dictionary in Python?**
   A: A collection of key-value pairs where each key is unique and used to access its associated value.
2. **Q: How do you safely access a key that might not exist?**
   A: Use `.get(key, default)`, which returns a default value instead of raising an error.
3. **Q: What happens if you assign a value to an existing key?**
   A: The old value is overwritten by the new one; keys are always unique.

**Intermediate**

4. **Q: What is the difference between `.keys()`, `.values()`, and `.items()`?**
   A: `.keys()` returns all keys, `.values()` returns all values, and `.items()` returns all key-value pairs as tuples.
5. **Q: Can a list be used as a dictionary key? Why or why not?**
   A: No, because dictionary keys must be hashable, and lists are mutable and therefore unhashable. Tuples can be used instead since they're immutable.
6. **Q: Since which Python version is dictionary insertion order guaranteed?**
   A: Python 3.7 (it was an implementation detail in 3.6 CPython, made an official language guarantee in 3.7).

**Advanced**

7. **Q: How does Python resolve hash collisions in a dictionary?**
   A: Using open addressing — when two keys hash to the same slot, Python probes for another open slot internally, so both entries are still stored and retrievable correctly.
8. **Q: What is the time complexity of dictionary lookups, and why?**
   A: O(1) on average because dictionaries use a hash table — the key's hash directly determines which slot to check, avoiding a full scan. Worst case is O(n) with many collisions, though rare in practice.
9. **Q: How would you count the frequency of words in a large text using a dictionary efficiently?**
   A: Loop through the words and use `word_counts[word] = word_counts.get(word, 0) + 1`, or use `collections.Counter(words)` which does this internally in optimized C code.

## 12. Revision

- Dictionary = mutable collection of unique **key-value pairs**, `{key: value}`.
- Keys must be hashable/immutable; values can be anything.
- Core operations: `[]` access (raises `KeyError` if missing), `.get()` (safe access), `del`, `.pop()`.
- Looping tools: `.keys()`, `.values()`, `.items()`.
- Since Python 3.7, dictionaries preserve **insertion order**.
- Internally backed by a **hash table** → average **O(1)** access/insert/delete/membership check.
- Common real-world use: counting frequencies, storing structured records, nesting lists/dicts for complex data (like JSON).
- **Interview keyword to remember**: "dict lookups are O(1) average due to hashing; keys must be hashable, values don't need to be."
