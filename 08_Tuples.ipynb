# Python Tuples

## 1. Introduction

- **What is it?**
  - A **tuple** is a built-in Python data type that stores an **ordered collection of items that cannot be changed** after creation.
- **Definition**
  - A tuple is an **ordered, immutable collection of items**, written inside parentheses `()`, separated by commas.
- **Why do we need it?**
  - Sometimes you want to guarantee that a collection of values **never changes** during the life of a program (e.g., fixed dimensions, coordinates, configuration). Python calls unchangeable values **immutable**, and an immutable list is called a tuple.
- **Where is it used?**
  - Fixed data like coordinates `(x, y)`, RGB colors `(255, 0, 0)`, function return values (returning multiple values at once), dictionary keys (since tuples are hashable), and any data that should stay constant and protected from accidental changes.

> A tuple is a list that is **locked** — you can look at it, but you can't modify it.

## 2. Real-Life Analogy

- A tuple is like a **printed birth certificate** — the date of birth, name, and place are fixed. You cannot edit them once issued; you can only read them.
- Another analogy: a **combination lock's factory-set numbers** — the sequence is fixed and won't change even if you use it a thousand times.
- Compare with a list: a **shopping list** (editable, cross things out, add new items) vs. a tuple: a **receipt** (fixed record of what happened, unchangeable after being printed).

## 3. Explanation

- A tuple looks just like a list, but uses **parentheses `()`** instead of square brackets `[]`.
- Once created, you **cannot add, remove, or change** items in a tuple — trying to do so raises a `TypeError`.
- Like lists, tuple items are accessed by **index**, starting from 0, and support slicing.
- Tuples can hold **mixed data types**, just like lists.
- Because tuples are immutable, they are **hashable** (as long as all their elements are also hashable) — meaning they can be used as **dictionary keys** or stored inside a **set**, unlike lists.
- A tuple with a **single item** needs a trailing comma to be recognized as a tuple: `(50,)` — without the comma, `(50)` is just the number 50 in parentheses.

## 4. Syntax

### Creating a tuple
```python
my_tuple = (item1, item2, item3)
single_item_tuple = (50,)    # comma is REQUIRED for a single-item tuple
empty_tuple = ()
my_tuple = tuple([1, 2, 3])   # convert a list to a tuple
```

### Accessing elements
```python
my_tuple[index]         # single item, same as list indexing
my_tuple[start:end]      # slice, same rules as list slicing
my_tuple[-1]               # last item (negative indexing)
```

### What you CANNOT do
```python
my_tuple[0] = 250   # TypeError: 'tuple' object does not support item assignment
my_tuple.append(x)   # AttributeError: tuples have no append method
```

### What you CAN do
```python
len(my_tuple)          # number of items
x in my_tuple            # membership check
my_tuple.count(value)     # count how many times a value appears
my_tuple.index(value)      # find the index of a value
a, b = my_tuple              # tuple unpacking, if tuple has 2 items
```

### "Writing over" a tuple
```python
my_tuple = (200, 50)
my_tuple = (400, 100)   # this is allowed -- you're creating a NEW tuple,
                          # not modifying the old one
```

### 5. Examples

**Basic Example** — creating and accessing a tuple

````python
dimensions = (200, 50)
print(dimensions[0])
print(dimensions[1])
````

**Intermediate Example** — trying to modify a tuple (shows immutability) and writing over it

````python
dimensions = (200, 50)
try:
    dimensions[0] = 250
except TypeError as e:
    print("Error:", e)

# You CAN reassign the whole variable to a brand new tuple
dimensions = (400, 100)
print(dimensions)
````

**Real-World Example** — looping through a tuple and tuple unpacking

````python
# A tuple of fixed menu items for a restaurant that won't change during service
menu_items = ('idli', 'dosa', 'vada')
for item in menu_items:
    print(item)

# Tuple unpacking: a function-like pattern for returning coordinates
point = (10, 20)
x, y = point
print("x =", x, ", y =", y)

# Real-world: representing an RGB color -- fixed values, used as a dictionary key
color_names = {
    (255, 0, 0): 'red',
    (0, 255, 0): 'green',
    (0, 0, 255): 'blue',
}
print(color_names[(255, 0, 0)])
````

- Line 2-3: a tuple of 3 fixed menu items looped over with a `for` loop, same syntax as looping a list.
- Line 6-8: **tuple unpacking** — assigns `10` to `x` and `20` to `y` in a single line, based on position.
- Line 11-15: tuples are used as **dictionary keys** here because they're hashable (a list could not be used this way).

## 6. Internal Working

- Internally, CPython implements a tuple as a **fixed-size array of references** to objects, similar to a list, but the array's size is **set once at creation time and never resized**.
- Because tuples never change size or content, Python can **optimize memory allocation** for them — tuples generally use less memory than an equivalent list.
- Since tuples cannot change, Python can also apply an optimization called **interning/caching** for small tuples in some cases, and the CPython compiler can sometimes treat tuple literals as **constants** built once, unlike lists, which are always rebuilt at runtime.
- Because tuples are immutable, their **hash value** can be computed once and reused — this is exactly why tuples (unlike lists) can be dictionary keys or set members.

## 7. Time & Space Complexity

- **Access by index** `my_tuple[i]` → **O(1)** — same direct-addressing as a list.
- **Slicing** `my_tuple[a:b]` → **O(k)** where k is the slice length — a new tuple is created.
- **Search (`in`, `.index()`)** → **O(n)** — linear scan, same as a list.
- **Creation** → **O(n)** to build from n items.
- **Space Complexity** → **O(n)**, generally **slightly less memory** than an equivalent list because tuples don't need extra capacity reserved for future growth (since they can't grow).
- No complexity for insert/append/delete/sort — **these operations don't exist** on tuples because tuples are immutable.

## 8. Common Mistakes

- **Forgetting the comma for a single-item tuple**: `(50)` is just an integer in parentheses, not a tuple. You need `(50,)`.
- **Trying to modify a tuple directly**: `my_tuple[0] = 5` raises `TypeError`. If you need to change data, use a list instead, or build a new tuple.
- **Confusing "tuple is immutable" with "tuple's contents are always immutable"**: if a tuple contains a **mutable object** like a list, that inner list CAN still be modified — only the tuple's own structure (which items it holds, in what order) is locked.
- **Using tuples when you actually need to grow/shrink a collection** — pick a list instead if the collection size will change over time.
- **Forgetting parentheses aren't strictly required** — Python allows `t = 1, 2, 3` without parentheses (this is still a tuple), which can be confusing to read; always use parentheses for clarity.

## 9. Best Practices

- Use a tuple whenever data should be **fixed and protected from accidental changes** — e.g., coordinates, RGB values, configuration constants.
- Prefer tuples over lists for **function return values with multiple pieces of data** — it signals to other developers that the values form a fixed, related group.
- Use tuple unpacking for cleaner code: `x, y = get_coordinates()` instead of manually indexing.
- Use tuples as **dictionary keys** or **set members** when you need a compound, hashable identifier.
- Name tuple variables clearly to reflect their fixed nature, e.g., `dimensions`, `rgb_color`, `point`.
- If you find yourself wanting to modify a "tuple" often, that's a sign you should be using a list instead.

````python
# Demonstrating that a list INSIDE a tuple can still be changed
data = (1, 2, [3, 4])
data[2].append(5)     # allowed -- we're modifying the LIST inside the tuple, not the tuple itself
print(data)            # (1, 2, [3, 4, 5])
````

## 10. Interview Questions

**Beginner**
1. **Q: What is the main difference between a list and a tuple?**
   A: A list is mutable (can be changed after creation) and uses `[]`; a tuple is immutable (cannot be changed) and uses `()`.
2. **Q: How do you create a tuple with only one item?**
   A: Add a trailing comma: `(50,)`. Without the comma, it's just a number in parentheses.
3. **Q: Can you access tuple items by index?**
   A: Yes, tuples support indexing and slicing exactly like lists.

**Intermediate**

4. **Q: Why would you use a tuple instead of a list?**
   A: When you want to guarantee the data won't be accidentally changed (fixed dimensions, coordinates), and because tuples can be used as dictionary keys or set elements since they're hashable.
5. **Q: Can a tuple contain a mutable object like a list? What happens if you modify it?**
   A: Yes. The tuple itself still can't be restructured (no adding/removing elements), but a mutable object stored inside it (like a list) can still be modified in place.
6. **Q: What is tuple unpacking, and give an example.**
   A: Assigning multiple variables at once from a tuple's items, positionally: `x, y = (10, 20)` assigns `x=10` and `y=20`.

**Advanced**

7. **Q: Why are tuples generally faster and more memory-efficient than lists?**
   A: Because tuples have a fixed size known at creation time, CPython can allocate exactly the memory needed without reserving extra space for future growth (which lists do for amortized O(1) append), and since tuples are immutable, some can be optimized/cached by the interpreter.
8. **Q: Why can tuples be used as dictionary keys but lists cannot?**
   A: Dictionary keys must be hashable, and hashability requires immutability — a tuple's contents can't change after creation, so its hash value stays constant and reliable, while a list's contents (and thus its "identity" for hashing purposes) could change at any time.
9. **Q: What happens internally when you do `my_tuple = (400, 100)` after `my_tuple` already pointed to a different tuple?**
   A: This does not modify the original tuple object; it simply makes the variable name `my_tuple` point to a brand-new tuple object in memory. The old tuple, if unreferenced elsewhere, becomes eligible for garbage collection.

## 12. Revision

- Tuple = ordered, **immutable** collection, created with `()`.
- Indexing and slicing work exactly like lists; indexing starts at **0**.
- Single-item tuple needs a trailing comma: `(50,)`.
- No `append`, `insert`, `remove`, `sort` — these don't exist because tuples can't change.
- You CAN reassign a variable to point to an entirely new tuple.
- A tuple containing a mutable object (like a list) still allows that inner object to be modified.
- Tuples are **hashable** (if their contents are hashable) → usable as dictionary keys / set elements, unlike lists.
- Internally: fixed-size array of references, generally more memory-efficient than an equivalent list.
- **Interview keyword to remember**: "tuples are immutable and hashable, which is why they can be dictionary keys — lists can't."
