## 1. Introduction

* **What is it?** Time and space complexity is a way to describe how the running time and memory usage of an algorithm grow as the input size grows.
* **Definition:** It is a measure of the number of operations (time) or the amount of memory (space) an algorithm needs, expressed as a function of input size `n`, written using **Big O notation**, e.g. `O(n)`.
* **Why do we need it?**
  * Lets you compare two algorithms without depending on the speed of a particular computer.
  * Predicts how an algorithm behaves on large inputs, not just the small input you tested with.
  * It is one of the most heavily tested concepts in coding interviews.
* **Where is it used?**
  * Choosing between two ways of solving the same problem (for example, searching a list).
  * Deciding which data structure to use (list vs set vs dict).
  * Estimating whether code will still run fast enough when the data grows from 100 rows to 10 million rows.

## 2. Real-Life Analogy

* **Looking up a word in a dictionary:** You don't read every page from the start. You open roughly the middle, see if your word comes before or after, and repeat on the smaller half. This halving process is what `O(log n)` feels like.
* **Looking up a name in an unsorted pile of papers:** You must check papers one by one until you find it, or reach the end. This is `O(n)` — the work grows directly with the pile size.
* **A cloakroom with numbered drawers:** Each item you store needs its own drawer. The more items you check in, the more drawers (memory) you need — this is the everyday version of space complexity.
* **Comparing every pair of people in a room to find two with the same birthday:** For every person, you check against every other person. Double the people, and the number of comparisons roughly quadruples. This is the feel of `O(n^2)`.
* **Folding a sheet of paper in half repeatedly:** Each fold doubles the number of layers. Very few folds are needed to get a huge number of layers — this is why `O(log n)` grows so slowly compared to `O(n)`.

---

## 3. Explanation

* Big O notation does not measure time in seconds. It measures how the **number of operations grows** as input size `n` increases.
* We almost always analyze the **worst case** — the maximum number of operations an algorithm could possibly take, so you have a guarantee it will never be slower than that.
* Best case and average case also exist:
  * Best case: the fewest operations possible (for example, finding the item on the very first check).
  * Average case: the typical number of operations across many different inputs.
  * Worst case: the most operations possible. This is the default unless stated otherwise.
* When writing Big O, we drop constants and lower-order terms, because we only care about how things scale for very large `n`:
  * `3n + 5` is written as `O(n)`.
  * `n^2 + n` is written as `O(n^2)`, because the `n^2` part dominates once `n` is large.
* Two algorithms can have the same Big O time but still run at different real-world speeds, because of constants Big O ignores (for example, one does more work per step than the other). Big O tells you about **growth rate**, not raw speed.
* Time complexity and space complexity are analyzed the same way, just counting different things: time complexity counts operations, space complexity counts extra memory used (not counting the input itself).

---

## 4. Syntax

Big O is written as:
```
O(f(n))
```
* `n` — the size of the input.
* `f(n)` — a function describing how operations or memory grow with `n`.

Rules for working out Big O from code:

* **Sequential statements** — add their complexities.
  ```python
  step_one()   # O(n)
  step_two()   # O(n)
  # total: O(n) + O(n) = O(n)
  ```
* **Nested loops** — multiply their complexities.
  ```python
  for i in range(n):        # runs n times
      for j in range(n):    # runs n times for each i
          do_something()
  # total: O(n) * O(n) = O(n^2)
  ```
* **Conditionals (if/else)** — take the branch with the worse complexity, since worst case considers the slowest path.
* **Recursive calls** — multiply the work per call by the number of calls (depth of recursion), and add the memory used by the call stack for space complexity.
* Drop constants: `O(2n)` becomes `O(n)`.
* Drop lower-order terms: `O(n^2 + n)` becomes `O(n^2)`.

## 5. Examples

### Basic Example — O(1)
```python
def get_first(my_list):
    return my_list[0]
```
* Accessing an item by index takes exactly one step, no matter how big `my_list` is.
* Time Complexity: O(1) — constant, doesn't depend on input size.
* Space Complexity: O(1) — no extra memory grows with input.

### Intermediate Example — O(n) vs O(n^2)
```python
def print_all(my_list):          # O(n)
    for item in my_list:
        print(item)

def print_all_pairs(my_list):    # O(n^2)
    for a in my_list:
        for b in my_list:
            print(a, b)
```
* `print_all` touches each item exactly once: O(n) time.
* `print_all_pairs` touches every combination of two items: for each of the `n` outer items, the inner loop runs `n` times, giving `n * n = n^2` total operations: O(n^2) time.
* Both use O(1) extra space — no new data structure grows with `n`.


### Real-World Example — Searching for a Contact
```python
def find_contact_simple(contacts, target):       # unsorted list
    for name in contacts:
        if name == target:
            return True
    return False
```
* `find_contact_simple` checks names one by one in an unsorted list: O(n) time, because in the worst case it checks every single contact.

## 6. Internal Working

* **How operations are actually counted:** every comparison, assignment, and arithmetic operation inside a loop or function counts as one "step" when reasoning about Big O — Python doesn't expose this directly, but you can think of each line inside a loop body as contributing to the total.
* **Recursion and the call stack:** every recursive function call adds a new frame onto Python's internal call stack, holding that call's local variables. If a recursive function calls itself `n` times deep, it uses O(n) extra space for the stack — this is why a recursive function can raise a `RecursionError` for very deep recursion, even though a loop doing the same logical work would use only O(1) space.
* **Hidden costs inside built-in operations** — this matters a lot in real Python code:
  * Indexing a list by position, `my_list[i]`, is O(1).
  * Checking membership with `in` is O(n) on a list, but O(1) on average for a `set` or `dict`, because of hashing.
  * `list.append(x)` is O(1) on average (amortized); `list.insert(0, x)` is O(n), because every existing element has to shift over by one position.
  * Looking up a value by key in a `dict`, `my_dict[key]`, is O(1) on average, again due to hashing.
* This is why putting an `in` check on a list inside a loop silently turns an O(n) algorithm into an O(n^2) one — the hidden cost inside the built-in operation is easy to miss.

## 7. Common Complexity Classes (Time and Space)

Ranked from fastest-growing-slowest to fastest-growing-fastest:

* **O(1)** — Constant. Example: accessing an array element by index.
* **O(log n)** — Logarithmic. Example: binary search on a sorted list.
* **O(n)** — Linear. Example: checking every item in a list once.
* **O(n log n)** — Linearithmic. Example: efficient sorting algorithms like merge sort.
* **O(n^2)** — Quadratic. Example: comparing every pair of items with nested loops, like selection sort.
* **O(2^n)** — Exponential. Example: trying every possible subset of a set of items.
* **O(n!)** — Factorial. Example: trying every possible ordering of `n` items, like a brute-force traveling salesperson solution.

For space complexity, the same classes apply, but they describe **extra memory used**, not counting the input itself:
* O(1) space — a fixed number of variables, regardless of input size (most simple loops).
* O(n) space — memory grows with input, for example building a new list of size `n`, or recursion that goes `n` levels deep.
* O(n^2) space — for example, storing a full grid or table of size `n` by `n`.

**Why these complexities are what they are:**
* O(log n) appears whenever a problem's size is repeatedly cut in half (or by some fixed fraction) at each step.
* O(n) appears whenever every item must be visited exactly once.
* O(n^2) appears whenever every item must be compared against every other item.
* O(2^n) and O(n!) appear when an algorithm must consider every possible combination or ordering — these become unusable very quickly as `n` grows.

## 8. Common Mistakes

* Thinking Big O measures time in seconds — it measures growth in number of operations, not real-world speed.
* Assuming two nested loops always mean O(n^2) — if the inner loop runs a fixed number of times (not depending on `n`), it's actually O(n).
* Forgetting the hidden cost of built-in operations, such as using `in` on a list inside a loop, accidentally creating O(n^2) behavior.
* Confusing best case with worst case — an algorithm that "got lucky" and found the answer immediately did not actually become a better algorithm; the worst case is still what matters.
* Ignoring space complexity, especially the memory used by recursion's call stack, which can cause a program to crash on large inputs even if the time complexity looks fine.
* Adding complexities when they should be multiplied (nested loops), or multiplying when they should be added (sequential, independent loops).

---

## 9. Best Practices

* Always state both time complexity and space complexity in an interview, even if only one was asked — it shows complete understanding.
* Default to analyzing the worst case unless the question specifically asks for average or best case.
* Choose the data structure that matches your access pattern: use a `set` or `dict` instead of a `list` whenever you need fast membership checks or lookups.
* Watch out for nested loops that compare every pair of items — often this can be reduced to a single pass using a `set` or `dict` (a very common interview pattern, for example the "two sum" problem).
* Don't over-optimize prematurely for code that doesn't need it — but always know the complexity of the code you're writing, especially before it goes into production with large data.
* When discussing complexity out loud, explain *why* the complexity is what it is, not just state the final O() answer — interviewers value the reasoning as much as the answer.

## 10. Interview Questions

**Beginner**
1. What does Big O notation measure?
   Answer: It measures how the number of operations (or memory used) grows as the input size grows, not the actual time in seconds.
2. What is the time complexity of accessing an element in a list by its index?
   Answer: O(1) — constant time, regardless of list size.
3. What is the time complexity of a single loop that runs once for every item in a list of size `n`?
   Answer: O(n) — linear time.

**Intermediate**
1. Why is checking membership with `in` faster on a `set` than on a `list`?
   Answer: A `list` requires checking elements one by one in the worst case, O(n), while a `set` uses hashing to jump almost directly to where the item would be, giving O(1) on average.
2. What is the time complexity of two nested loops, each running `n` times?
   Answer: O(n^2), because the inner loop runs fully for every iteration of the outer loop.
3. What's the difference between best case, average case, and worst case complexity?
   Answer: Best case is the minimum operations possible for some lucky input; average case is the typical number across many inputs; worst case is the maximum possible, and is the default used for comparing algorithms reliably.

**Advanced**
1. Why can a recursive solution use more space than an equivalent loop-based solution, even if both have the same time complexity?
   Answer: Each recursive call adds a frame to the call stack that stays in memory until that call returns, so a recursion that goes `n` levels deep uses O(n) extra space, while a loop doing the same logical work typically uses only O(1) extra space.
2. Why does Big O notation ignore constants, and when might constants actually matter in practice?
   Answer: Big O describes growth rate for large `n`, where constants become insignificant compared to the dominant term; however, constants can matter in practice when comparing two algorithms with the *same* Big O complexity, since one may still be consistently faster due to less work per operation.
3. A function loops through a list of size `n`, and inside the loop it checks membership in another list of size `m` using `in`. What is the overall time complexity, and how could it be improved?
   Answer: O(n * m), because each of the `n` iterations does an O(m) membership check; this can be improved to O(n + m) by converting the second list into a `set` first, making each membership check O(1).

## 12. Revision

* Big O notation describes how operations (time) or memory (space) grow as input size `n` grows — not actual seconds or bytes.
* We typically analyze the worst case, since it gives a reliable upper bound.
* Sequential code adds complexities; nested loops multiply complexities; conditionals take the worse branch.
* Common classes, fastest to slowest: O(1), O(log n), O(n), O(n log n), O(n^2), O(2^n), O(n!).
* O(log n) comes from repeatedly halving the problem (like binary search); O(n^2) comes from comparing every pair of items.
* Built-in operations have their own hidden complexity: list `in` is O(n), but set/dict lookup is O(1) on average due to hashing.
* Recursion uses extra space proportional to its depth, because of the call stack — this is a space complexity cost that loops usually avoid.
* Big O ignores constants and lower-order terms, focusing only on the dominant term for large `n`.
* In interviews, always state both time and space complexity, and explain the reasoning behind the answer, not just the final notation.
