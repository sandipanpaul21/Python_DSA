## 1. Introduction

- An **array** is a collection of items stored at contiguous memory locations, all of the **same data type**
- Python does **not** have a built-in low-level array type like C/Java
- In Python, "array" can mean one of three different things - this confusion is a very common interview trap:
  1. **`list`** - Python's built-in dynamic array, can hold mixed types (most commonly what people mean by "array" in Python)
  2. **`array` module** (`array.array`) - a real typed array, all elements must be the same type, more memory-efficient than a list
  3. **NumPy `ndarray`** - the array used in Data Science (fixed type, vectorized operations, contiguous memory) - covered in depth on Day 3
- Why do we need arrays?
  - Store multiple related values under one variable name
  - Enable fast indexed access, iteration, and bulk operations
  - Foundation for almost every data structure used in interviews (matrices, grids, sliding window problems)
- Where used?
  - Storing a sequence of numbers/records (temperatures, prices, scores)
  - Any algorithm question involving sequences (searching, sorting, two-pointer)
  - Numerical computing (NumPy arrays are THE core data structure of Data Science)

> **Interview trap:** If asked "does Python have arrays?", the safe answer is: "Python's `list` acts as a dynamic array. For a real fixed-type array, Python has the `array` module, and for numerical work, NumPy's `ndarray` is used." Naming all three shows depth.

## 2. Real-Life Analogy

- A **list** (Python's general-purpose array) is like a **row of lockers that can hold anything** - one locker has a book, the next has a shoe, the next has a laptop. Flexible, but a bit wasteful because each locker must be able to hold "anything"
- The **`array` module** is like a **row of identical labeled boxes** - every box is built to hold exactly one type of item (say, only integers). More compact and efficient, but less flexible
- A **NumPy array** is like a **factory conveyor belt with fixed-size slots** - extremely efficient, built for bulk processing of numbers, all slots are the same type and size, and operations can be applied to the whole belt at once (vectorization)

## 3. Explanation

- **Python `list`**:
  - Dynamic - grows/shrinks automatically
  - Can store mixed types (`[1, "a", 3.5, True]`)
  - Internally implemented as a dynamic array of **pointers/references** to objects, not the raw values themselves
- **`array` module**:
  - Must specify a **typecode** (e.g. `'i'` for int, `'d'` for float) when creating
  - All elements must match that type - raises `TypeError` otherwise
  - Stores actual values compactly (not pointers), so it uses less memory than a list of the same length
- **NumPy `ndarray`**:
  - Fixed type, fixed size (resizing creates a new array)
  - Stored in one contiguous block of raw memory (no per-element Python object overhead)
  - Supports vectorized math (`arr * 2` multiplies every element without an explicit loop)
  - This is why NumPy is fast - operations run in compiled C, not the Python interpreter loop

> **Trick to remember:** "list = flexible but slow for numeric bulk-ops, array module = typed but rarely used in practice, NumPy ndarray = the real workhorse for Data Science."

## 4. Syntax

```python
# Python list (acts as a general-purpose dynamic array)
my_list = [1, 2, 3, 4]

# array module - typed array
from array import array
int_array = array('i', [1, 2, 3, 4])   # 'i' = signed int typecode
float_array = array('d', [1.5, 2.5])   # 'd' = double (float) typecode

# NumPy array (most relevant for Data Science, deep dive on Day 3)
import numpy as np
np_array = np.array([1, 2, 3, 4])
```

- `'i'`, `'d'`, `'f'`, `'b'` etc. are **typecodes** that tell the `array` module what type of data it will hold
- `array(typecode, iterable)` - the constructor always needs both arguments
- Common `array` methods: `.append()`, `.insert()`, `.remove()`, `.pop()`, `.index()` - same names as list methods

````python
# List as a general-purpose array
scores = [88, 92, 79, 95]
print(scores, type(scores))
````

### Basic Example

````python
# Basic: using array module for a strictly typed integer array
from array import array

ages = array('i', [25, 30, 35, 40])
print(ages)
print(ages[1])          # indexed access, same as list
ages.append(45)         # supported, just like list
print(ages)
````

- `array('i', [...])` creates a typed integer array - trying to insert a string would raise `TypeError`
- Indexing (`ages[1]`), appending, and iteration work just like a list

### Intermediate Example

````python
# Intermediate: comparing memory usage - list vs array module
import sys
from array import array

py_list = list(range(1000))
typed_array = array('i', range(1000))

print("List size (bytes):", sys.getsizeof(py_list))
print("Array size (bytes):", sys.getsizeof(typed_array))
````

- `sys.getsizeof()` shows the memory footprint
- The typed `array` is smaller because it stores raw integers directly, while a `list` stores pointers to separate Python integer objects scattered in memory
- This exact gap - but far larger - is why NumPy (contiguous, typed, no per-element object overhead) crushes plain lists for numeric work

### Real-World Example

````python
# Real-world: NumPy array for a tiny "sensor readings" dataset
import numpy as np

temperature_readings = np.array([21.5, 22.3, 19.8, 23.1, 20.0])

print("Readings:", temperature_readings)
print("Average temperature:", temperature_readings.mean())
print("Above 21 degrees:", temperature_readings[temperature_readings > 21])
````

- `np.array([...])` builds a NumPy array (deep dive on Day 3) - shown here to make clear WHY "array" matters for Data Science
- `.mean()` is a vectorized operation - computes the average over the whole array in one call, no manual loop
- `temperature_readings > 21` is **boolean masking** - it filters the array using a condition directly (previewed here, covered fully on Day 3)
- This is the real reason interviewers ask about arrays: everything in NumPy/Pandas builds on this concept

## 6. Internal Working

- **`list`** internally:
  - Implemented in CPython as a **dynamic array of pointers** to `PyObject`s
  - When it grows beyond current capacity, Python allocates a **larger block** (over-allocates a bit extra) and copies existing pointers over - this is why occasional `append()` calls are more expensive than others (amortized O(1), not strictly O(1) every time)
- **`array` module**:
  - Stores raw C-type values directly in one contiguous block (no per-element Python object wrapper)
  - Still resizes dynamically like a list, but each element takes far less memory
- **NumPy `ndarray`**:
  - One contiguous block of memory, fixed dtype, fixed shape at creation
  - Operations are implemented in **compiled C loops**, avoiding Python's slow per-element interpreter loop entirely
  - This is called **vectorization** - deep dive on Day 3

> **Trick to remember:** Python `list` = array of pointers (flexible, more memory, slower bulk math). NumPy array = array of raw values (rigid, less memory, much faster bulk math).

## 7. Time and Space Complexity

- **List / array module** (same asymptotic behavior for these operations):
  - Indexed access `arr[i]`: **O(1)** - direct memory offset calculation
  - Append at end: **O(1)** amortized (occasional resize is O(n), but averages out)
  - Insert/delete at beginning or middle: **O(n)** - all following elements must shift
  - Search (`in`, `.index()`): **O(n)** - linear scan
  - Space: **O(n)**
- **NumPy array**: same Big-O pattern for indexing (O(1)), but with a much smaller constant factor and support for vectorized O(n) bulk operations that avoid Python-level loop overhead entirely

> **Interview line:** "Indexing is O(1) because arrays store elements at contiguous memory - the address of index `i` is calculated directly as `base_address + i * element_size`, no traversal needed."

## 8. Common Mistakes

- Calling a Python `list` an "array" without knowing the distinction between `list`, `array` module, and `np.ndarray` - fine casually, but weak in an interview if you can't explain the difference when pushed
- Assuming inserting/deleting from the **front** of a list is O(1) - it's actually O(n) because every other element must shift
- Mixing types in an `array.array` (it will raise `TypeError` - unlike a regular list)
- Forgetting that resizing a NumPy array (`np.append`) creates a **brand-new array** every time - unlike a Python list's amortized growth, this is O(n) on every call, so it's a real performance trap in loops

## 9. Best Practices

- Use a plain `list` for small, flexible, mixed-type collections
- Use `array.array` only when memory efficiency matters and all elements are one primitive type (rare in practice)
- Use **NumPy arrays** for any numeric/bulk computation - this is the actual standard in Data Science work
- Never grow a NumPy array inside a loop with repeated `np.append()` - build a Python list first, then convert once with `np.array(my_list)` at the end
- Prefer vectorized operations (`arr * 2`, `arr[arr > 5]`) over manual `for` loops on NumPy arrays for both speed and readability

## 10. Interview Questions

**Beginner**
- Q: Does Python have a built-in array type?
  A: Not a low-level fixed-type array like C. Python's `list` behaves as a general-purpose dynamic array; there is also the `array` module for typed arrays, and NumPy's `ndarray` for numerical computing.
- Q: What is the time complexity of indexing into a list, `arr[i]`?
  A: O(1) - constant time, because the memory address of any index can be calculated directly.

**Intermediate**
- Q: What is the key difference between a Python `list` and the `array` module's `array`?
  A: A `list` can hold mixed data types and stores pointers to objects; `array.array` requires all elements to be the same primitive type and stores raw values contiguously, using less memory.
- Q: Why is inserting at the beginning of a list O(n) instead of O(1)?
  A: Because all existing elements must shift one position to make room, which takes time proportional to the list's length.

**Advanced**
- Q: Why are NumPy arrays faster than Python lists for numerical operations?
  A: NumPy arrays store data as a contiguous block of a single fixed type with no per-element Python object overhead, and operations are executed as vectorized, compiled C loops instead of Python's interpreted per-element loop.
- Q: What does "amortized O(1)" mean in the context of list `.append()`?
  A: Most appends are O(1) because Python over-allocates extra capacity when resizing; occasionally an append triggers a full resize (O(n)) to copy elements into a bigger block, but averaged over many appends, the cost per append is still O(1).

## 12. Revision Summary

- Python has no native fixed-type array - `list` is the general-purpose dynamic array
- Three "array" options: `list` (flexible), `array` module (typed, rare), `np.ndarray` (numeric powerhouse)
- Indexing is O(1) for all three; insert/delete in the middle or front is O(n)
- Lists store pointers to objects; `array`/NumPy store raw values contiguously - this is why NumPy is memory-efficient and fast
- NumPy uses vectorization (compiled C loops) to avoid Python's slow per-element interpreter loop
- Never repeatedly `np.append()` inside a loop - build a list, convert once

## Interview Q&A Bank (60 Questions)

### Easy (1-20)

1. **Q: What is an array?**
   A: A collection of elements stored in contiguous memory, typically of the same data type, accessed via an index.
2. **Q: What is the Python equivalent of an array?**
   A: The built-in `list` is the most common equivalent; it's a dynamic array that can also hold mixed types.
3. **Q: How do you access the third element of a list `arr`?**
   A: `arr[2]` - indexing starts at 0.
4. **Q: How do you find the length of a list?**
   A: Using the built-in `len(arr)` function.
5. **Q: How do you add an element to the end of a list?**
   A: `arr.append(value)`.
6. **Q: How do you remove the last element of a list?**
   A: `arr.pop()` - removes and returns the last element by default.
7. **Q: Can a Python list hold different data types at once?**
   A: Yes, e.g. `[1, "two", 3.0, True]` is valid because list elements are references to objects, not raw typed values.
8. **Q: What does negative indexing mean, e.g. `arr[-1]`?**
   A: It accesses elements from the end; `-1` is the last element, `-2` the second-last, and so on.
9. **Q: How do you check if a value exists in a list?**
   A: Using the `in` operator, e.g. `5 in arr`.
10. **Q: What module provides a real typed array in Python?**
    A: The built-in `array` module (`from array import array`).
11. **Q: What is a typecode in the `array` module?**
    A: A single character (e.g. `'i'` for int, `'d'` for double/float) that specifies what type of data the array will hold.
12. **Q: How do you create an empty list?**
    A: `arr = []` or `arr = list()`.
13. **Q: What is the index of the first element in a Python list?**
    A: 0 (zero-based indexing).
14. **Q: How do you reverse a list?**
    A: `arr[::-1]` (slicing) or `arr.reverse()` (in-place).
15. **Q: How do you sort a list in ascending order?**
    A: `arr.sort()` (in-place) or `sorted(arr)` (returns a new sorted list).
16. **Q: What happens if you access an index that doesn't exist, e.g. `arr[100]` on a 5-element list?**
    A: Python raises an `IndexError`.
17. **Q: How do you combine (concatenate) two lists?**
    A: Using `+`, e.g. `list1 + list2`, or `list1.extend(list2)` to modify in place.
18. **Q: What is the difference between `append()` and `extend()`?**
    A: `append()` adds a single item (even if it's a list) as one element; `extend()` adds each item of an iterable individually.
19. **Q: Is a Python list mutable or immutable?**
    A: Mutable - its contents can be changed after creation.
20. **Q: How do you copy a list without affecting the original?**
    A: Using `arr.copy()`, `list(arr)`, or slicing `arr[:]` for a shallow copy.

### Medium (21-40)

21. **Q: What is the time complexity of `list.append()`?**
    A: O(1) amortized - most calls are constant time, with occasional O(n) resizes averaged out.
22. **Q: What is the time complexity of inserting an element at index 0 in a list?**
    A: O(n), because every existing element must shift one position to the right.
23. **Q: What is the time complexity of searching for a value using `in` on a list?**
    A: O(n) - Python performs a linear scan through the list.
24. **Q: How would you remove duplicate elements from a list while preserving order?**
    A: Iterate through the list, add elements to a result list only if not already seen (tracked via a set for O(1) lookups), e.g. using `dict.fromkeys(arr)` which preserves insertion order in modern Python.
25. **Q: What is the difference between `list.sort()` and `sorted(list)`?**
    A: `.sort()` sorts in place and returns `None`; `sorted()` returns a new sorted list, leaving the original unchanged.
26. **Q: How is a Python list implemented internally?**
    A: As a dynamic array of pointers to Python objects, with over-allocated capacity so that appends are usually O(1).
27. **Q: What is the difference between a shallow copy and using the original list directly (no copy)?**
    A: Without copying, both variables point to the same list object - changes through one affect the other. A shallow copy creates a new outer list, but nested mutable objects inside are still shared.
28. **Q: Why does `array.array` use less memory than a `list` of the same numbers?**
    A: Because `array` stores raw primitive values contiguously, while `list` stores pointers to separate Python integer/float objects that carry their own object overhead.
29. **Q: What is list slicing, and what does `arr[1:4]` return?**
    A: Slicing extracts a sub-list; `arr[1:4]` returns elements at index 1, 2, 3 (stop index is exclusive).
30. **Q: How do you flatten a nested list, e.g. `[[1,2],[3,4]]` into `[1,2,3,4]`?**
    A: Using a list comprehension: `[item for sublist in nested for item in sublist]`.
31. **Q: What does `list(range(5))` produce and why is it useful?**
    A: `[0, 1, 2, 3, 4]` - `range()` is a memory-efficient sequence generator; converting to a list materializes all its values.
32. **Q: How would you find the maximum and minimum of a list without using `max()`/`min()`?**
    A: Iterate through the list, tracking the current largest and smallest value seen so far, updating them as needed - O(n) time.
33. **Q: What is the difference between `list.remove(value)` and `list.pop(index)`?**
    A: `remove()` deletes the first occurrence of a given value; `pop()` deletes and returns the element at a given index (or the last one by default).
34. **Q: Can you multiply a list, e.g. `[1,2] * 3`? What does it produce?**
    A: Yes, it produces `[1, 2, 1, 2, 1, 2]` - repetition, not element-wise multiplication (that requires NumPy).
35. **Q: What is a 2D list (list of lists) used for?**
    A: Representing grid/matrix-like data, e.g. a table or board, where each inner list is a row.
36. **Q: How do you create a 2D list of size m x n filled with zeros correctly?**
    A: `[[0] * n for _ in range(m)]` - using a list comprehension avoids all rows referencing the same inner list object (a common bug with `[[0]*n]*m`).
37. **Q: What bug occurs with `[[0]*n]*m` for creating a 2D list, and why?**
    A: All `m` rows end up being references to the exact same inner list object, so modifying one row modifies all of them - because `*` on a list of lists copies references, not the inner lists themselves.
38. **Q: What is the space complexity of storing an array/list of n elements?**
    A: O(n) - space grows linearly with the number of elements.
39. **Q: How do you convert a list to a NumPy array and why would you do it?**
    A: `np.array(my_list)` - done to enable fast vectorized numeric operations instead of manual Python loops.
40. **Q: What does `arr.index(value)` do, and what happens if the value isn't found?**
    A: Returns the index of the first occurrence of `value`; raises a `ValueError` if the value is not present.

### Hard (41-60)

41. **Q: Explain why list `.append()` is amortized O(1) rather than always exactly O(1).**
    A: Python lists over-allocate extra capacity when they resize. Most appends just fill existing spare capacity (O(1)). When capacity runs out, Python allocates a new, larger block and copies all existing elements (O(n)) - but because this happens infrequently and capacity grows geometrically, the average cost per append across many operations is still O(1).
42. **Q: Given a rotated sorted array, how would you search for a target value efficiently?**
    A: Use a modified binary search - at each step, determine which half of the array is properly sorted, check if the target lies within that half's range, and recurse/iterate into the correct half; this achieves O(log n) instead of a linear scan.
43. **Q: How would you find all pairs in an array that sum to a target value, in less than O(n^2) time?**
    A: Use a hash set - iterate through the array once, for each element check if `(target - element)` is already in the set; if yes, a pair is found; otherwise add the current element to the set. This achieves O(n) time using O(n) space.
44. **Q: What is the difference in memory layout between a Python list and a NumPy array, at a low level?**
    A: A Python list stores an array of pointers, each pointing to a separately allocated Python object (with type info, reference count, and value) scattered in memory. A NumPy array stores raw values of one fixed type in a single contiguous memory block with no per-element object overhead, enabling fast cache-friendly access and vectorized operations.
45. **Q: How would you rotate an array by k positions in-place with O(1) extra space?**
    A: Use the "reversal algorithm" - reverse the entire array, then reverse the first k elements, then reverse the remaining n-k elements; this achieves the rotation in O(n) time and O(1) extra space.
46. **Q: What is the Kadane's Algorithm used for, and what is its time complexity?**
    A: Finding the maximum sum of a contiguous subarray; it runs in O(n) time by tracking a running "current max sum ending here" and a global maximum, resetting the running sum to zero whenever it goes negative.
47. **Q: Explain the difference between `array.array` growth strategy and NumPy array growth when appending.**
    A: `array.array` (like `list`) grows dynamically with over-allocation, so appends are amortized O(1). NumPy arrays are fixed-size once created - `np.append()` always allocates an entirely new array and copies all data over, making it O(n) on every single call, which is a common performance trap in loops.
48. **Q: How would you find the missing number in an array containing n distinct numbers from 0 to n?**
    A: Compute the expected sum of 0 to n using the formula `n*(n+1)/2`, subtract the actual sum of the array elements - the difference is the missing number. O(n) time, O(1) space.
49. **Q: What does "cache locality" mean, and why do NumPy arrays benefit from it more than Python lists?**
    A: Cache locality refers to how close in memory related data is stored - CPUs load memory in blocks (cache lines), so contiguous data loads faster on repeated access. NumPy arrays store values contiguously, so iterating over them benefits heavily from cache locality; Python lists store scattered pointers to objects, which causes more cache misses.
50. **Q: How would you find the longest increasing subsequence conceptually (no need for full DP code)?**
    A: For each element, track the length of the longest increasing subsequence ending at that index by checking all previous smaller elements; this naive approach is O(n^2), though an O(n log n) approach exists using binary search with a patience-sorting-like technique.
51. **Q: What's the difference between `is` and `==` when comparing two lists with the same contents?**
    A: `==` checks if the contents are equal element-by-element (returns True for two separately created lists with the same values), while `is` checks whether they are the exact same object in memory (returns False unless they're literally the same reference).
52. **Q: How would you merge two sorted arrays into a single sorted array in O(n+m) time?**
    A: Use two pointers, one for each array; compare the current elements pointed to, append the smaller to the result and advance that pointer, repeating until one array is exhausted, then append the remainder of the other array.
53. **Q: What is the Dutch National Flag problem, and how does it relate to arrays?**
    A: It's the problem of sorting an array containing only three distinct values (e.g. 0s, 1s, 2s) in a single O(n) pass using three pointers (low, mid, high), avoiding a full comparison sort.
54. **Q: Why can inserting into the middle of a NumPy array not be done in-place efficiently?**
    A: NumPy arrays are fixed-size contiguous memory blocks; inserting requires creating a new array of the new size and copying all elements, shifting everything after the insertion point - this is O(n) and NumPy's `np.insert()` always returns a new array rather than mutating in place.
55. **Q: How would you detect if an array contains a duplicate value in O(n) time and O(n) space?**
    A: Iterate through the array while adding each element to a set; if an element is already present in the set before adding, a duplicate has been found.
56. **Q: What is the two-pointer technique, and give an example of an array problem it solves.**
    A: Using two index pointers (often from opposite ends or moving at different speeds) to traverse an array in a single pass instead of nested loops - e.g. finding a pair that sums to a target in a sorted array by moving a left pointer up and a right pointer down based on comparison to the target, achieving O(n) instead of O(n^2).
57. **Q: What is the sliding window technique used for with arrays?**
    A: Efficiently computing something over every contiguous subarray of a fixed or variable size by maintaining a running window and updating it incrementally (adding one new element, removing one old element) instead of recomputing from scratch, reducing time complexity from O(n*k) to O(n).
58. **Q: Why is `list.pop(0)` considered a performance anti-pattern in a loop?**
    A: Removing the first element requires shifting every remaining element left by one position, making each call O(n); doing this repeatedly in a loop of n iterations produces O(n^2) total time - a `collections.deque` (O(1) pops from both ends) should be used instead for such cases.
59. **Q: How would you find the intersection of two large arrays efficiently?**
    A: Convert one array to a set (O(n) build time), then iterate through the second array checking membership in that set (O(1) average per check) - overall O(n+m) time instead of O(n*m) with nested loops.
60. **Q: Explain why NumPy's broadcasting mechanism allows operations between arrays of different shapes, and why this matters for performance.**
    A: Broadcasting lets NumPy implicitly expand smaller arrays across a larger array's shape (following specific compatibility rules) without physically copying data, so operations like adding a scalar to a whole array or a 1D array to each row of a 2D array run in a single vectorized C loop instead of requiring explicit Python loops or manual replication of data - covered in full depth on Day 3.
