## 1. Introduction

- These are Python's four core built-in **data structures** used to store collections of values
- **List** - ordered, changeable collection (allows duplicates)
- **Tuple** - ordered, unchangeable collection (allows duplicates)
- **Set** - unordered collection of unique values (no duplicates)
- **Dictionary** - collection of key-value pairs (like a real-world lookup table)
- Why needed?
  - Almost no real dataset is a single number - data always comes as collections (rows, columns, records)
  - Choosing the right structure impacts both correctness and performance
- Where used?
  - A Pandas DataFrame column is essentially built on these ideas
  - JSON API responses map directly to nested dicts/lists
  - Removing duplicates, fast lookups, and grouping data all rely on choosing the right structure

## 2. Real-Life Analogy

- **List** = a shopping list on paper - ordered, you can add/remove/edit items, duplicates allowed ("milk" can appear twice)
- **Tuple** = a printed receipt - ordered, but once printed, it cannot be changed
- **Set** = a guest list at a private party - no duplicate names allowed, and the order they're called doesn't matter
- **Dictionary** = a phone contact book - you look up a person's number by their **name** (key), not by position

## 3. Explanation

- **List**
  - Mutable (changeable), ordered, allows duplicates
  - Created with `[]`
- **Tuple**
  - Immutable (cannot change after creation), ordered, allows duplicates
  - Created with `()`
  - Faster than lists and safer when data shouldn't change (e.g. coordinates, fixed records)
- **Set**
  - Mutable, unordered, no duplicates allowed
  - Created with `{}` or `set()`
  - Best for fast membership checks and removing duplicates
- **Dictionary**
  - Mutable, ordered (insertion order preserved since Python 3.7), key-value pairs, keys must be unique
  - Created with `{key: value}`

> **Trick to remember:** "LTSD" - List (changeable), Tuple (frozen), Set (unique), Dict (labeled) - think of it as increasing structure/restriction from List to Dict.

## 4. Syntax

```python
# List
my_list = [1, 2, 3, "a"]
my_list.append(4)
my_list[0]              # indexing
my_list[0:2]             # slicing

# Tuple
my_tuple = (1, 2, 3)
my_tuple[0]              # indexing works, but no append/remove

# Set
my_set = {1, 2, 3}
my_set.add(4)
my_set.remove(2)

# Dictionary
my_dict = {"name": "Alice", "age": 25}
my_dict["name"]           # access by key
my_dict["city"] = "Pune"  # add new key
my_dict.get("age")        # safe access (no error if key missing)
```

````
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)
my_set = {1, 2, 3}
my_dict = {"a": 1, "b": 2}

print(my_list, my_tuple, my_set, my_dict)
````

## 5. Examples

### Basic Example

````
# Basic: creating and accessing each structure
fruits_list = ["apple", "banana", "cherry"]
coordinates_tuple = (10.5, 20.3)
unique_ids_set = {101, 102, 103}
person_dict = {"name": "Ravi", "age": 30}

print(fruits_list[1])          # banana - index access
print(coordinates_tuple[0])    # 10.5
print(103 in unique_ids_set)   # True - membership check
print(person_dict["name"])     # Ravi - key access
````

### Intermediate Example

````
# Intermediate: modifying list/set/dict (tuple is immutable, skipped)
scores = [55, 70, 40, 90]
scores.append(85)
scores.remove(40)
print("Updated list:", scores)

departments = {"Sales", "HR", "Sales", "Engineering"}  # duplicate "Sales" auto-removed
print("Unique departments:", departments)

employee = {"name": "Meena", "role": "Analyst"}
employee["role"] = "Senior Analyst"   # update existing key
employee["team"] = "Data"             # add new key
print("Updated dict:", employee)
````

- `scores.append(85)` adds to the end; `scores.remove(40)` removes the first matching value
- The set automatically dropped the duplicate `"Sales"` - a built-in deduplication feature
- Dictionary keys are updated/added the same way (`dict[key] = value`)

### Real-World Example

````
# Real-world: representing one row of a dataset as a dict, and a dataset as a list of dicts
# This is EXACTLY how JSON API data / MongoDB documents look before becoming a DataFrame

dataset = [
    {"emp_id": 101, "name": "Asha", "dept": "Sales", "salary": 52000},
    {"emp_id": 102, "name": "Vikram", "dept": "Engineering", "salary": 68000},
    {"emp_id": 103, "name": "Neha", "dept": "Sales", "salary": 49000},
]

# Get unique departments using a set + list comprehension idea (basic loop version here)
departments = set()
for record in dataset:
    departments.add(record["dept"])

print("Departments:", departments)

# Find total salary for Sales department
sales_total = 0
for record in dataset:
    if record["dept"] == "Sales":
        sales_total += record["salary"]

print("Total Sales salary:", sales_total)
````

- A "list of dictionaries" is the most common raw shape of real-world data before it becomes a Pandas DataFrame (`pd.DataFrame(dataset)` would convert this directly)
- Using a `set` to collect `departments` automatically avoids duplicates without extra logic
- This exact filter-and-sum pattern is the manual version of what `groupby().sum()` does in Pandas later in the course

## 6. Internal Working

- **List**: implemented as a dynamic array - contiguous memory, resizes automatically when it grows beyond capacity
- **Tuple**: also a fixed-size array internally, but since it can't change size, Python can store it more compactly and access it slightly faster than a list
- **Set**: implemented as a **hash table** - each element's hash determines where it's stored, enabling O(1) average lookup/membership check
- **Dictionary**: also a **hash table**, where each key is hashed to determine its storage location; the value is stored alongside it

> **Trick to remember:** "Set and Dict are hash-powered, List and Tuple are array-powered." This single fact explains almost every complexity/interview question about these structures.

## 7. Time and Space Complexity

- **List**
  - Access by index: O(1)
  - Search (`in`): O(n)
  - Append: O(1) amortized; Insert/Delete at start/middle: O(n)
- **Tuple**
  - Access by index: O(1)
  - Search (`in`): O(n)
  - (No insert/delete - immutable)
- **Set**
  - Add/Remove/Search (`in`): O(1) average (hash-based)
- **Dictionary**
  - Access/Add/Remove by key: O(1) average (hash-based)
- Space: all four are O(n) where n = number of elements

> **Interview line:** If I need fast lookups, I reach for a set or dict, not a list - that's an O(n) to O(1) upgrade.

## 8. Common Mistakes

- Trying to modify a tuple (`my_tuple[0] = 5`) -> `TypeError`, tuples are immutable
- Assuming sets preserve order or allow indexing (`my_set[0]`) -> `TypeError`, sets are unordered and unindexed
- Accessing a missing dictionary key directly (`my_dict["missing"]`) -> `KeyError` - use `.get()` instead for safety
- Using a list when a set would be better for membership checks, causing unnecessary O(n) scans in loops
- Confusing `{}` (empty dict) with an empty set - `{}` always creates a dict, use `set()` for an empty set

````
my_dict = {"name": "Kiran"}

# Safe access - won't crash if key is missing
print(my_dict.get("age"))          # None
print(my_dict.get("age", "N/A"))   # N/A - default value if missing

# {} is a dict, not a set!
print(type({}))       # <class 'dict'>
print(type(set()))    # <class 'set'>
````

## 9. Best Practices

- Use a **tuple** for fixed data that shouldn't change (e.g. coordinates, RGB values, function returns with multiple values)
- Use a **set** whenever you need uniqueness or fast membership checks
- Use `.get()` instead of `dict[key]` when the key might not exist, to avoid crashes
- Use a **list of dicts** as the natural intermediate structure before converting to a Pandas DataFrame
- Prefer dictionaries over parallel lists (e.g. separate `names` list and `ages` list) - keeping related data together avoids index-mismatch bugs

## 10. Interview Questions

**Beginner**
- Q: What is the main difference between a list and a tuple?
  A: Lists are mutable (can be changed after creation); tuples are immutable (cannot be changed once created). Both are ordered and allow duplicates.
- Q: Can a dictionary have duplicate keys?
  A: No - keys must be unique. If you assign to an existing key, it overwrites the previous value rather than creating a duplicate.

**Intermediate**
- Q: Why is membership checking (`in`) faster in a set than in a list?
  A: A set is backed by a hash table, giving O(1) average lookup time, while a list requires a linear O(n) scan through its elements.
- Q: When would you use a tuple instead of a list?
  A: When the data shouldn't change after creation (e.g. fixed records, function return values, dictionary keys since tuples are hashable and lists are not).

**Advanced**
- Q: Why can a tuple be used as a dictionary key but a list cannot?
  A: Dictionary keys must be hashable (i.e. have a fixed hash value that doesn't change). Tuples are immutable so they're hashable; lists are mutable, so their contents (and thus hash) could change, which is disallowed.
- Q: How does Python's dictionary maintain insertion order since 3.7, given it's hash-based?
  A: CPython's dict implementation internally keeps a separate compact array that tracks insertion order alongside the hash table used for key lookups, so iteration follows insertion order while still preserving O(1) average lookup.

## 12. Revision Summary

- List: ordered, mutable, duplicates allowed - `[]`
- Tuple: ordered, immutable, duplicates allowed - `()`
- Set: unordered, mutable, unique values only - `{}` or `set()`
- Dict: key-value pairs, unique keys, insertion order preserved - `{key: value}`
- Set and Dict are hash-based -> O(1) average lookup; List and Tuple are array-based -> O(n) search
- Use `.get()` on dicts to avoid `KeyError` crashes
- Tuples are hashable (can be dict keys), lists are not
- "List of dicts" is the natural bridge structure between raw/JSON data and a Pandas DataFrame
