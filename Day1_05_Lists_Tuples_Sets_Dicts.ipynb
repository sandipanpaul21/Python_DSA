## 1. Introduction

- These are Python's four core built-in **data structures** used to store collections of values
- **List** - ordered, changeable collection (allows duplicates)
- **Tuple** - ordered, unchangeable collection (allows duplicates)
- **Set** - unordered collection of unique values (no duplicates)
- **Dictionary** - collection of key-value pairs (like a real-world lookup table)
- Why needed?
  - Almost no real dataset is a single number - data always comes as collections (rows, columns, records)
  - Choosing the right structure impacts both correctness and performance
- Where used?
  - A Pandas DataFrame column is essentially built on these ideas
  - JSON API responses map directly to nested dicts/lists
  - Removing duplicates, fast lookups, and grouping data all rely on choosing the right structure

## 2. Real-Life Analogy

- **List** = a shopping list on paper - ordered, you can add/remove/edit items, duplicates allowed ("milk" can appear twice)
- **Tuple** = a printed receipt - ordered, but once printed, it cannot be changed
- **Set** = a guest list at a private party - no duplicate names allowed, and the order they're called doesn't matter
- **Dictionary** = a phone contact book - you look up a person's number by their **name** (key), not by position

## 3. Explanation

- **List**
  - Mutable (changeable), ordered, allows duplicates
  - Created with `[]`
- **Tuple**
  - Immutable (cannot change after creation), ordered, allows duplicates
  - Created with `()`
  - Faster than lists and safer when data shouldn't change (e.g. coordinates, fixed records)
- **Set**
  - Mutable, unordered, no duplicates allowed
  - Created with `{}` or `set()`
  - Best for fast membership checks and removing duplicates
- **Dictionary**
  - Mutable, ordered (insertion order preserved since Python 3.7), key-value pairs, keys must be unique
  - Created with `{key: value}`

> **Trick to remember:** "LTSD" - List (changeable), Tuple (frozen), Set (unique), Dict (labeled) - think of it as increasing structure/restriction from List to Dict.

### List Methods (Complete Set)

**Adding elements**
- `append(x)` - add ONE item at the end
- `extend(iterable)` - add ALL items from another iterable at the end
- `insert(index, x)` - add item at a specific position

**Removing elements**
- `remove(x)` - remove FIRST occurrence of a value (error if not found)
- `pop(index)` - remove and RETURN item at index (default: last item)
- `clear()` - empty the whole list

**Searching / counting**
- `index(x)` - position of first occurrence (error if not found)
- `count(x)` - how many times a value appears

**Ordering**
- `sort()` - sort IN PLACE (modifies the list, returns None)
- `sorted(list)` - built-in, returns a NEW sorted list (original untouched)
- `reverse()` - reverse IN PLACE
- `copy()` - shallow copy of the list

> **Trick to remember:** append = one guest, extend = whole family. pop = remove AND take away, remove = just kick out. sort() changes me, sorted() gives a copy.

````python
nums = [3, 1, 4, 1, 5]

# Adding
nums.append(9)              # [3, 1, 4, 1, 5, 9]
nums.extend([2, 6])         # [3, 1, 4, 1, 5, 9, 2, 6]
nums.insert(0, 10)          # [10, 3, 1, 4, 1, 5, 9, 2, 6]
print("After adding:   ", nums)

# Removing
nums.remove(1)              # removes FIRST 1 only
last = nums.pop()           # removes and returns 6
second = nums.pop(1)        # removes and returns item at index 1
print("After removing: ", nums, "| popped:", last, second)

# Searching / counting
print("index of 4:", nums.index(4))
print("count of 1:", nums.count(1))

# Ordering
nums.sort()                          # in place, ascending
print("sorted asc:    ", nums)
nums.sort(reverse=True)              # in place, descending
print("sorted desc:   ", nums)
nums.reverse()                       # just flips current order
print("reversed:      ", nums)

fresh = sorted(nums)                 # NEW list, original untouched
print("sorted() copy: ", fresh, "| original:", nums)

backup = nums.copy()                 # shallow copy
backup.append(999)
print("copy modified: ", backup, "| original safe:", nums)
````

**Useful built-ins that work on lists (not methods, but interview staples)**
- `len(lst)` - number of items
- `max(lst)` / `min(lst)` - largest / smallest
- `sum(lst)` - total of numeric list
- `any(lst)` - True if AT LEAST ONE item is truthy
- `all(lst)` - True if EVERY item is truthy
- `list(reversed(lst))` - reversed COPY (vs .reverse() in place)
- `enumerate(lst)` - index + value pairs
- `zip(lst1, lst2)` - pair up two lists element-wise

````python
scores = [45, 88, 62, 91]
names = ["Asha", "Vikram", "Neha", "Ravi"]

print(len(scores), max(scores), min(scores), sum(scores))
print(any(s > 90 for s in scores))     # is there any topper?
print(all(s >= 40 for s in scores))    # did everyone pass?

for name, score in zip(names, scores):
    print(name, "->", score)
````

### Tuple Methods (Complete Set)

- Tuples are immutable, so they have ONLY 2 methods:
- `count(x)` - how many times a value appears
- `index(x)` - position of first occurrence
- Everything read-only works: indexing, slicing, `len()`, `min()`, `max()`, `sum()`, `in`
- Special tuple superpower: **unpacking** (including `*` rest-unpacking)

> **Trick to remember:** Tuple = list with hands tied - it can only LOOK (count/index), never TOUCH (no append/remove/sort).

````python
point = (3, 7, 3, 9, 3)

print(point.count(3))     # 3 appears three times
print(point.index(9))     # position of 9

# Read-only operations all work
print(len(point), max(point), sum(point))
print(7 in point)

# Unpacking - THE tuple skill
x, y, *rest = point
print(x, y, rest)          # 3 7 [3, 9, 3]

# Swapping variables - classic tuple trick
a, b = 1, 2
a, b = b, a
print(a, b)                # 2 1
````

### Set Methods (Complete Set)

**Adding / removing**
- `add(x)` - add one element
- `update(iterable)` - add many elements
- `remove(x)` - remove element (ERROR if missing)
- `discard(x)` - remove element (SILENT if missing)
- `pop()` - remove and return an arbitrary element
- `clear()` - empty the set

**Set mathematics (the interview favorites)**
- `union(other)` or `|` - all elements from both
- `intersection(other)` or `&` - only common elements
- `difference(other)` or `-` - in me but NOT in other
- `symmetric_difference(other)` or `^` - in either, but NOT both

**Relationship checks**
- `issubset(other)` or `<=` - am I fully inside other?
- `issuperset(other)` or `>=` - does other fit fully inside me?
- `isdisjoint(other)` - do we share NOTHING?

> **Trick to remember:** remove = strict bouncer (complains if person absent), discard = chill bouncer (shrugs). Union | = OR, Intersection & = AND, Difference - = subtract, ^ = XOR (exclusive).

````python
team_python = {"Asha", "Vikram", "Neha"}
team_sql    = {"Neha", "Ravi", "Meena"}

# Adding / removing
team_python.add("Kiran")
team_python.discard("Ghost")          # no error even though absent
print("Python team:", team_python)

# Set math - who knows what?
print("Either skill:  ", team_python | team_sql)     # union
print("Both skills:   ", team_python & team_sql)     # intersection
print("Only Python:   ", team_python - team_sql)     # difference
print("Exactly one:   ", team_python ^ team_sql)     # symmetric difference

# Relationship checks
juniors = {"Asha", "Vikram"}
print("Juniors all in Python team?", juniors.issubset(team_python))
print("Teams share nobody?", team_python.isdisjoint(team_sql))
````

**Real-world set-math pattern (EDA):** comparing customer lists across two months
- `march & april` - retained customers
- `march - april` - churned customers
- `april - march` - new customers
- This exact pattern appears in cohort/churn analysis constantly

### Dictionary Methods (Complete Set)

**Reading**
- `get(key, default)` - safe read, returns default (None) if key missing - NEVER crashes
- `keys()` - view of all keys
- `values()` - view of all values
- `items()` - view of (key, value) pairs - THE way to loop a dict

**Writing / removing**
- `update(other_dict)` - merge another dict in (overwrites matching keys)
- `setdefault(key, default)` - return value if key exists, else INSERT default and return it
- `pop(key, default)` - remove key and return its value
- `popitem()` - remove and return the LAST inserted (key, value) pair
- `clear()` - empty the dict
- `copy()` - shallow copy

**Class-level helper**
- `dict.fromkeys(keys, value)` - build a dict giving every key the same starting value

> **Trick to remember:** get = safe read, setdefault = safe write. items() for looping, update() for merging. Since Python 3.9, `d1 | d2` merges dicts just like sets union.

````python
emp = {"name": "Asha", "dept": "Sales", "salary": 52000}

# Reading
print(emp.get("salary"))
print(emp.get("bonus", 0))            # default instead of KeyError
print(list(emp.keys()))
print(list(emp.values()))

# THE looping pattern
for key, value in emp.items():
    print(f"  {key}: {value}")

# Writing / removing
emp.update({"salary": 55000, "city": "Pune"})   # overwrite + add
removed = emp.pop("dept")
print("popped:", removed, "| now:", emp)

# setdefault - read-or-create in one step
emp.setdefault("skills", []).append("Python")
emp.setdefault("skills", []).append("SQL")      # key exists now - just appends
print("skills:", emp["skills"])

# fromkeys - initialize many keys at once
tracker = dict.fromkeys(["mon", "tue", "wed"], 0)
print("tracker:", tracker)

# Dict merge operator (Python 3.9+)
defaults = {"theme": "light", "lang": "en"}
user_prefs = {"lang": "hi"}
final = defaults | user_prefs          # right side wins conflicts
print("merged:", final)
````

**The grouping pattern - most important dict idiom in Data Science interviews**
- Problem: group records by a field (this is what SQL GROUP BY / Pandas groupby do)
- Tool: `setdefault` (or `collections.defaultdict`)

````python
records = [
    {"name": "Asha", "dept": "Sales"},
    {"name": "Vikram", "dept": "Engineering"},
    {"name": "Neha", "dept": "Sales"},
    {"name": "Ravi", "dept": "Engineering"},
]

# Group names by department - THE interview pattern
by_dept = {}
for r in records:
    by_dept.setdefault(r["dept"], []).append(r["name"])

print(by_dept)
# {'Sales': ['Asha', 'Neha'], 'Engineering': ['Vikram', 'Ravi']}

# Counting pattern - second most common dict idiom
counts = {}
for r in records:
    counts[r["dept"]] = counts.get(r["dept"], 0) + 1
print(counts)
````

> **Interview tip:** These two idioms (grouping via setdefault, counting via get) solve a huge share of dict-based interview problems. `collections.defaultdict(list)` and `collections.Counter` are their professional upgrades - mention them for bonus points.

## 5. Examples

### Basic Example

````python
# Basic: creating and accessing each structure
fruits_list = ["apple", "banana", "cherry"]
coordinates_tuple = (10.5, 20.3)
unique_ids_set = {101, 102, 103}
person_dict = {"name": "Ravi", "age": 30}

print(fruits_list[1])          # banana - index access
print(coordinates_tuple[0])    # 10.5
print(103 in unique_ids_set)   # True - membership check
print(person_dict["name"])     # Ravi - key access
````

### Intermediate Example

````python
# Intermediate: modifying list/set/dict (tuple is immutable, skipped)
scores = [55, 70, 40, 90]
scores.append(85)
scores.remove(40)
print("Updated list:", scores)

departments = {"Sales", "HR", "Sales", "Engineering"}  # duplicate "Sales" auto-removed
print("Unique departments:", departments)

employee = {"name": "Meena", "role": "Analyst"}
employee["role"] = "Senior Analyst"   # update existing key
employee["team"] = "Data"             # add new key
print("Updated dict:", employee)
````

- `scores.append(85)` adds to the end; `scores.remove(40)` removes the first matching value
- The set automatically dropped the duplicate `"Sales"` - a built-in deduplication feature
- Dictionary keys are updated/added the same way (`dict[key] = value`)

### Real-World Example

````python
# Real-world: representing one row of a dataset as a dict, and a dataset as a list of dicts
# This is EXACTLY how JSON API data / MongoDB documents look before becoming a DataFrame

dataset = [
    {"emp_id": 101, "name": "Asha", "dept": "Sales", "salary": 52000},
    {"emp_id": 102, "name": "Vikram", "dept": "Engineering", "salary": 68000},
    {"emp_id": 103, "name": "Neha", "dept": "Sales", "salary": 49000},
]

# Get unique departments using a set + list comprehension idea (basic loop version here)
departments = set()
for record in dataset:
    departments.add(record["dept"])

print("Departments:", departments)

# Find total salary for Sales department
sales_total = 0
for record in dataset:
    if record["dept"] == "Sales":
        sales_total += record["salary"]

print("Total Sales salary:", sales_total)
````

- A "list of dictionaries" is the most common raw shape of real-world data before it becomes a Pandas DataFrame (`pd.DataFrame(dataset)` would convert this directly)
- Using a `set` to collect `departments` automatically avoids duplicates without extra logic
- This exact filter-and-sum pattern is the manual version of what `groupby().sum()` does in Pandas later in the course

## 6. Internal Working

- **List**: implemented as a dynamic array - contiguous memory, resizes automatically when it grows beyond capacity
- **Tuple**: also a fixed-size array internally, but since it can't change size, Python can store it more compactly and access it slightly faster than a list
- **Set**: implemented as a **hash table** - each element's hash determines where it's stored, enabling O(1) average lookup/membership check
- **Dictionary**: also a **hash table**, where each key is hashed to determine its storage location; the value is stored alongside it

> **Trick to remember:** "Set and Dict are hash-powered, List and Tuple are array-powered." This single fact explains almost every complexity/interview question about these structures.

## 7. Time and Space Complexity

- **List**
  - Access by index: O(1)
  - Search (`in`): O(n)
  - Append: O(1) amortized; Insert/Delete at start/middle: O(n)
- **Tuple**
  - Access by index: O(1)
  - Search (`in`): O(n)
  - (No insert/delete - immutable)
- **Set**
  - Add/Remove/Search (`in`): O(1) average (hash-based)
- **Dictionary**
  - Access/Add/Remove by key: O(1) average (hash-based)
- Space: all four are O(n) where n = number of elements

> **Interview line:** If I need fast lookups, I reach for a set or dict, not a list - that's an O(n) to O(1) upgrade.

## 8. Common Mistakes

- Trying to modify a tuple (`my_tuple[0] = 5`) -> `TypeError`, tuples are immutable
- Assuming sets preserve order or allow indexing (`my_set[0]`) -> `TypeError`, sets are unordered and unindexed
- Accessing a missing dictionary key directly (`my_dict["missing"]`) -> `KeyError` - use `.get()` instead for safety
- Using a list when a set would be better for membership checks, causing unnecessary O(n) scans in loops
- Confusing `{}` (empty dict) with an empty set - `{}` always creates a dict, use `set()` for an empty set

````python
my_dict = {"name": "Kiran"}

# Safe access - won't crash if key is missing
print(my_dict.get("age"))          # None
print(my_dict.get("age", "N/A"))   # N/A - default value if missing

# {} is a dict, not a set!
print(type({}))       # <class 'dict'>
print(type(set()))    # <class 'set'>
````

## 9. Best Practices

- Use a **tuple** for fixed data that shouldn't change (e.g. coordinates, RGB values, function returns with multiple values)
- Use a **set** whenever you need uniqueness or fast membership checks
- Use `.get()` instead of `dict[key]` when the key might not exist, to avoid crashes
- Use a **list of dicts** as the natural intermediate structure before converting to a Pandas DataFrame
- Prefer dictionaries over parallel lists (e.g. separate `names` list and `ages` list) - keeping related data together avoids index-mismatch bugs

## 10. Interview Questions

**Beginner**
- Q: What is the main difference between a list and a tuple?
  A: Lists are mutable (can be changed after creation); tuples are immutable (cannot be changed once created). Both are ordered and allow duplicates.
- Q: Can a dictionary have duplicate keys?
  A: No - keys must be unique. If you assign to an existing key, it overwrites the previous value rather than creating a duplicate.

**Intermediate**
- Q: Why is membership checking (`in`) faster in a set than in a list?
  A: A set is backed by a hash table, giving O(1) average lookup time, while a list requires a linear O(n) scan through its elements.
- Q: When would you use a tuple instead of a list?
  A: When the data shouldn't change after creation (e.g. fixed records, function return values, dictionary keys since tuples are hashable and lists are not).

**Advanced**
- Q: Why can a tuple be used as a dictionary key but a list cannot?
  A: Dictionary keys must be hashable (i.e. have a fixed hash value that doesn't change). Tuples are immutable so they're hashable; lists are mutable, so their contents (and thus hash) could change, which is disallowed.
- Q: How does Python's dictionary maintain insertion order since 3.7, given it's hash-based?
  A: CPython's dict implementation internally keeps a separate compact array that tracks insertion order alongside the hash table used for key lookups, so iteration follows insertion order while still preserving O(1) average lookup.

## 12. Revision Summary

- List: ordered, mutable, duplicates allowed - `[]`
- Tuple: ordered, immutable, duplicates allowed - `()`
- Set: unordered, mutable, unique values only - `{}` or `set()`
