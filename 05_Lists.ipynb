# Python Lists and List Comprehension

## 1. Introduction

- **What is it?**
  - A **list** is a built-in Python data type that stores multiple items in a single variable, in order.
- **Definition**
  - A list is an **ordered, mutable (changeable) collection** of items written inside square brackets `[]`, separated by commas.
- **Why do we need it?**
  - Without a list, you would need a separate variable for every value (`name1`, `name2`, `name3`...). Lists let you group related data together and work on it as one unit.
- **Where is it used?**
  - Storing user records, to-do items, shopping carts, sensor readings, results of database queries, rows of data, queues of tasks — anywhere you need "many items, one box."

> A list is a **container** that remembers the order in which you put things in.

## 2. Real-Life Analogy

- A list is like a **train** with numbered compartments (0, 1, 2, 3...).
  - Each compartment holds one passenger (item).
  - You can ask "who is in compartment 0?" and get an answer instantly.
  - You can add a new compartment at the end, or remove one.
- Another analogy: a **shopping list on paper**.
  - Items are written in order.
  - You can cross out an item (remove), add a new item at the bottom (append), or insert one in the middle.

## 3. Explanation

- A list is created using square brackets: `my_list = [1, 2, 3]`.
- Each item in a list has a position called an **index**, starting from **0**, not 1.
- Lists are **mutable** — you can change, add, or remove items after creation.
- Lists can hold **any data type**, and even **mixed types**: `[1, "hello", 3.5, True]`.
- A **list comprehension** is a short, one-line way to build a list using a loop and (optionally) a condition, instead of writing a multi-line `for` loop.

## 4. Syntax

### Creating a list
```python
my_list = [item1, item2, item3]
```
- `[ ]` — square brackets define a list.
- Items are separated by `,` (commas).
- List name should usually be **plural** for readability.

### Accessing elements
```python
my_list[index]        # single item
my_list[start:end]    # slice, end is NOT included
my_list[-1]            # last item (negative indexing)
```

### Common list methods
```python
my_list.append(item)      # add item at the end
my_list.insert(i, item)   # add item at index i
my_list.remove(item)      # remove first matching item by value
my_list.pop(i)             # remove and return item at index i (default: last)
del my_list[i]              # delete item at index i
my_list.sort()               # sort permanently
sorted(my_list)               # sort temporarily (returns new list)
my_list.reverse()              # reverse order permanently
len(my_list)                    # number of items
```

### List Comprehension syntax
```python
new_list = [expression for item in iterable if condition]
```
- `expression` — what to store (can transform `item`).
- `for item in iterable` — the loop.
- `if condition` — **optional** filter.
- No colon at the end of the `for` part, unlike a normal `for` loop.

### 5. Examples

**Basic Example** — creating and accessing a list

````python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
print(bicycles[0])          # trek
print(bicycles[0].title())  # Trek
````

**Intermediate Example** — loop vs list comprehension

````python
# Using a for loop
squares = []
for value in range(1, 11):
    squares.append(value ** 2)
print(squares)

# Same result using list comprehension
squares = [value ** 2 for value in range(1, 11)]
print(squares)
````

**Real-World Example** — filtering data with a condition

````python
scores = [45, 78, 90, 32, 88, 55, 67]

# Get only scores that are passing (>= 50)
passing_scores = [s for s in scores if s >= 50]
print(passing_scores)

# Convert passing scores into grade labels
grades = ["Pass" if s >= 50 else "Fail" for s in scores]
print(grades)
````

## 6. Internal Working

- Internally, a Python list is a **dynamic array** (resizable array of references), not a linked list.
- Each slot stores a **reference (pointer)** to an object, not the object itself — this is why lists can hold mixed types.
- When `append()` fills the current array, Python allocates a **new, larger block** (roughly 1.125x extra) and copies old references — making append **amortized O(1)**.
- Indexing is O(1) because the array knows each slot's memory address directly.
- List comprehensions run in an optimized internal loop, often slightly faster than manual `for` + `.append()`.

## 7. Time & Space Complexity

- **Access by index** `my_list[i]` → **O(1)**
- **Append** `my_list.append(x)` → **O(1)** amortized
- **Insert at position** `my_list.insert(i, x)` → **O(n)** — shifts elements
- **Delete/pop from middle** → **O(n)** — shifts elements
- **Pop from end** → **O(1)**
- **Search (`in`, `.index()`)** → **O(n)**
- **List comprehension** → **O(n)** time, **O(n)** space (new list created)
- **Space Complexity** overall → **O(n)**

## 8. Common Mistakes

- **Off-by-one in slicing**: `my_list[0:3]` excludes index `3`.
- **Forgetting index starts at 0**.
- **Modifying a list while looping over it** — loop over a copy instead: `for item in my_list[:]`.
- **Confusing `.sort()` (in-place, returns `None`) vs `sorted()` (returns new list)**.
- **Confusing `.remove()` (by value) vs `.pop()` (by index)**.
- **Copying incorrectly**: `new_list = old_list` does not copy; use `old_list[:]` or `.copy()`.
- **IndexError**: accessing an index that doesn't exist.
- **Overcomplicating comprehensions** — use a normal loop if logic gets too nested.

## 9. Best Practices

- Use **plural, descriptive names**: `names`, `scores`, `active_users`.
- Prefer **list comprehensions** for simple transforms/filters; use loops for complex logic.
- Use `enumerate()` instead of manual index counters.
- Use `in` for membership checks.
- Use a **set** or **dict** instead of a list when uniqueness or key lookup is needed.
- Use `.copy()` or `[:]` for independent copies; `copy.deepcopy()` for nested lists.
- Keep comprehensions short and readable — one clear line.

````python
names = ['alice', 'bob', 'charlie']
for i, name in enumerate(names):
    print(i, name.title())
````

## 10. Interview Questions

**Beginner**
1. **Q: Difference between a list and a tuple?**
   A: List is mutable, uses `[]`. Tuple is immutable, uses `()`.
2. **Q: How to access the last element of a list?**
   A: `my_list[-1]`.
3. **Q: What does `len(my_list)` return?**
   A: Number of items in the list.

**Intermediate**

4. **Q: Difference between `.remove()` and `.pop()`?**
   A: `.remove(value)` deletes by value; `.pop(index)` deletes/returns by index.
5. **Q: What is a list comprehension and why use it?**
   A: A concise one-line way to build a list from an iterable with optional filtering/transformation; often more readable and slightly faster than a manual loop.
6. **Q: How to copy a list safely?**
   A: `old_list[:]` or `old_list.copy()` (shallow); `copy.deepcopy()` for nested lists.

**Advanced**

7. **Q: Why is `append()` amortized O(1)?**
   A: Because of dynamic array resizing — most appends are O(1), occasional resize is O(n), averaging to O(1).
8. **Q: Danger of mutable default arguments like `def f(x, lst=[])`?**
   A: The list is created once at function definition and shared across calls, causing unexpected accumulation. Fix with `lst=None` and `lst = lst or []`.
9. **Q: Flatten a nested list using a list comprehension?**
   A: `[item for sublist in nested for item in sublist]`

````python
# Q9 demo: flatten a nested list
nested = [[1, 2], [3, 4], [5]]
flat = [item for sublist in nested for item in sublist]
print(flat)
````

## 12. Revision

- List = ordered, mutable collection, `[]`.
- Indexing starts at **0**; `-1` is the last item.
- Slicing `list[a:b]` excludes index `b`.
- Key methods: `append`, `insert`, `remove`, `pop`, `sort`, `sorted`, `reverse`, `len`.
- Comprehension: `[expr for item in iterable if condition]` — O(n) time and space.
- `.sort()` mutates in place; `sorted()` returns a new list.
- Copying: `[:]` / `.copy()` (shallow), `copy.deepcopy()` (nested).
- Internally: dynamic array → O(1) index, O(1) amortized append, O(n) insert/delete middle.
- **Interview keyword**: "amortized O(1) append."
