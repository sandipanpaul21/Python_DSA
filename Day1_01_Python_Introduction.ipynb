## 1. Introduction

- Python is a **high-level, interpreted, general-purpose programming language**
- "High-level" means you write code close to human language, not machine code
- "Interpreted" means code runs line-by-line (no separate compile step like C/Java)
- Created by **Guido van Rossum**, released in **1991**
- Why do we need it?
  - Easy to read and write (less code than Java/C++ for the same task)
  - Huge ecosystem of libraries (NumPy, Pandas, Scikit-learn, TensorFlow, PyTorch)
  - One language covers scripting, web backend, data science, ML, automation
- Where is it used?
  - Data Science and Machine Learning (primary reason you are learning it)
  - Web backend (Django, Flask, FastAPI)
  - Automation / scripting
  - Data engineering pipelines (ETL)

> **Interview angle:** "Why Python for Data Science?" is a common opener.
> Answer with: readability + library ecosystem + community + integration with C (NumPy is fast because it wraps C code).

## 2. Real-Life Analogy

- Think of programming languages like modes of communication:
  - **Machine code** = Morse code (machine understands instantly, humans struggle)
  - **C/C++** = Formal written English with strict grammar (compiler checks every rule before it runs)
  - **Python** = Talking to a smart assistant in plain English (interpreter understands you as you speak, one line at a time)
- Python trades a little speed for a lot of human speed (faster to write, debug, and iterate)
- This is exactly why Data Science favors Python: you are experimenting with data constantly, and want fast iteration, not fast execution (heavy lifting is delegated to NumPy/Pandas which use C internally)

## 3. Explanation

- Python code is executed by the **Python Interpreter** (CPython is the default/most common implementation)
- Execution flow (simplified):
  1. You write code in a `.py` file (or a Jupyter cell)
  2. The interpreter reads it line by line
  3. Each line is translated to **bytecode**
  4. Bytecode runs on the **Python Virtual Machine (PVM)**
- Python is **dynamically typed**: you don't declare variable types, Python figures it out at runtime
- Python is **case-sensitive**: `Age` and `age` are different variables

> **Trick to remember:** Python = "Interpreted + Dynamically Typed + Easy Syntax"
> If an interviewer asks "what kind of language is Python", these 3 words are your answer.

## 4. Syntax

- No semicolons needed to end a line (unlike C/Java)
- No curly braces `{}` for blocks - Python uses **indentation** (spaces) to define blocks
- A colon `:` starts a new block (after `if`, `for`, `def`, `class`, etc.)
- Comments start with `#` (single line) or triple quotes `""" """` (multi-line / docstring)

```python
# This is a single-line comment

"""
This is a multi-line comment / docstring
"""

print("Hello, World")   # print() is the built-in function to display output
```

> **Trick to remember:** "Indentation is not style in Python, it is syntax." Wrong indentation = program crashes (`IndentationError`).

## 5. Examples

### Basic Example

````python
# Basic: printing output
print("Hello, Data Science")
````

- `print()` is a built-in function that displays text/values on the screen
- Everything inside the parentheses is the **argument** passed to the function

### Intermediate Example

````python
# Intermediate: taking input and using a variable
name = input("Enter your name: ")   # input() always returns a string
print("Welcome,", name)
````

- `input()` pauses the program, waits for user to type, returns whatever was typed as a **string**
- `name = input(...)` stores that string inside a variable called `name`
- `print("Welcome,", name)` prints multiple items separated by a comma (Python auto-adds a space between them)

### Real-World Example

````python
# Real-world: a tiny "data check" script
# Common first step in any EDA (Exploratory Data Analysis) task

dataset_name = "employee_salary.csv"
num_rows = 1000
num_columns = 8

print(f"Dataset: {dataset_name}")
print(f"Shape: {num_rows} rows x {num_columns} columns")
````

- `f"..."` is an **f-string** (formatted string) - lets you insert variable values directly inside `{}`
- This pattern (printing dataset shape/name) is literally the first thing you do in real EDA work, e.g. `print(df.shape)`
- f-strings are the modern, Pythonic way to build strings (preferred over `%` formatting or `.format()` in interviews)

## 6. Internal Working

- When you run a `.py` file:
  1. CPython interpreter compiles your source code into **bytecode** (`.pyc` files, cached)
  2. Bytecode is executed by the **Python Virtual Machine (PVM)**
- Python manages memory automatically using a **Garbage Collector** (you don't free memory manually like in C)
- Variables in Python are **references (names) pointing to objects in memory**, not boxes holding values directly

> **Trick to remember:** In Python, "variables are labels, not boxes." `a = 5` means the label `a` points to the object `5` in memory - it does not mean `a` IS a box containing 5.
> This exact mental model is why `mutable vs immutable` behavior later makes sense.

## 7. Time and Space Complexity

- Not directly applicable to "Python Introduction" itself (no algorithm here)
- But one core interview fact belongs here:
  - Python is generally **slower than C/C++/Java** for raw computation (interpreted + dynamic typing overhead)
  - Data Science libraries (NumPy, Pandas) get around this by doing heavy computation in **C under the hood**, and just exposing a Python interface
- **Interview line to remember:** "Python itself is slow, but NumPy/Pandas are fast because they call optimized C code - Python is just the wrapper you interact with."</br>

## 8. Common Mistakes

- Mixing **tabs and spaces** for indentation -> `IndentationError` (always use spaces, 4 spaces is standard)
- Forgetting the colon `:` after `if`, `for`, `def`, etc.
- Assuming `input()` returns a number - it always returns a **string**, must convert manually (e.g. `int(input())`)
- Using `print` without parentheses (Python 2 habit) - Python 3 requires `print(...)`
- Confusing `=` (assignment) with `==` (comparison)

## 9. Best Practices

- Use 4 spaces for indentation (PEP 8 standard) - never tabs
- Use meaningful variable names: `num_rows` not `n`
- Follow **PEP 8** (Python's official style guide) - lowercase_with_underscores for variables/functions
- Add docstrings to explain what a script/function does
- Prefer f-strings for building strings over `%` or `.format()`

> **Interview tip:** Mentioning "PEP 8" by name signals you know Python conventions, not just syntax.

## 10. Interview Questions

**Beginner**
- Q: Is Python compiled or interpreted?
  A: Interpreted (technically compiled to bytecode first, then interpreted by the PVM - so it's a hybrid, but commonly called "interpreted").
- Q: Is Python statically or dynamically typed?
  A: Dynamically typed - you don't declare variable types; Python infers them at runtime.

**Intermediate**
- Q: What is CPython?
  A: The default and most widely used implementation of Python, written in C. When people say "Python," they usually mean CPython.
- Q: Why is Python popular for Data Science despite being slower than C?
  A: Because performance-critical libraries (NumPy, Pandas, TensorFlow) are written in C/C++ under the hood; Python just provides a simple interface to call them.

**Advanced**
- Q: What is the difference between a `.py` script execution and bytecode?
  A: Source code (`.py`) is first compiled to bytecode (`.pyc`), which is a lower-level, platform-independent representation. The Python Virtual Machine then executes this bytecode - this is why Python is described as "compiled to bytecode, then interpreted."
- Q: Why is Python considered a "glue language"?
  A: Because it easily integrates and connects components written in other languages (C, C++, Java) - common in Data Science where the heavy math is in C but the interface is Python.
