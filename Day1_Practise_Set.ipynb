### 1. Two Sum  (Easy)

Given an array of integers `nums` and an integer `target`, return the indices of the two numbers that add up to `target`.

**Example:** `nums=[2,7,11,15], target=9` → `[0,1]`

```python
def two_sum(nums, target):
    seen = {}
    for i, n in enumerate(nums):
        complement = target - n
        if complement in seen:
            return [seen[complement], i]
        seen[n] = i
    return []

print(two_sum([2,7,11,15], 9))
```

**Complexity:** Time O(n), Space O(n) — single pass using a hash map.

### 2. Reverse Integer  (Easy)

Given a 32-bit signed integer, reverse its digits. Return 0 if the result overflows 32-bit signed range.

**Example:** `x=123` → `321`, `x=-123` → `-321`

```python
def reverse_integer(x):
    sign = -1 if x < 0 else 1
    x = abs(x)
    reversed_num = int(str(x)[::-1])
    reversed_num *= sign
    if reversed_num < -2**31 or reversed_num > 2**31 - 1:
        return 0
    return reversed_num

print(reverse_integer(123), reverse_integer(-123))
```

**Complexity:** Time O(d) where d = number of digits, Space O(d) for the string conversion.

### 3. Palindrome Number  (Easy)

Determine if an integer is a palindrome without converting it entirely to a string (though a simple string check is also acceptable for interviews).

**Example:** `121` → `True`, `-121` → `False`

```python
def is_palindrome(x):
    if x < 0:
        return False
    s = str(x)
    return s == s[::-1]

print(is_palindrome(121), is_palindrome(-121))
```

**Complexity:** Time O(d), Space O(d) where d is number of digits.

### 4. Valid Parentheses  (Easy)

Given a string of brackets `()[]{}`, determine if the input is valid (every opening bracket has a matching closing bracket in the correct order).

**Example:** `"()[]{}"` → `True`, `"(]"` → `False`

```python
def is_valid(s):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}
    for ch in s:
        if ch in pairs.values():
            stack.append(ch)
        elif ch in pairs:
            if not stack or stack.pop() != pairs[ch]:
                return False
    return not stack

print(is_valid("()[]{}"), is_valid("(]"))
```

**Complexity:** Time O(n), Space O(n) — using a stack.

### 5. Remove Duplicates from Sorted Array  (Easy)

Given a sorted array, remove duplicates in-place so each element appears once, and return the new length.

**Example:** `[1,1,2]` → length `2`, array becomes `[1,2,...]`

```python
def remove_duplicates(nums):
    if not nums:
        return 0
    write = 1
    for read in range(1, len(nums)):
        if nums[read] != nums[write - 1]:
            nums[write] = nums[read]
            write += 1
    return write

arr = [1,1,2]
print(remove_duplicates(arr), arr[:2])
```

**Complexity:** Time O(n), Space O(1) — two-pointer technique.

### 6. Search Insert Position  (Easy)

Given a sorted array and a target, return the index if found, or the index where it would be inserted to keep the array sorted.

**Example:** `[1,3,5,6], target=5` → `2`; `target=2` → `1`

```python
def search_insert(nums, target):
    lo, hi = 0, len(nums)
    while lo < hi:
        mid = (lo + hi) // 2
        if nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid
    return lo

print(search_insert([1,3,5,6], 5), search_insert([1,3,5,6], 2))
```

**Complexity:** Time O(log n), Space O(1) — binary search.

### 9. Sqrt(x)  (Easy)

Compute and return the integer (floor) square root of a non-negative integer x, without using `x ** 0.5` directly (binary search preferred).

**Example:** `x=8` → `2`

```python
def my_sqrt(x):
    if x < 2:
        return x
    lo, hi = 1, x // 2
    while lo <= hi:
        mid = (lo + hi) // 2
        if mid * mid == x:
            return mid
        elif mid * mid < x:
            lo = mid + 1
        else:
            hi = mid - 1
    return hi

print(my_sqrt(8))
```

**Complexity:** Time O(log x), Space O(1) — binary search.

### 10. Climbing Stairs  (Easy)

You can climb 1 or 2 steps at a time. Given n steps, count how many distinct ways to reach the top.

**Example:** `n=3` → `3` (1+1+1, 1+2, 2+1)

```python
def climb_stairs(n):
    if n <= 2:
        return n
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b

print(climb_stairs(3), climb_stairs(5))
```

**Complexity:** Time O(n), Space O(1) — Fibonacci-style DP.

### 11. Merge Sorted Array  (Easy)

Merge two sorted arrays `nums1` (with extra space) and `nums2` into `nums1` in sorted order.

**Example:** `nums1=[1,2,3,0,0,0], m=3, nums2=[2,5,6], n=3` → `[1,2,2,3,5,6]`

```python
def merge(nums1, m, nums2, n):
    i, j, k = m - 1, n - 1, m + n - 1
    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
    return nums1

print(merge([1,2,3,0,0,0], 3, [2,5,6], 3))
```

**Complexity:** Time O(m+n), Space O(1) — merge from the back.

### 12. Single Number  (Easy)

Given a list where every element appears twice except one, find that single one.

**Example:** `[4,1,2,1,2]` → `4`

```python
def single_number(nums):
    result = 0
    for n in nums:
        result ^= n
    return result

print(single_number([4,1,2,1,2]))
```

**Complexity:** Time O(n), Space O(1) — XOR cancels out pairs.

### 13. Majority Element  (Easy)

Given an array, find the element that appears more than n/2 times (guaranteed to exist).

**Example:** `[2,2,1,1,1,2,2]` → `2`

```python
def majority_element(nums):
    count, candidate = 0, None
    for n in nums:
        if count == 0:
            candidate = n
        count += 1 if n == candidate else -1
    return candidate

print(majority_element([2,2,1,1,1,2,2]))
```

**Complexity:** Time O(n), Space O(1) — Boyer-Moore voting algorithm.

### 14. Move Zeroes  (Easy)

Given an array, move all zeros to the end while maintaining the relative order of non-zero elements, in-place.

**Example:** `[0,1,0,3,12]` → `[1,3,12,0,0]`

```python
def move_zeroes(nums):
    write = 0
    for read in range(len(nums)):
        if nums[read] != 0:
            nums[write], nums[read] = nums[read], nums[write]
            write += 1
    return nums

print(move_zeroes([0,1,0,3,12]))
```

**Complexity:** Time O(n), Space O(1) — two-pointer swap.

### 15. Missing Number  (Easy)

Given an array containing n distinct numbers from 0 to n, find the missing one.

**Example:** `[3,0,1]` → `2`

```python
def missing_number(nums):
    n = len(nums)
    expected = n * (n + 1) // 2
    return expected - sum(nums)

print(missing_number([3,0,1]))
```

**Complexity:** Time O(n), Space O(1) — sum formula.

### 18. Best Time to Buy and Sell Stock  (Easy)

Given daily stock prices, find the maximum profit from a single buy and single sell (buy before sell).

**Example:** `[7,1,5,3,6,4]` → `5` (buy at 1, sell at 6)

```python
def max_profit(prices):
    min_price = float('inf')
    max_profit = 0
    for p in prices:
        min_price = min(min_price, p)
        max_profit = max(max_profit, p - min_price)
    return max_profit

print(max_profit([7,1,5,3,6,4]))
```

**Complexity:** Time O(n), Space O(1) — track running minimum.

### 22. Reverse String  (Easy)

Reverse a list of characters in-place.

**Example:** `['h','e','l','l','o']` → `['o','l','l','e','h']`

```python
def reverse_string(s):
    left, right = 0, len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left]
        left += 1
        right -= 1
    return s

print(reverse_string(['h','e','l','l','o']))
```

**Complexity:** Time O(n), Space O(1) — two-pointer swap.

### 24. Implement strStr() / Substring Search  (Easy)

Return the index of the first occurrence of `needle` in `haystack`, or -1 if not found.

**Example:** `haystack="hello", needle="ll"` → `2`

```python
def str_str(haystack, needle):
    if needle == "":
        return 0
    n, m = len(haystack), len(needle)
    for i in range(n - m + 1):
        if haystack[i:i+m] == needle:
            return i
    return -1

print(str_str("hello", "ll"))
```

**Complexity:** Time O(n*m) worst case, Space O(1) — brute-force sliding window.

### 25. Longest Common Prefix  (Easy)

Find the longest common prefix string among an array of strings.

**Example:** `["flower","flow","flight"]` → `"fl"`

```python
def longest_common_prefix(strs):
    if not strs:
        return ""
    prefix = strs[0]
    for s in strs[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if not prefix:
                return ""
    return prefix

print(longest_common_prefix(["flower","flow","flight"]))
```

**Complexity:** Time O(n*m), Space O(1) — shrinking prefix comparison.

### 26. Roman to Integer  (Easy)

Convert a Roman numeral string to an integer.

**Example:** `"MCMXCIV"` → `1994`

```python
def roman_to_int(s):
    values = {'I':1,'V':5,'X':10,'L':50,'C':100,'D':500,'M':1000}
    total = 0
    for i in range(len(s)):
        if i + 1 < len(s) and values[s[i]] < values[s[i+1]]:
            total -= values[s[i]]
        else:
            total += values[s[i]]
    return total

print(roman_to_int("MCMXCIV"))
```

**Complexity:** Time O(n), Space O(1) — single pass with lookahead.

### 27. Integer to Roman  (Easy)

Convert an integer (1-3999) into a Roman numeral string.

**Example:** `1994` → `"MCMXCIV"`

```python
def int_to_roman(num):
    values = [1000,900,500,400,100,90,50,40,10,9,5,4,1]
    symbols = ["M","CM","D","CD","C","XC","L","XL","X","IX","V","IV","I"]
    result = []
    for v, s in zip(values, symbols):
        while num >= v:
            result.append(s)
            num -= v
    return ''.join(result)

print(int_to_roman(1994))
```

**Complexity:** Time O(1) bounded (13 symbols max), Space O(1).

### 31. Power of Three  (Easy)

```python
def is_power_of_three(n):
    if n < 1:
        return False
    while n % 3 == 0:
        n //= 3
    return n == 1

print(is_power_of_three(27), is_power_of_three(45))
```

### 36. Happy Number  (Easy)

A happy number eventually reaches 1 when replaced by the sum of the square of its digits, repeated; determine if n is happy.

```python
def is_happy(n):
    seen = set()
    while n != 1 and n not in seen:
        seen.add(n)
        n = sum(int(d)**2 for d in str(n))
    return n == 1

print(is_happy(19))
```

### 37. Isomorphic Strings  (Easy)

Determine if two strings are isomorphic — characters in s can be replaced to get t, with a consistent one-to-one mapping.

**Example:** `"egg", "add"` → `True`

```python
def is_isomorphic(s, t):
    if len(s) != len(t):
        return False
    map_st, map_ts = {}, {}
    for a, b in zip(s, t):
        if a in map_st and map_st[a] != b:
            return False
        if b in map_ts and map_ts[b] != a:
            return False
        map_st[a] = b
        map_ts[b] = a
    return True

print(is_isomorphic("egg", "add"))
```

**Complexity:** Time O(n), Space O(n) — two hash maps for bidirectional mapping.

### 38. Word Pattern  (Easy)

Given a pattern and a string of words, determine if the string follows the same pattern (bijective mapping).

**Example:** `pattern="abba", s="dog cat cat dog"` → `True`

```python
def word_pattern(pattern, s):
    words = s.split()
    if len(pattern) != len(words):
        return False
    return is_isomorphic(pattern, ' '.join(words)) if False else _wp(pattern, words)

def _wp(pattern, words):
    p2w, w2p = {}, {}
    for p, w in zip(pattern, words):
        if p in p2w and p2w[p] != w:
            return False
        if w in w2p and w2p[w] != p:
            return False
        p2w[p] = w
        w2p[w] = p
    return True

print(word_pattern("abba", "dog cat cat dog"))
```

**Complexity:** Time O(n), Space O(n) — bidirectional mapping, same idea as Isomorphic Strings.

### 39. Ugly Number  (Easy)

An ugly number is a positive integer whose prime factors are limited to 2, 3, and 5. Determine if n is ugly.

**Example:** `6` → `True`; `14` → `False`

```python
def is_ugly(n):
    if n <= 0:
        return False
    for factor in (2, 3, 5):
        while n % factor == 0:
            n //= factor
    return n == 1

print(is_ugly(6), is_ugly(14))
```

**Complexity:** Time O(log n), Space O(1) — repeated division by 2, 3, 5.

### 40. Factorial Trailing Zeroes  (Easy)

Given an integer n, return the number of trailing zeroes in n!.

**Example:** `n=5` → `1` (5! = 120)

```python
def trailing_zeroes(n):
    count = 0
    power = 5
    while power <= n:
        count += n // power
        power *= 5
    return count

print(trailing_zeroes(5), trailing_zeroes(25))
```

**Complexity:** Time O(log₅ n), Space O(1) — count factors of 5, since 2s are always more plentiful.

### 41. Pascal's Triangle  (Easy)

Generate the first n rows of Pascal's Triangle.

**Example:** `n=5` → `[[1],[1,1],[1,2,1],[1,3,3,1],[1,4,6,4,1]]`

```python
def generate(n):
    triangle = []
    for i in range(n):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[i-1][j-1] + triangle[i-1][j]
        triangle.append(row)
    return triangle

print(generate(5))
```

**Complexity:** Time O(n²), Space O(n²) — build row by row using the previous row.

### 42. Pascal's Triangle II  (Easy)

### 43. Two Sum II - Sorted Input  (Easy)

Given a sorted array, find two numbers that add up to target, returning 1-indexed positions.

**Example:** `[2,7,11,15], target=9` → `[1,2]`

```python
def two_sum_sorted(numbers, target):
    left, right = 0, len(numbers) - 1
    while left < right:
        s = numbers[left] + numbers[right]
        if s == target:
            return [left + 1, right + 1]
        elif s < target:
            left += 1
        else:
            right -= 1
    return []

print(two_sum_sorted([2,7,11,15], 9))
```

**Complexity:** Time O(n), Space O(1) — two-pointer technique on sorted data.

### 44. Third Maximum Number  (Easy)

Given an array, return the third distinct maximum number; if it doesn't exist, return the maximum.

**Example:** `[3,2,1]` → `1`; `[1,2]` → `2`

```python
def third_max(nums):
    top3 = set()
    for n in nums:
        top3.add(n)
        if len(top3) > 3:
            top3.remove(min(top3))
    return min(top3) if len(top3) == 3 else max(top3)

print(third_max([3,2,1]), third_max([1,2]))
```

**Complexity:** Time O(n), Space O(1) (bounded set of size ≤4) — maintain top-3 distinct values.

### 47. Assign Cookies  (Easy)

Given children's greed factors and cookie sizes, maximize the number of content children (a child is content if cookie size >= greed).

**Example:** `g=[1,2,3], s=[1,1]` → `1`

```python
def find_content_children(g, s):
    g.sort()
    s.sort()
    child = cookie = 0
    while child < len(g) and cookie < len(s):
        if s[cookie] >= g[child]:
            child += 1
        cookie += 1
    return child

print(find_content_children([1,2,3], [1,1]))
```

**Complexity:** Time O(n log n), Space O(1) extra — greedy two-pointer after sorting.

### 49. Find the Difference  (Easy)

String t is s shuffled with one extra letter added. Find that extra letter.

**Example:** `s="abcd", t="abcde"` → `"e"`

```python
def find_the_difference(s, t):
    result = 0
    for ch in s + t:
        result ^= ord(ch)
    return chr(result)

print(find_the_difference("abcd", "abcde"))
```

**Complexity:** Time O(n), Space O(1) — XOR cancels matching characters.

### 50. Two Sum Sorted Duplicate Check (Binary Search Variant)  (Easy)

Given a sorted array, determine if target exists using binary search.

**Example:** `[1,3,5,7,9], target=7` → `True`

```python
def binary_search(nums, target):
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return True
        elif nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return False

print(binary_search([1,3,5,7,9], 7))
```

**Complexity:** Time O(log n), Space O(1) — classic binary search.

### 51. Longest Substring Without Repeating Characters  (Medium)

Find the length of the longest substring without repeating characters.

**Example:** `"abcabcbb"` → `3` ("abc")

```python
def length_of_longest_substring(s):
    seen = {}
    left = 0
    best = 0
    for right, ch in enumerate(s):
        if ch in seen and seen[ch] >= left:
            left = seen[ch] + 1
        seen[ch] = right
        best = max(best, right - left + 1)
    return best

print(length_of_longest_substring("abcabcbb"))
```

**Complexity:** Time O(n), Space O(min(n, alphabet size)) — sliding window with a hash map.

### 52. Longest Palindromic Substring  (Medium)

Find the longest palindromic substring in s.

**Example:** `"babad"` → `"bab"` or `"aba"`

```python
def longest_palindrome(s):
    def expand(l, r):
        while l >= 0 and r < len(s) and s[l] == s[r]:
            l -= 1
            r += 1
        return s[l+1:r]

    result = ""
    for i in range(len(s)):
        odd = expand(i, i)
        even = expand(i, i + 1)
        result = max(result, odd, even, key=len)
    return result

print(longest_palindrome("babad"))
```

**Complexity:** Time O(n²), Space O(1) extra — expand around center for each position.

### 53. 3Sum  (Medium)

Find all unique triplets in an array that sum to zero.

**Example:** `[-1,0,1,2,-1,-4]` → `[[-1,-1,2],[-1,0,1]]`

```python
def three_sum(nums):
    nums.sort()
    result = []
    for i in range(len(nums) - 2):
        if i > 0 and nums[i] == nums[i-1]:
            continue
        left, right = i + 1, len(nums) - 1
        while left < right:
            s = nums[i] + nums[left] + nums[right]
            if s == 0:
                result.append([nums[i], nums[left], nums[right]])
                left += 1
                right -= 1
                while left < right and nums[left] == nums[left-1]:
                    left += 1
                while left < right and nums[right] == nums[right+1]:
                    right -= 1
            elif s < 0:
                left += 1
            else:
                right -= 1
    return result

print(three_sum([-1,0,1,2,-1,-4]))
```

**Complexity:** Time O(n²), Space O(1) extra (excluding output) — sort + two-pointer per fixed element.

### 55. Spiral Matrix  (Medium)

Given an m x n matrix, return all elements in spiral order.

**Example:** `[[1,2,3],[4,5,6],[7,8,9]]` → `[1,2,3,6,9,8,7,4,5]`

```python
def spiral_order(matrix):
    result = []
    while matrix:
        result += matrix.pop(0)
        if matrix and matrix[0]:
            for row in matrix:
                result.append(row.pop())
        if matrix:
            result += matrix.pop()[::-1]
        if matrix and matrix[0]:
            for row in matrix[::-1]:
                result.append(row.pop(0))
    return result

print(spiral_order([[1,2,3],[4,5,6],[7,8,9]]))
```

### 57. Merge Intervals  (Medium)

Given a list of intervals, merge all overlapping intervals.

**Example:** `[[1,3],[2,6],[8,10],[15,18]]` → `[[1,6],[8,10],[15,18]]`

```python
def merge_intervals(intervals):
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    for start, end in intervals[1:]:
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return merged

print(merge_intervals([[1,3],[2,6],[8,10],[15,18]]))
```

### 58. Insert Interval  (Medium)

Given a sorted, non-overlapping list of intervals and a new interval, insert and merge as needed.

**Example:** `[[1,3],[6,9]], newInterval=[2,5]` → `[[1,5],[6,9]]`

```python
def insert(intervals, new_interval):
    result = []
    i, n = 0, len(intervals)
    while i < n and intervals[i][1] < new_interval[0]:
        result.append(intervals[i])
        i += 1
    while i < n and intervals[i][0] <= new_interval[1]:
        new_interval[0] = min(new_interval[0], intervals[i][0])
        new_interval[1] = max(new_interval[1], intervals[i][1])
        i += 1
    result.append(new_interval)
    result += intervals[i:]
    return result

print(insert([[1,3],[6,9]], [2,5]))
```

**Complexity:** Time O(n), Space O(n) — linear scan with three phases (before, overlap, after).

### 59. Unique Paths  (Medium)

```python
def unique_paths(m, n):
    dp = [[1] * n for _ in range(m)]
    for i in range(1, m):
        for j in range(1, n):
            dp[i][j] = dp[i-1][j] + dp[i][j-1]
    return dp[m-1][n-1]

print(unique_paths(3, 7))
```

### 60. Set Matrix Zeroes  (Medium)

### 61. Sort Colors (Dutch National Flag)  (Medium)

Given an array with values 0, 1, 2, sort it in-place in one pass.

**Example:** `[2,0,2,1,1,0]` → `[0,0,1,1,2,2]`

```python
def sort_colors(nums):
    low, mid, high = 0, 0, len(nums) - 1
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
    return nums

print(sort_colors([2,0,2,1,1,0]))
```

**Complexity:** Time O(n), Space O(1) — Dutch National Flag three-pointer partition.

### 63. Word Search  (Medium)

Given a 2D board of letters and a word, determine if the word exists by moving to adjacent cells (no reuse).

**Example:** `board=[['A','B','C','E'],['S','F','C','S'],['A','D','E','E']], word="ABCCED"` → `True`

```python
def exist(board, word):
    rows, cols = len(board), len(board[0])

    def backtrack(r, c, i):
        if i == len(word):
            return True
        if r < 0 or r >= rows or c < 0 or c >= cols or board[r][c] != word[i]:
            return False
        temp, board[r][c] = board[r][c], '#'
        found = (backtrack(r+1, c, i+1) or backtrack(r-1, c, i+1) or
                 backtrack(r, c+1, i+1) or backtrack(r, c-1, i+1))
        board[r][c] = temp
        return found

    for r in range(rows):
        for c in range(cols):
            if backtrack(r, c, 0):
                return True
    return False

print(exist([['A','B','C','E'],['S','F','C','S'],['A','D','E','E']], "ABCCED"))
```

**Complexity:** Time O(n * 4^L) where L=word length, Space O(L) recursion depth — backtracking DFS.

### 64. Search in Rotated Sorted Array  (Medium)

Given a rotated sorted array, search for a target value in O(log n).

**Example:** `[4,5,6,7,0,1,2], target=0` → `4`

```python
def search_rotated(nums, target):
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        if nums[lo] <= nums[mid]:
            if nums[lo] <= target < nums[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        else:
            if nums[mid] < target <= nums[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1

print(search_rotated([4,5,6,7,0,1,2], 0))
```

**Complexity:** Time O(log n), Space O(1) — modified binary search.

### 65. Find First and Last Position of Element in Sorted Array  (Medium)

Given a sorted array, find the starting and ending position of a target value.

**Example:** `[5,7,7,8,8,10], target=8` → `[3,4]`

```python
import bisect

def search_range(nums, target):
    left = bisect.bisect_left(nums, target)
    if left == len(nums) or nums[left] != target:
        return [-1, -1]
    right = bisect.bisect_right(nums, target) - 1
    return [left, right]

print(search_range([5,7,7,8,8,10], 8))
```

**Complexity:** Time O(log n), Space O(1) — two binary searches (bisect left/right).

### 66. Combination Sum  (Medium)

Given distinct candidates and a target, find all unique combinations that sum to target (numbers can repeat).

**Example:** `candidates=[2,3,6,7], target=7` → `[[2,2,3],[7]]`

```python
def combination_sum(candidates, target):
    result = []

    def backtrack(start, path, remaining):
        if remaining == 0:
            result.append(path[:])
            return
        if remaining < 0:
            return
        for i in range(start, len(candidates)):
            path.append(candidates[i])
            backtrack(i, path, remaining - candidates[i])
            path.pop()

    backtrack(0, [], target)
    return result

print(combination_sum([2,3,6,7], 7))
```

**Complexity:** Time O(2^target) worst case, Space O(target) recursion — backtracking.

### 67. Permutations  (Medium)

```python
def permute(nums):
    if len(nums) <= 1:
        return [nums]
    result = []
    for i in range(len(nums)):
        rest = nums[:i] + nums[i+1:]
        for p in permute(rest):
            result.append([nums[i]] + p)
    return result

print(permute([1,2,3]))
```

### 69. Product of Array Except Self  (Medium)

### 72. Container With Most Water  (Medium)

Given heights representing vertical lines, find two lines that form a container holding the most water.

```python
def max_area(height):
    left, right = 0, len(height) - 1
    best = 0
    while left < right:
        best = max(best, (right - left) * min(height[left], height[right]))
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    return best

print(max_area([1,8,6,2,5,4,8,3,7]))
```

### 73. Longest Increasing Subsequence  (Medium)

```python
import bisect

def length_of_lis(nums):
    tails = []
    for n in nums:
        idx = bisect.bisect_left(tails, n)
        if idx == len(tails):
            tails.append(n)
        else:
            tails[idx] = n
    return len(tails)

print(length_of_lis([10,9,2,5,3,7,101,18]))
```

**Complexity:** Time O(n log n), Space O(n) — patience sorting with binary search.

### 74. Coin Change  (Medium)

Given coin denominations and a target amount, find the fewest coins needed to make that amount (-1 if impossible).

**Example:** `coins=[1,2,5], amount=11` → `3` (5+5+1)

```python
def coin_change(coins, amount):
    dp = [0] + [float('inf')] * amount
    for a in range(1, amount + 1):
        for c in coins:
            if c <= a:
                dp[a] = min(dp[a], dp[a - c] + 1)
    return dp[amount] if dp[amount] != float('inf') else -1

print(coin_change([1,2,5], 11))
```

### 77. Minimum Path Sum  (Medium)

Given a grid of non-negative numbers, find a path from top-left to bottom-right minimizing the sum (only right/down moves).

**Example:** `[[1,3,1],[1,5,1],[4,2,1]]` → `7`

```python
def min_path_sum(grid):
    m, n = len(grid), len(grid[0])
    for i in range(m):
        for j in range(n):
            if i == 0 and j == 0:
                continue
            elif i == 0:
                grid[i][j] += grid[i][j-1]
            elif j == 0:
                grid[i][j] += grid[i-1][j]
            else:
                grid[i][j] += min(grid[i-1][j], grid[i][j-1])
    return grid[-1][-1]

print(min_path_sum([[1,3,1],[1,5,1],[4,2,1]]))
```

### 78. Decode Ways  (Medium)

### 80. Binary Search  (Medium)

### 81. Trapping Rain Water  (Hard)

Given elevation heights, compute how much water can be trapped after raining.

**Example:** `[0,1,0,2,1,0,1,3,2,1,2,1]` → `6`

```python
def trap(height):
    if not height:
        return 0
    left, right = 0, len(height) - 1
    left_max, right_max = height[left], height[right]
    water = 0
    while left < right:
        if left_max < right_max:
            left += 1
            left_max = max(left_max, height[left])
            water += left_max - height[left]
        else:
            right -= 1
            right_max = max(right_max, height[right])
            water += right_max - height[right]
    return water

print(trap([0,1,0,2,1,0,1,3,2,1,2,1]))
```

**Complexity:** Time O(n), Space O(1) — two-pointer with running max from each side.

### 82. Median of Two Sorted Arrays  (Hard)

Given two sorted arrays, find the median of the combined array in O(log(m+n)).

**Example:** `nums1=[1,3], nums2=[2]` → `2.0`

```python
def find_median_sorted_arrays(nums1, nums2):
    merged = sorted(nums1 + nums2)
    n = len(merged)
    mid = n // 2
    if n % 2 == 0:
        return (merged[mid - 1] + merged[mid]) / 2
    return float(merged[mid])

print(find_median_sorted_arrays([1,3], [2]))
```

**Complexity:** Time O((m+n) log(m+n)) with this simple merge-sort approach (true O(log(min(m,n))) needs binary-search partitioning), Space O(m+n).

### 83. Merge K Sorted Arrays  (Hard)

Given k sorted arrays, merge them into one sorted array.

**Example:** `[[1,4,5],[1,3,4],[2,6]]` → `[1,1,2,3,4,4,5,6]`

```python
import heapq

def merge_k_sorted(arrays):
    heap = []
    for i, arr in enumerate(arrays):
        if arr:
            heapq.heappush(heap, (arr[0], i, 0))
    result = []
    while heap:
        val, arr_idx, elem_idx = heapq.heappop(heap)
        result.append(val)
        if elem_idx + 1 < len(arrays[arr_idx]):
            heapq.heappush(heap, (arrays[arr_idx][elem_idx+1], arr_idx, elem_idx+1))
    return result

print(merge_k_sorted([[1,4,5],[1,3,4],[2,6]]))
```

**Complexity:** Time O(N log k) where N=total elements, k=number of arrays, Space O(k) heap — min-heap merge.

### 84. Largest Rectangle in Histogram  (Hard)

Given bar heights, find the area of the largest rectangle in the histogram.

**Example:** `[2,1,5,6,2,3]` → `10`

```python
def largest_rectangle_area(heights):
    stack = []
    max_area = 0
    heights = heights + [0]
    for i, h in enumerate(heights):
        while stack and heights[stack[-1]] > h:
            height = heights[stack.pop()]
            width = i if not stack else i - stack[-1] - 1
            max_area = max(max_area, height * width)
        stack.append(i)
    return max_area

print(largest_rectangle_area([2,1,5,6,2,3]))
```

**Complexity:** Time O(n), Space O(n) — monotonic increasing stack.

### 85. N-Queens (Count Solutions)  (Hard)

Given n, count the number of ways to place n queens on an n x n board so none attack each other.

**Example:** `n=4` → `2` solutions

```python
def total_n_queens(n):
    count = 0
    cols, diag1, diag2 = set(), set(), set()

    def backtrack(row):
        nonlocal count
        if row == n:
            count += 1
            return
        for col in range(n):
            if col in cols or (row - col) in diag1 or (row + col) in diag2:
                continue
            cols.add(col); diag1.add(row - col); diag2.add(row + col)
            backtrack(row + 1)
            cols.remove(col); diag1.remove(row - col); diag2.remove(row + col)

    backtrack(0)
    return count

print(total_n_queens(4))
```

**Complexity:** Time O(n!) worst case, Space O(n) recursion — backtracking with column/diagonal sets.

### 86. Edit Distance  (Hard)

Find the minimum number of insert/delete/replace operations to convert word1 into word2.

**Example:** `"horse", "ros"` → `3`

```python
def min_distance(word1, word2):
    m, n = len(word1), len(word2)
    dp = [[0]*(n+1) for _ in range(m+1)]
    for i in range(m+1):
        dp[i][0] = i
    for j in range(n+1):
        dp[0][j] = j
    for i in range(1, m+1):
        for j in range(1, n+1):
            if word1[i-1] == word2[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(dp[i-1][j], dp[i][j-1], dp[i-1][j-1])
    return dp[m][n]

print(min_distance("horse", "ros"))
```

**Complexity:** Time O(m*n), Space O(m*n) — classic 2D dynamic programming.

### 87. Longest Valid Parentheses  (Hard)

Given a string of '(' and ')', find the length of the longest valid parentheses substring.

**Example:** `"(()"` → `2`; `")()())"` → `4`

```python
def longest_valid_parentheses(s):
    stack = [-1]
    best = 0
    for i, ch in enumerate(s):
        if ch == '(':
            stack.append(i)
        else:
            stack.pop()
            if not stack:
                stack.append(i)
            else:
                best = max(best, i - stack[-1])
    return best

print(longest_valid_parentheses(")()())"))
```

**Complexity:** Time O(n), Space O(n) — stack tracks indices of unmatched positions.

### 88. First Missing Positive  (Hard)

Given an unsorted array, find the smallest missing positive integer in O(n) time and O(1) extra space.

**Example:** `[3,4,-1,1]` → `2`

```python
def first_missing_positive(nums):
    n = len(nums)
    for i in range(n):
        while 1 <= nums[i] <= n and nums[nums[i]-1] != nums[i]:
            nums[nums[i]-1], nums[i] = nums[i], nums[nums[i]-1]
    for i in range(n):
        if nums[i] != i + 1:
            return i + 1
    return n + 1

print(first_missing_positive([3,4,-1,1]))
```

**Complexity:** Time O(n), Space O(1) — index-based cyclic placement (each value placed at its 'home' index).

### 89. Sliding Window Maximum  (Hard)

Given an array and window size k, return the max of each sliding window.

**Example:** `[1,3,-1,-3,5,3,6,7], k=3` → `[3,3,5,5,6,7]`

```python
from collections import deque

def max_sliding_window(nums, k):
    dq = deque()
    result = []
    for i, n in enumerate(nums):
        while dq and nums[dq[-1]] < n:
            dq.pop()
        dq.append(i)
        if dq[0] <= i - k:
            dq.popleft()
        if i >= k - 1:
            result.append(nums[dq[0]])
    return result

print(max_sliding_window([1,3,-1,-3,5,3,6,7], 3))
```

**Complexity:** Time O(n), Space O(k) — monotonic decreasing deque of indices.

### 90. Minimum Window Substring  (Hard)

Given strings s and t, find the smallest substring of s containing all characters of t (with counts).

**Example:** `s="ADOBECODEBANC", t="ABC"` → `"BANC"`

```python
from collections import Counter

def min_window(s, t):
    if not s or not t:
        return ""
    need = Counter(t)
    missing = len(t)
    left = start = end = 0
    for right, ch in enumerate(s, 1):
        if need[ch] > 0:
            missing -= 1
        need[ch] -= 1
        if missing == 0:
            while left < right and need[s[left]] < 0:
                need[s[left]] += 1
                left += 1
            if end == 0 or right - left < end - start:
                start, end = left, right
            need[s[left]] += 1
            missing += 1
            left += 1
    return s[start:end]

print(min_window("ADOBECODEBANC", "ABC"))
```

**Complexity:** Time O(n), Space O(k) where k = distinct chars in t — sliding window with a need-counter.

### 91. Word Ladder (Shortest Transformation Length)  (Hard)

Given a beginWord, endWord, and wordList, find the length of the shortest transformation sequence changing one letter at a time.

**Example:** `beginWord="hit", endWord="cog", wordList=["hot","dot","dog","lot","log","cog"]` → `5`

```python
from collections import deque

def ladder_length(begin_word, end_word, word_list):
    words = set(word_list)
    if end_word not in words:
        return 0
    queue = deque([(begin_word, 1)])
    while queue:
        word, steps = queue.popleft()
        if word == end_word:
            return steps
        for i in range(len(word)):
            for c in 'abcdefghijklmnopqrstuvwxyz':
                next_word = word[:i] + c + word[i+1:]
                if next_word in words:
                    words.remove(next_word)
                    queue.append((next_word, steps + 1))
    return 0

print(ladder_length("hit", "cog", ["hot","dot","dog","lot","log","cog"]))
```

**Complexity:** Time O(n * L * 26) where n=word count, L=word length, Space O(n) — BFS shortest path.

### 92. Longest Consecutive Sequence  (Hard)

Given an unsorted array, find the length of the longest run of consecutive integers, in O(n).

**Example:** `[100,4,200,1,3,2]` → `4` ([1,2,3,4])

```python
def longest_consecutive(nums):
    num_set = set(nums)
    best = 0
    for n in num_set:
        if n - 1 not in num_set:
            length = 1
            while n + length in num_set:
                length += 1
            best = max(best, length)
    return best

print(longest_consecutive([100,4,200,1,3,2]))
```

**Complexity:** Time O(n), Space O(n) — only expand from sequence starts, using a set for O(1) lookups.

### 93. Best Time to Buy and Sell Stock III  (Hard)

Find the max profit with at most two buy-sell transactions.

**Example:** `[3,3,5,0,0,3,1,4]` → `6`

```python
def max_profit_two(prices):
    buy1 = buy2 = float('-inf')
    sell1 = sell2 = 0
    for p in prices:
        buy1 = max(buy1, -p)
        sell1 = max(sell1, buy1 + p)
        buy2 = max(buy2, sell1 - p)
        sell2 = max(sell2, buy2 + p)
    return sell2

print(max_profit_two([3,3,5,0,0,3,1,4]))
```

**Complexity:** Time O(n), Space O(1) — state machine DP tracking 4 rolling states.

### 94. Palindrome Partitioning (All Partitions)  (Hard)

Given a string, partition it so every substring is a palindrome; return all such partitionings.

**Example:** `"aab"` → `[["a","a","b"],["aa","b"]]`

```python
def partition(s):
    result = []

    def is_pal(sub):
        return sub == sub[::-1]

    def backtrack(start, path):
        if start == len(s):
            result.append(path[:])
            return
        for end in range(start + 1, len(s) + 1):
            if is_pal(s[start:end]):
                path.append(s[start:end])
                backtrack(end, path)
                path.pop()

    backtrack(0, [])
    return result

print(partition("aab"))
```

**Complexity:** Time O(n * 2^n) worst case, Space O(n) recursion depth — backtracking with palindrome checks.

### 95. Longest Common Subsequence  (Hard)

Given two strings, find the length of their longest common subsequence.

**Example:** `"abcde", "ace"` → `3`

```python
def longest_common_subsequence(text1, text2):
    m, n = len(text1), len(text2)
    dp = [[0]*(n+1) for _ in range(m+1)]
    for i in range(1, m+1):
        for j in range(1, n+1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    return dp[m][n]

print(longest_common_subsequence("abcde", "ace"))
```

**Complexity:** Time O(m*n), Space O(m*n) — classic 2D DP table.

### 96. Maximum Product Subarray  (Hard)

Find the contiguous subarray within an array that has the largest product.

**Example:** `[2,3,-2,4]` → `6`; `[-2,0,-1]` → `0`

```python
def max_product(nums):
    max_prod = min_prod = result = nums[0]
    for n in nums[1:]:
        candidates = (n, max_prod * n, min_prod * n)
        max_prod, min_prod = max(candidates), min(candidates)
        result = max(result, max_prod)
    return result

print(max_product([2,3,-2,4]), max_product([-2,0,-1]))
```

**Complexity:** Time O(n), Space O(1) — track both running max and min (negatives can flip the sign).

### 97. Word Break  (Hard)

Given a string and a dictionary of words, determine if the string can be segmented into a space-separated sequence of dictionary words.

**Example:** `s="leetcode", wordDict=["leet","code"]` → `True`

```python
def word_break(s, word_dict):
    words = set(word_dict)
    n = len(s)
    dp = [False] * (n + 1)
    dp[0] = True
    for i in range(1, n + 1):
        for j in range(i):
            if dp[j] and s[j:i] in words:
                dp[i] = True
                break
    return dp[n]

print(word_break("leetcode", ["leet","code"]))
```

**Complexity:** Time O(n²), Space O(n) — DP where dp[i] means s[:i] is segmentable.

### 98. Subsets II (With Duplicates)  (Hard)

Given an array that may contain duplicates, return all unique subsets.

**Example:** `[1,2,2]` → `[[],[1],[1,2],[1,2,2],[2],[2,2]]`

```python
def subsets_with_dup(nums):
    nums.sort()
    result = []

    def backtrack(start, path):
        result.append(path[:])
        for i in range(start, len(nums)):
            if i > start and nums[i] == nums[i-1]:
                continue
            path.append(nums[i])
            backtrack(i + 1, path)
            path.pop()

    backtrack(0, [])
    return result

print(subsets_with_dup([1,2,2]))
```

**Complexity:** Time O(n * 2^n), Space O(n) recursion — sort first, then skip duplicate branches.

### 99. Combination Sum II (No Reuse, With Duplicates)  (Hard)

Given candidates (may contain duplicates) and a target, find all unique combinations where each number is used once.

**Example:** `candidates=[10,1,2,7,6,1,5], target=8` → includes `[1,1,6],[1,2,5],[1,7],[2,6]`

```python
def combination_sum2(candidates, target):
    candidates.sort()
    result = []

    def backtrack(start, path, remaining):
        if remaining == 0:
            result.append(path[:])
            return
        for i in range(start, len(candidates)):
            if i > start and candidates[i] == candidates[i-1]:
                continue
            if candidates[i] > remaining:
                break
            path.append(candidates[i])
            backtrack(i + 1, path, remaining - candidates[i])
            path.pop()

    backtrack(0, [], target)
    return result

print(combination_sum2([10,1,2,7,6,1,5], 8))
```

**Complexity:** Time O(2^n) worst case, Space O(n) recursion — backtracking with duplicate-skip logic.

### 100. Integer to English Words  (Hard)

Convert a non-negative integer into its English words representation.

**Example:** `123` → `"One Hundred Twenty Three"`

```python
def number_to_words(num):
    if num == 0:
        return "Zero"

    below20 = ["", "One","Two","Three","Four","Five","Six","Seven","Eight","Nine","Ten",
               "Eleven","Twelve","Thirteen","Fourteen","Fifteen","Sixteen","Seventeen","Eighteen","Nineteen"]
    tens = ["", "", "Twenty","Thirty","Forty","Fifty","Sixty","Seventy","Eighty","Ninety"]
    thousands = ["", "Thousand", "Million", "Billion"]

    def helper(n):
        if n == 0:
            return ""
        elif n < 20:
            return below20[n] + " "
        elif n < 100:
            return tens[n // 10] + " " + helper(n % 10)
        else:
            return below20[n // 100] + " Hundred " + helper(n % 100)

    result = ""
    i = 0
    while num > 0:
        if num % 1000 != 0:
            result = helper(num % 1000) + thousands[i] + " " + result
        num //= 1000
        i += 1
    return result.strip()

print(number_to_words(123))
print(number_to_words(1234567))
```

**Complexity:** Time O(log n) (digit groups processed), Space O(1) extra — recursive chunking by powers of 1000.
