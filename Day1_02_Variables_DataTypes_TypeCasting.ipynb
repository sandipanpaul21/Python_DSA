## 1. Introduction

- A **variable** is a name that points to a value stored in memory
- **Data type** tells Python what kind of value is stored (number, text, true/false, etc.)
- **Type casting** means converting one data type into another (e.g. text "5" into number 5)
- Why needed?
  - Every real dataset has mixed types (numbers, text, dates, booleans) - you must know how Python represents and converts them
  - Wrong type = wrong calculation = silently broken data pipeline (a classic real-world bug)
- Where used?
  - Reading raw data (CSV/API) - values often arrive as strings and need casting
  - Every single line of Data Science code touches variables and types

## 2. Real-Life Analogy

- A **variable** is like a labeled storage box (label = variable name, content = value)
- A **data type** is like the category printed on the box: "Fragile - Glass", "Liquid", "Documents" - it tells you how to handle the content
- **Type casting** is like repackaging: pouring liquid from a bottle into a different-shaped container - same content, different container/format
- Example: `age = "25"` is a box labeled `age` containing the text "25" (like a sticky note with numbers written on it, not an actual number) - you must "repackage" it with `int()` to actually do math with it

## 3. Explanation

- Python has these core built-in data types:
  - `int` - whole numbers (10, -5, 1000)
  - `float` - decimal numbers (3.14, -0.5)
  - `str` - text, written in quotes ("hello", 'data')
  - `bool` - True or False
  - `complex` - numbers with imaginary part (rarely used in DS, good to just know it exists)
- Python is **dynamically typed**: you don't write the type, Python detects it automatically based on the value
- Use `type(variable)` to check a variable's data type at any point

> **Trick to remember:** "IFSB" - Int, Float, String, Bool = the 4 types you will use 95% of the time.

## 4. Syntax

```python
# Variable assignment - no need to declare type
x = 10          # int
y = 3.14        # float
name = "Alice"  # str
is_active = True  # bool

# Checking type
type(x)   # <class 'int'>

# Type casting functions
int("10")      # string -> int
float("3.14")  # string -> float
str(10)        # int -> string
bool(0)        # any value -> bool (0 is False, non-zero is True)
```

- Variable naming rules:
  - Must start with a letter or underscore (`_`), not a digit
  - Can contain letters, digits, underscores
  - Case-sensitive (`Age` != `age`)
  - Cannot use Python reserved keywords (`for`, `if`, `class`, etc.)

````python
x = 10
y = 3.14
name = "Sandipan"
is_active = True

print(type(x), type(y), type(name), type(is_active))
````

## 5. Examples

### Basic Example

````python
# Basic: declaring variables of different types and checking type
age = 25
height = 5.9
city = "Pune"
is_student = False

print(age, type(age))
print(height, type(height))
print(city, type(city))
print(is_student, type(is_student))
````

### Intermediate Example

````python
# Intermediate: type casting between types
score_text = "85"          # comes in as string (e.g. from user input or a CSV)
score_number = int(score_text)   # cast string -> int
print(score_number + 10)         # now math works

pi_text = "3.14159"
pi_float = float(pi_text)
print(round(pi_float, 2))
````

- `int(score_text)` converts the string `"85"` into the actual number `85`
- Without casting, `score_text + 10` would raise a `TypeError` (cannot add string and int)
- `round(pi_float, 2)` rounds the float to 2 decimal places

### Real-World Example

````python
# Real-world: cleaning a "dirty" data row before analysis
# Common in EDA - raw CSV/API data often arrives entirely as strings

raw_row = {"employee_id": "101", "salary": "55000.50", "is_manager": "True"}

employee_id = int(raw_row["employee_id"])
salary = float(raw_row["salary"])
is_manager = raw_row["is_manager"] == "True"   # explicit comparison, safer than bool("False")

print(f"ID: {employee_id} ({type(employee_id).__name__})")
print(f"Salary: {salary} ({type(salary).__name__})")
print(f"Is Manager: {is_manager} ({type(is_manager).__name__})")

'''
type(employee_id) returns the class/type object itself, e.g. <class 'int'> — not just the word "int"
__name__ is an attribute every class/type object has, which stores its name as a plain string
So type(employee_id).__name__ extracts just "int" (clean string), instead of printing the whole <class 'int'> representation
'''
````

- Raw data from CSVs/APIs is almost always strings - this cleaning pattern is something you will write constantly in real EDA work
- Note: `bool("False")` would incorrectly return `True` because any non-empty string is truthy - hence the explicit `== "True"` comparison instead
- This exact "gotcha" is a favorite interview trap question

## 6. Internal Working

- Every value in Python (even `10` or `"a"`) is an **object** stored in memory
- A variable is just a **name/reference pointing to that object** (not a container holding the value directly)
- `int`, `float`, `bool` are **immutable** - once created, the object's value cannot change; reassigning a variable just points the name to a new object
- `bool` is technically a subclass of `int` in Python: `True == 1` and `False == 0` evaluate to `True`

> **Trick to remember:** `True` and `False` behave like `1` and `0` in math operations - `True + True` gives `2`. Interviewers love asking this to test if you understand Python's type system deeply.

````python
print(True + True)      # 2 (bool behaves like int)
print(True == 1)        # True
print(isinstance(True, int))  # True - bool is a subclass of int
````

## 7. Time and Space Complexity

- `type()` check and type casting (`int()`, `float()`, `str()`) are all **O(1)** operations for numbers
- Casting a string to a number is technically **O(n)** where n = number of digits/characters (the interpreter has to scan the whole string), but this is treated as constant time in practice for typical short values
- Space: primitive types (`int`, `float`, `bool`) take a small, mostly fixed amount of memory (though Python ints can grow arbitrarily large, unlike C)

## 8. Common Mistakes

- Forgetting that `input()` always returns a `str`, then trying to do math directly without casting -> `TypeError`
- Using `bool("False")` expecting `False` - any non-empty string is truthy, this returns `True`
- Trying `int("3.14")` directly -> fails; you must go `int(float("3.14"))`
- Confusing `str(10)` (converts number to text) with `int("10")` (converts text to number) - opposite directions
- Comparing types incorrectly, e.g. `"5" == 5` -> `False` (different types, no auto-casting in comparison)

````python
try:
    int("3.14")
except ValueError as e:
    print("Error:", e)

print(int(float("3.14")))   # correct way
````

## 9. Best Practices

- Always cast/validate incoming data (from files, APIs, user input) before using it in calculations
- Use `isinstance(x, int)` instead of `type(x) == int` for type checks (handles inheritance correctly)
- Use descriptive variable names (`salary`, not `s`)
- Prefer explicit casting (`int(x)`) over relying on Python's implicit type coercion - implicit behavior is a common source of bugs
- When casting booleans from strings, always compare explicitly (`x == "True"`), never use `bool(x)` blindly

## 10. Interview Questions

**Beginner**
- Q: What is the difference between `int()` and `str()`?
  A: `int()` converts a value into an integer number; `str()` converts a value into its text representation. They go in opposite directions.
- Q: What does `type()` return?
  A: The data type (class) of the given object, e.g. `<class 'int'>`.

**Intermediate**
- Q: Why does `bool("False")` return `True`?
  A: Because any non-empty string is considered "truthy" in Python - the string `"False"` has characters in it, so it's truthy, regardless of its text content.
- Q: Is Python statically typed or dynamically typed, and what does that mean for variables?
  A: Dynamically typed - a variable's type is determined at runtime based on the assigned value, and the same variable name can be reassigned to a different type later.

**Advanced**
- Q: Why does `True + True` equal `2` in Python?
  A: Because `bool` is implemented as a subclass of `int` in Python - `True` is `1` and `False` is `0` internally, so arithmetic operations treat them as integers.
- Q: What's the difference between `is` and `==` when comparing variables, and how does it relate to types/objects?
  A: `==` compares values for equality; `is` compares object identity (whether both variables point to the exact same object in memory). Two variables can hold equal values (`==` True) but be different objects (`is` False), except for small cached integers and interned strings where Python may reuse objects.
