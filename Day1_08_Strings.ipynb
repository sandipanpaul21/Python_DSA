## 1. Introduction

- A **string** is a sequence of characters, used to represent text
- In Python, strings are the built-in type `str`
- Why do we need strings?
  - Almost all real-world data has a text component - names, categories, file paths, log messages, JSON/CSV content
  - Data cleaning in Data Science relies heavily on string operations (trimming whitespace, fixing case, splitting fields)
- Where used?
  - Reading/cleaning column names and text data in Pandas
  - Parsing file paths, URLs, logs
  - Any interview question involving pattern matching, palindromes, anagrams, parsing

> **Interview angle:** String manipulation questions are extremely common because they test both language fluency (know your string methods) and algorithmic thinking (two pointers, hashing, sliding window) at once.

## 2. Real-Life Analogy

- A string is like a **necklace of beads**, where each bead is one character, strung together in a fixed order
- You can look at any bead by its position (indexing), take out a section of beads (slicing), but you **cannot swap out a single bead directly** - strings are immutable, so "editing" a string actually means making a **brand-new necklace** with the desired change
- Searching for a substring is like looking for a specific sequence of beads somewhere along the necklace

## 3. Explanation

- Strings in Python are **immutable** - once created, a string object's characters cannot be changed in place
- Any operation that appears to "modify" a string (like `.upper()` or concatenation) actually creates and returns a **new string object**
- Strings support **indexing** (`s[0]`) and **slicing** (`s[1:4]`), just like lists, because a string is a sequence type
- Strings can be defined with single quotes `'...'`, double quotes `"..."`, or triple quotes `'''...'''` / `"""..."""` for multi-line strings
- Python strings are Unicode by default - they can represent virtually any character/language out of the box

> **Trick to remember:** "Strings are immutable sequences of characters." Both words - immutable AND sequence - matter for interviews; immutability explains why repeated concatenation in a loop is inefficient, and sequence behavior explains why indexing/slicing work like lists.

## 4. Syntax

```python
s = "Hello, World"

# Indexing and slicing
s[0]        # 'H'
s[-1]       # 'd'
s[0:5]      # 'Hello'
s[::-1]     # reversed string

# Common methods
s.upper()          # "HELLO, WORLD"
s.lower()          # "hello, world"
s.strip()          # removes leading/trailing whitespace
s.replace("l", "L")# replaces occurrences
s.split(",")       # splits into a list by a delimiter
"-".join(["a","b"])# joins list elements into a string
s.find("World")    # returns index, or -1 if not found
s.startswith("He") # True/False
s.endswith("ld")   # True/False
len(s)             # length of string

# f-strings (formatting)
name = "Sandipan"
f"Hello, {name}!"
```

### Basic Example

````python
# Basic: indexing, slicing, and case conversion
text = "Data Science"
print(text[0])
print(text[5:])
print(text.upper())
print(text.lower())
````

- `text[0]` gets the first character
- `text[5:]` slices from index 5 to the end
- `.upper()`/`.lower()` return new strings - the original `text` is unchanged

### Intermediate Example

````python
# Intermediate: splitting, cleaning, and joining
raw = "  Sandipan, Data Scientist, Pune  "

cleaned = raw.strip()                 # remove leading/trailing whitespace
parts = cleaned.split(",")            # split into a list by comma
parts = [p.strip() for p in parts]    # strip whitespace from each part

print(parts)
print(" | ".join(parts))
````

- `.strip()` removes outer whitespace only, not whitespace between words
- `.split(",")` breaks the string into a list wherever a comma appears
- The list comprehension re-strips each part because splitting on `", "` still leaves stray spaces
- `" | ".join(parts)` glues the list back into one string with `" | "` between each item

### Real-World Example

````python
# Real-world: cleaning a messy column value before loading into a DataFrame
# Common first step in real data cleaning pipelines

raw_value = "  New York City  "

cleaned_value = raw_value.strip().title()
print(f"Cleaned city name: '{cleaned_value}'")

# Checking for a valid email-like pattern (simplified, not full validation)
email = "sandipan@example.com"
is_plausible_email = "@" in email and email.endswith(".com")
print("Looks like a valid email:", is_plausible_email)
````

- `.title()` capitalizes the first letter of every word - common when standardizing text categories (e.g. city names) before analysis
- Checking `"@" in email` and `.endswith(...)` is a lightweight sanity check (not real validation, which needs `re` - regex)
- This exact pattern (`strip()` + case normalization) is one of the first things done in real-world data cleaning

## 6. Internal Working

- Python strings are stored as **immutable sequences of Unicode code points**
- Since Python 3.3, CPython uses **flexible string representation** - internally it picks the smallest storage width (1, 2, or 4 bytes per character) needed based on the widest character actually present, saving memory for typical ASCII/Latin text
- Because strings are immutable, methods like `.upper()`, `.replace()`, or concatenation with `+` always create and return a **brand-new string object** - the original is untouched
- String **interning**: Python may reuse the same memory object for small, simple strings (like short identifiers) to save memory - this is why `a = "hi"; b = "hi"; a is b` can be `True` for short strings (implementation detail, not guaranteed for all strings)

> **Trick to remember:** Every string "modification" method actually builds a new object. This is exactly why building a large string with repeated `+=` in a loop is a classic performance mistake.

````python
# Demonstrating immutability: repeated concatenation creates new objects each time
s = "a"
print(id(s))
s += "b"
print(id(s))   # different id - a brand-new string object was created
````

## 7. Time and Space Complexity

- Indexing `s[i]`: **O(1)**
- Slicing `s[a:b]`: **O(k)** where k is the length of the slice (a new string of that length must be built)
- Concatenation `s1 + s2`: **O(n)** where n is the combined length (a new string is built)
- Repeated concatenation in a loop (`s += x`, n times): **O(n^2)** total, because each `+=` creates a new string and copies all previous characters again
- `in` (substring search), `.find()`, `.replace()`: **O(n)** in the average/worst case (Python's implementation uses an efficient substring search algorithm, but the worst case is still linear-ish relative to string length)
- `str.join(list_of_strings)`: **O(n)** where n is the total combined length - far more efficient than repeated `+=`

> **Interview line:** "Never build a large string with `+=` inside a loop - use a list to collect pieces and call `"".join(list)` once at the end, turning O(n^2) into O(n)."

````python
# Demonstrating the O(n) join pattern vs O(n^2) concatenation pattern
words = ["Data", "Science", "is", "fun"]

# Inefficient pattern (avoid in real code, shown for illustration)
result = ""
for w in words:
    result += w + " "

# Efficient pattern
result_efficient = " ".join(words)

print(result.strip())
print(result_efficient)
````

## 8. Common Mistakes

- Trying to modify a string in place, e.g. `s[0] = "H"` - raises `TypeError` because strings are immutable
- Using `+=` repeatedly inside a loop to build a large string - O(n^2) performance trap; use `"".join(list)` instead
- Forgetting that `.strip()` only removes whitespace from the **ends**, not the middle
- Confusing `.find()` (returns `-1` if not found) with `.index()` (raises `ValueError` if not found)
- Assuming `split()` with no arguments splits only on a single space - it actually splits on **any whitespace** and collapses multiple spaces automatically
- Off-by-one errors in slicing (`s[0:5]` gives 5 characters, indices 0 through 4 - the stop index is exclusive)

## 9. Best Practices

- Use f-strings (`f"{variable}"`) for building formatted strings - clearer and faster than `%` formatting or `.format()`
- Use `"".join(list_of_strings)` instead of repeated `+=` when building strings in a loop
- Always `.strip()` and normalize case (`.lower()`/`.upper()`/`.title()`) when comparing or cleaning user-provided or raw text data
- Use raw strings (`r"..."`) when writing regex patterns or Windows file paths, to avoid escape character confusion
- Prefer `.startswith()`/`.endswith()` over slicing when just checking a prefix/suffix - more readable and intention-revealing

## 10. Interview Questions

**Beginner**
- Q: Are strings mutable or immutable in Python?
  A: Immutable - once created, a string's characters cannot be changed in place; any "modification" creates a new string object.
- Q: How do you reverse a string in Python?
  A: Using slicing: `s[::-1]`.

**Intermediate**
- Q: Why is building a string with repeated `+=` in a loop inefficient?
  A: Because each `+=` creates a brand-new string object and copies all previous characters into it, resulting in O(n^2) total time for n concatenations; using `"".join(list)` avoids this by building the result in one O(n) pass.
- Q: What is the difference between `.find()` and `.index()`?
  A: Both search for a substring and return its starting index; `.find()` returns -1 if not found, while `.index()` raises a `ValueError` if not found.

**Advanced**
- Q: How does CPython store strings internally to save memory (flexible string representation)?
  A: Since Python 3.3, CPython chooses the smallest fixed-width storage (1, 2, or 4 bytes per character) needed to represent the widest character actually present in the string, rather than always using 4 bytes per character.
- Q: How would you check if two strings are anagrams of each other efficiently?
  A: Sort both strings and compare (O(n log n)), or use a character frequency count with a dictionary/Counter for both strings and compare the counts (O(n) time, O(1) space if the character set is fixed like lowercase English letters).

## 12. Revision Summary

### Medium (21-40)

21. **Q: Why can't you do `s[0] = "H"` to modify a string?**
    A: Because strings are immutable in Python - item assignment on an immutable type raises a `TypeError`.
22. **Q: What is the time complexity of string concatenation with `+`?**
    A: O(n), where n is the length of the resulting string, since a new string object must be created and both operands copied into it.
23. **Q: Why is using `+=` in a loop to build a large string inefficient?**
    A: Each `+=` creates an entirely new string object and copies all previously accumulated characters again, leading to O(n^2) total time across n iterations.
24. **Q: What is the more efficient alternative to repeated `+=` for building strings?**
    A: Collect pieces in a list and call `"".join(list)` once at the end - O(n) total time.
25. **Q: What is the difference between `.find()` and `.index()` when a substring isn't found?**
    A: `.find()` returns `-1`; `.index()` raises a `ValueError`.
26. **Q: How would you check if a string is a palindrome?**
    A: Compare the string to its reverse, e.g. `s == s[::-1]` (after normalizing case/spaces if needed).
27. **Q: What does `str.split(",")` do differently from `str.split()`?**
    A: `split(",")` splits only on the comma character exactly; `split()` with no argument splits on any run of whitespace and ignores leading/trailing whitespace automatically.
28. **Q: How would you count the frequency of each character in a string?**
    A: Use a dictionary or `collections.Counter` to tally occurrences while iterating through the string once - O(n) time.
29. **Q: What is string interning in Python?**
    A: An optimization where CPython may reuse the same memory object for identical short/simple strings (e.g. identifiers), so two variables holding the same short string literal can sometimes point to the same object in memory.
30. **Q: How do you check if two strings are anagrams using sorting?**
    A: Sort the characters of both strings and compare the results; if equal, they're anagrams. This is O(n log n) due to sorting.
31. **Q: What does `"  hello  ".strip()` return, and why doesn't it affect the space between "hello" and other words if there were any?**
    A: It returns `"hello"` - `.strip()` only removes whitespace from the very start and end of the string, not from the interior.
32. **Q: How would you remove all vowels from a string?**
    A: Using a list/generator comprehension with a join: `"".join(ch for ch in s if ch.lower() not in "aeiou")`.
33. **Q: What is the difference between `str.isalpha()` and `str.isalnum()`?**
    A: `.isalpha()` checks if all characters are alphabetic letters; `.isalnum()` checks if all characters are letters or digits (no symbols/spaces).
34. **Q: How would you count the number of words in a sentence?**
    A: `len(sentence.split())` - splitting on whitespace and counting the resulting list length.
35. **Q: What does slicing with a step, `s[::2]`, do?**
    A: Returns every second character of the string, starting from index 0.
36. **Q: How would you check if a string contains only uppercase letters?**
    A: `s.isupper()`.
37. **Q: What is the difference between `str.lower()` and `str.casefold()`?**
    A: Both lowercase a string, but `.casefold()` is more aggressive and handles certain special Unicode casing rules (like German sharp S) more correctly for case-insensitive comparisons.
38. **Q: How would you check if a sentence is a pangram (contains every letter of the alphabet at least once)?**
    A: Convert the sentence to lowercase, build a set of its letters, and check whether it's a superset of the set of all 26 alphabet letters.
39. **Q: What does `"Hello".encode()` return, and why would you need it?**
    A: It returns the string encoded as bytes (default UTF-8), needed when writing text to binary files, sending over a network, or working with byte-level APIs.
40. **Q: How would you remove all digits from a string?**
    A: `"".join(ch for ch in s if not ch.isdigit())`.

### Hard (41-60)

41. **Q: Explain how CPython's flexible string representation reduces memory usage compared to always using 4 bytes per character.**
    A: CPython inspects the widest character actually present in the string and stores every character using the smallest fixed width that fits (1 byte for pure Latin-1/ASCII text, 2 bytes if any character needs UCS-2 range, 4 bytes only if a character needs full UCS-4 range) - so a plain ASCII string uses just 1 byte per character instead of always paying for 4-byte Unicode storage.
42. **Q: How would you find the longest palindromic substring in a given string?**
    A: A common O(n^2) approach is "expand around center" - for each index (and each gap between indices, to handle even-length palindromes), expand outward while characters match, tracking the longest palindrome found; a more advanced O(n) solution uses Manacher's algorithm.
43. **Q: How would you implement string compression, e.g. "aaabcc" becomes "a3b1c2"?**
    A: Iterate through the string tracking the current character and a run-length counter; whenever the character changes, append the character and its count to the result, reset the counter, and continue - O(n) time, and if the compressed string isn't shorter than the original, you'd typically return the original.
44. **Q: What is the KMP (Knuth-Morris-Pratt) algorithm used for, and why is it better than naive substring search?**
    A: It's used for substring/pattern searching in O(n + m) time (text length n, pattern length m) by precomputing a "failure function" for the pattern that avoids re-checking characters already known to match, unlike the naive approach which can degrade to O(n*m) in the worst case with repetitive patterns.
45. **Q: How would you check if one string is a rotation of another, e.g. is "erbottlewat" a rotation of "waterbottle"?**
    A: Concatenate the second string with itself (`s2 + s2`) and check if the first string is a substring of that concatenation - if the lengths match and it's a rotation, this will always be true, done in O(n) using an efficient substring search.
46. **Q: How would you find all permutations of a given string?**
    A: Use backtracking - recursively fix one character at a time and permute the remainder, swapping characters in and out of position; this generates n! permutations, so the algorithm is inherently O(n * n!) time.
47. **Q: What is the time complexity of Python's `in` operator for substring search (e.g. `"abc" in text`), and what algorithm does CPython actually use?**
    A: Average case is close to O(n), though naive worst case is O(n*m); CPython uses a hybrid algorithm (a variant inspired by Boyer-Moore/Horspool with some elements of the "two-way" string matching algorithm) to make common cases fast while bounding worst-case behavior.
48. **Q: How would you determine if a string can be rearranged into a palindrome?**
    A: Count the frequency of each character; a string can be rearranged into a palindrome if at most one character has an odd frequency count (for even-length strings, zero odd counts are allowed; for odd-length, exactly one is allowed).
49. **Q: Explain the difference in behavior between `str.split()` with no arguments and `str.split(" ")` when there are multiple consecutive spaces.**
    A: `split()` with no arguments treats consecutive whitespace as a single delimiter and ignores leading/trailing whitespace entirely, producing no empty strings in the result; `split(" ")` splits on every single literal space character, producing empty strings `""` in the result wherever there were consecutive spaces.
50. **Q: How would you implement a basic run-length decoder for a string like "a3b1c2"?**
    A: Iterate through the string two characters (or a character plus a multi-digit number) at a time, parsing out a letter and its following digit(s), then repeat that letter that many times in the output - careful parsing is needed if counts can be multi-digit.
51. **Q: What is the Z-algorithm used for in string processing?**
    A: It computes, for every position in a string, the length of the longest substring starting there that matches a prefix of the string, in O(n) time - useful for pattern matching problems, often as an alternative to KMP.
52. **Q: How would you find the minimum window substring of string S that contains all characters of string T?**
    A: Use the sliding window technique with two pointers and a frequency count of T's characters - expand the right pointer to include characters until the window contains all of T's characters, then shrink from the left to find the minimum valid window, tracking the smallest valid window seen; this runs in O(n) time.
53. **Q: Why might using regular expressions (`re` module) be less performant than direct string methods for simple checks like `.startswith()`?**
    A: Regex involves compiling a pattern into a state machine and running a more general matching engine, which carries more overhead than a direct, purpose-built C-level string method for simple fixed checks - regex should be reserved for genuinely pattern-based (not fixed/literal) matching needs.
54. **Q: How would you check if string A can be transformed into string B using exactly one edit (insert, delete, or replace one character)?**
    A: If the lengths are equal, count mismatched character positions and check if the count is exactly one (replace case). If lengths differ by exactly one, walk both strings with two pointers, and once a mismatch is found, check if skipping one character in the longer string realigns the rest of the strings; more than one mismatch after that means more than one edit was needed. This is O(n).
55. **Q: What is the time and space complexity of checking if two strings are anagrams using a character frequency array, assuming lowercase English letters only?**
    A: O(n) time to build/compare frequency counts, and O(1) space, since the character set is fixed at 26 possible letters regardless of input size.
56. **Q: How does Python handle string comparison (`"apple" < "banana"`) internally?**
    A: It performs lexicographic comparison character by character based on Unicode code point values, similar to dictionary ordering, stopping at the first differing character (or by length if one is a prefix of the other).
57. **Q: What edge cases should you consider when checking if a string is a valid palindrome, ignoring non-alphanumeric characters and case?**
    A: Empty strings and single-character strings (trivially palindromes), strings with only non-alphanumeric characters (become empty after filtering, typically treated as palindromes), mixed case, and Unicode/accented characters if the definition of "alphanumeric" needs to be Unicode-aware rather than ASCII-only.
58. **Q: How would you efficiently find the first non-repeating character in a string?**
    A: Build a frequency count of all characters in one pass (O(n)), then iterate through the string again in order and return the first character whose count is exactly 1 (O(n)) - overall O(n) time, O(1) space if character set is fixed.
59. **Q: Why is it generally a bad idea to use string concatenation to build SQL queries or HTML from user input?**
    A: It creates injection vulnerabilities - malicious input can break out of the intended string structure and inject unintended commands/markup; parameterized queries or proper escaping/templating libraries should be used instead of raw string concatenation.
60. **Q: How would you compare Python's string immutability behavior to Java's, and why do both languages choose immutable strings?**
    A: Both Java and Python make strings immutable so that strings can be safely shared, hashed (used as dictionary/set keys), interned/cached for memory efficiency, and passed around without risk of unexpected mutation by another part of the program - the trade-off is that any modification requires creating a new object rather than mutating in place.
