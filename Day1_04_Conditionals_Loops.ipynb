## 1. Introduction

- **Conditional statements** let a program make decisions - run different code depending on whether a condition is `True` or `False`
- **Loops** let a program repeat a block of code multiple times, without rewriting it
- Why needed?
  - Real programs must react differently to different data (e.g. flag rows where `salary < 0` as invalid)
  - Real datasets have thousands/millions of rows - loops (or vectorized equivalents) process them without manual repetition
- Where used?
  - Data cleaning: `if value is missing: fill with mean`
  - Iterating through files, API responses, or rows of data
  - Every ML training loop is fundamentally a loop (epochs, batches)

## 2. Real-Life Analogy

- **`if/elif/else`** is like a decision at a traffic signal:
  - If light is red -> stop
  - Else if light is yellow -> slow down
  - Else (green) -> go
- **`for` loop** is like going through a checklist item by item, one at a time, until the list ends
- **`while` loop** is like eating popcorn "while the bowl is not empty" - you don't know exactly how many pieces, you just keep going until the condition becomes false
- **`break`** is like a fire alarm - stop everything immediately, exit the loop
- **`continue`** is like skipping a question you don't know in an exam and moving to the next one

## 3. Explanation

- `if` runs a block only when its condition is `True`
- `elif` (else-if) checks another condition only if the previous one(s) were `False`
- `else` runs when none of the above conditions were `True`
- `for` loop iterates over a sequence (list, string, range, etc.) - a fixed, known iteration
- `while` loop keeps running as long as a condition stays `True` - used when you don't know the exact number of iterations in advance
- `break` exits the loop completely
- `continue` skips the rest of the current iteration and moves to the next one
- `pass` does nothing - a placeholder when syntax requires a statement but you have no code yet

> **Trick to remember:** for = I know how many times. while = I don't know, I just watch a condition.

## 4. Syntax

```python
# Conditional
if condition1:
    # code
elif condition2:
    # code
else:
    # code

# For loop
for item in sequence:
    # code

for i in range(start, stop, step):
    # code

# While loop
while condition:
    # code

# Loop control
break      # exit loop entirely
continue   # skip to next iteration
pass       # do nothing (placeholder)
```

- `range(start, stop, step)` generates numbers from `start` to `stop-1`, incrementing by `step` (default start=0, step=1)

````python
for i in range(5):
    print(i, end=" ")
print()

for i in range(2, 10, 2):
    print(i, end=" ")
````

## 5. Examples

### Basic Example

````python
# Basic: if-elif-else
marks = 72

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Grade: {grade}")
````

### Intermediate Example

````python
# Intermediate: loop with break and continue
numbers = [4, 7, 2, 9, 15, 3, 20]

for num in numbers:
    if num > 15:
        print(f"Stopping - found {num} which exceeds 15")
        break
    if num % 2 != 0:
        continue   # skip odd numbers
    print(f"Even number: {num}")
````

- `continue` skips printing for odd numbers (7, 9, 3) and moves to the next item
- `break` stops the loop completely the moment it hits a number > 15 (here, 20 is never even reached)

### Real-World Example

````python
# Real-world: basic data cleaning loop
# Simulates checking a column of ages for invalid entries before analysis

raw_ages = [25, -5, 34, 150, 41, "unknown", 29]
clean_ages = []
invalid_count = 0

for entry in raw_ages:
    if not isinstance(entry, int):
        invalid_count += 1
        continue
    if entry < 0 or entry > 120:
        invalid_count += 1
        continue
    clean_ages.append(entry)

print(f"Clean ages: {clean_ages}")
print(f"Invalid entries removed: {invalid_count}")
````

- `isinstance(entry, int)` filters out non-numeric junk data (like the string `"unknown"`)
- Range check (`entry < 0 or entry > 120`) filters out impossible ages - a real EDA data-cleaning pattern
- `continue` cleanly skips bad entries without breaking the whole loop - this exact pattern appears constantly in real data pipelines before vectorized Pandas operations take over

## 6. Internal Working

- `for item in sequence` internally calls `iter(sequence)` to get an **iterator**, then repeatedly calls `next()` on it until it raises `StopIteration`
- This is why `for` loops work on lists, strings, dicts, files, and any "iterable" object - they all implement the iterator protocol
- `while` loop simply re-checks its condition expression before every iteration - no iterator involved
- `if/elif/else` evaluates conditions top to bottom and executes the **first** matching block only, then skips the rest

> **Trick to remember:** Every `for` loop in Python is secretly powered by the iterator protocol (`__iter__` / `__next__`) - this connects directly to the Generators topic coming up later in Day 1.

## 7. Time and Space Complexity

- A single `for`/`while` loop over `n` items: **O(n)** time
- A loop inside a loop (nested loop) over `n` items: **O(n^2)** time - a very common interview red flag to watch for
- Space complexity depends on what you store during the loop - if you build a new list of size `n`, that's **O(n)** space; if you just track a counter, it's **O(1)** space

> **Interview tip:** If you see nested loops filtering/searching data, the interviewer often wants you to recognize it's O(n^2) and suggest a faster approach (e.g. using a set/dict for O(1) lookups instead of a nested loop).

## 8. Common Mistakes

- Writing infinite `while` loops by forgetting to update the condition variable inside the loop
- Using `=` instead of `==` in a condition (syntax error, but a common typo)
- Modifying a list while iterating over it with a `for` loop - causes skipped elements or bugs (iterate over a copy instead: `for x in my_list[:]`)
- Forgetting `elif` and instead chaining multiple separate `if` statements - all of them get checked even after a match, wasting time and possibly causing bugs
- Off-by-one errors with `range()` - forgetting that `range(5)` goes from 0 to 4, not 1 to 5

````python
# Common mistake: modifying a list while looping over it directly
nums = [1, 2, 3, 4, 5]
# Wrong way (can skip elements): for n in nums: if n % 2 == 0: nums.remove(n)

# Correct way: iterate over a copy
for n in nums[:]:
    if n % 2 == 0:
        nums.remove(n)

print(nums)
````

## 9. Best Practices

- Prefer `for` loops over `while` when the number of iterations is known/finite
- Use `elif` instead of multiple separate `if` statements when conditions are mutually exclusive
- Avoid deeply nested loops (O(n^2) or worse) when a set/dict lookup can replace the inner loop
- Never modify a list while directly iterating over it - iterate over a copy or build a new list instead
- Use `enumerate()` when you need both index and value in a loop, instead of manually tracking a counter

````python
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, fruit)
````

## 10. Interview Questions

**Beginner**
- Q: What is the difference between `for` and `while` loops?
  A: `for` iterates over a known sequence for a fixed number of steps; `while` keeps looping as long as a condition remains `True`, useful when the number of iterations isn't known in advance.
- Q: What does `break` do versus `continue`?
  A: `break` exits the loop entirely; `continue` skips the rest of the current iteration and moves to the next one.

**Intermediate**
- Q: Why is modifying a list while iterating over it dangerous?
  A: Because the list's indices shift as elements are removed/added, causing the loop to skip elements or process them incorrectly. Iterate over a copy (`list[:]`) instead.
- Q: What does `range(2, 10, 2)` produce?
  A: 2, 4, 6, 8 - starts at 2, stops before 10, stepping by 2.

**Advanced**
- Q: How does a `for` loop work internally in Python?
  A: It calls `iter()` on the sequence to get an iterator object, then repeatedly calls `next()` on it, catching `StopIteration` to know when to stop - this is the iterator protocol.
- Q: Why should nested loops be flagged in a code review, and what's an alternative?
  A: Nested loops over the same data often mean O(n^2) time complexity. If the inner loop is doing membership checks or lookups, converting the target to a set or dict brings the lookup to O(1), reducing overall complexity to O(n).

## 12. Revision Summary

- `if / elif / else` - decision-making, only the first True condition's block runs
- `for` - iterate over a known sequence (uses iterator protocol internally)
- `while` - repeat while a condition holds; risk of infinite loops if condition never updates
- `break` exits loop entirely, `continue` skips to next iteration, `pass` does nothing (placeholder)
- Nested loops = O(n^2) - a common interview red flag; consider set/dict lookups instead
- Never mutate a list while iterating it directly - use a copy
- `enumerate()` gives index + value together, cleaner than manual counters
